### Q1: Names File Creation & Reading (`names.txt`)

**Sawal:** Ek program banayein jo:
1. Ek file `"names.txt"` ko write mode (`'w'`) me khole.
2. User ke diye hue 5 names (har line par ek naam) usme likhe.
3. Fir usi file ko read mode (`'r'`) me khol kar saare names screen par print kare.

**Logic (Hinglish):**
`with open("names.txt", "w")` se 5 names `\n` lagakar write karenge, fir `with open("names.txt", "r")` se read karke print karenge.

```python
# Q1 Solution
names_list = ["Soban", "Aman", "Priya", "Rahul", "Neha"]

# 1 & 2: Write mode me 5 names likhna
with open("names.txt", "w") as f:
    for name in names_list:
        f.write(name + "\n")
print("5 names successfully 'names.txt' me save ho gaye!")

# 3: Read mode me khol kar names print karna
print("\n--- 'names.txt' Content ---")
with open("names.txt", "r") as f:
    for line in f:
        print("Name:", line.strip())
```

### Q2: Logging System in Append Mode (`log.txt`)

**Sawal:** Ek program banayein jo:
1. Ek file `"log.txt"` ko append mode (`'a'`) me khole.
2. Nayi log entry dale (jaise `"Program run successfully"`).
3. Fir usi file ko read mode me khol kar saare logs print kare.

**Logic (Hinglish):** Append mode purana data erase nahi karta, isliye har run par naya log add hota jayega.

```python
from datetime import datetime

# 1 & 2: Append mode me naya log jodna
current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
new_log = f"[{current_time}] Program run successfully\n"

with open("log.txt", "a") as f:
    f.write(new_log)
print("Naya log append kar diya gaya!")

# 3: Read mode me saare logs print karna
print("\n--- 'log.txt' ke saare records ---")
with open("log.txt", "r") as f:
    print(f.read().strip())
```

### Q3: List Comprehension Filtering (`> 15`)

**Sawal:** Ek program banayein jo:
1. Numbers ki list rakhe: `[5, 10, 15, 20, 25]`.
2. List comprehension use karke nayi list banaye jisme sirf 15 se bade numbers hon.
3. Nayi list ko print kare.

**Logic (Hinglish):** Syntax: `[n for n in numbers if n > 15]`.

```python
# Q3 Solution
numbers = [5, 10, 15, 20, 25]

# List comprehension with condition
filtered_nums = [n for n in numbers if n > 15]

print("Original Numbers: ", numbers)
print("Numbers > 15 Only: ", filtered_nums)
```

### Q4: City Population Database with JSON (`cities.json`)

**Sawal:** 3 cities aur unki population ka Python dictionary banayein aur use `"cities.json"` me save karein:
1. JSON load karke har city aur uski population print karein.
2. Nayi city aur population lekar json file ko update karein.

```python
import json

# 1: Initial dictionary create karke JSON me save karna
cities_data = {
    "Bhopal": 2400000,
    "Mumbai": 20500000,
    "Delhi": 32000000
}

with open("cities.json", "w") as jf:
    json.dump(cities_data, jf, indent=4)
print("Initial cities data 'cities.json' me save ho gaya!")

# 2: JSON file read karke print karna
with open("cities.json", "r") as jf:
    loaded_cities = json.load(jf)

print("\n--- Current Cities Database ---")
for city, pop in loaded_cities.items():
    print(f"  {city}: {pop:,} log")

# 3: Nayi city add karke update karna
loaded_cities["Bengaluru"] = 13500000
with open("cities.json", "w") as jf:
    json.dump(loaded_cities, jf, indent=4)
print("\nUpdated 'cities.json' me 'Bengaluru' add ho gaya!")
```

### Q5: Missing File Exception Handling (`FileNotFoundError`)

**Sawal:** Ek program likho jo `"data.txt"` ko read mode me open karne ki koshish kare. Agar file exist na kare, to crash hone ke bajaye exception catch karke print kare:
`"File not found!"`

**Logic (Hinglish):** `try` ke andar `open("data.txt", "r")` karenge aur `except FileNotFoundError:` block se error catch karenge.

```python
# Q5 Solution
try:
    with open("data.txt", "r") as f:
        content = f.read()
        print("File content:", content)
except FileNotFoundError:
    print("File not found!")
else:
    print("File successfully read!")
finally:
    print("Safe file opening attempt completed.")
```
