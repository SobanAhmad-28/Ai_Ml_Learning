### Operation 1: File I/O Basics aur File Opening Modes

**Kyu zaroorat hai:**
Python me kisi bhi file ke sath kaam karne ke liye hume pehle use `open("filename", "mode")` se open karna hota hai.

**Common Modes:**
- `'r'` (Read - Default): File padhne ke liye. Agar file exist nahi karti to `FileNotFoundError` aata hai.
- `'w'` (Write): Nayi file banata hai ya existing file ka purana data mita kar naya likhta hai (overwrite/truncate).
- `'a'` (Append): Purane data ko bina mitaye aakhiri me naya text jodata hai.
- `'r+'`: Read + Write (File pehle se honi chahiye).
- `'w+'`: Write + Read (Purana data mita deta hai).
- `'b'` (Binary mode): Images, videos, ya serialized model files (`'rb'`, `'wb'`) ke liye.

```python
# File write mode ('w') me open karke naya content likhna
f = open("demo_sample.txt", "w")
f.write("Hello Python Learners!\nYeh hamari pehli file I/O line hai.\n")
f.close()  # Hamesha file close karna zaroori hai
print("demo_sample.txt file create ho gayi!")
```

### Operation 2: File Se Data Padhna (`read`, `readline`, `readlines`)

**Teeno Functions me Farq:**
1. `read()`: Puri file ka sara content ek single string ke roop me read karta hai.
2. `readline()`: Ek baar me sirf ek line padhta hai (har baar cursor agli line par chala jata hai).
3. `readlines()`: Saari lines ko ek List of strings ke roop me return karta hai (`['line 1\n', 'line 2\n']`).

```python
# 1. read() - Poora content ek string me
f = open("demo_sample.txt", "r")
full_text = f.read()
print("--- Full Text (read) ---\n", full_text)
f.close()

# 2. readlines() - Saari lines list me
f = open("demo_sample.txt", "r")
lines_list = f.readlines()
print("--- Lines List (readlines) ---\n", lines_list)
f.close()
```

### Operation 3: Data Append Karna (`'a'`) aur `writelines()`

**Kyu zaroorat hai:**
- Append mode (`'a'`): Jab hume server logs ya training history save karni ho bina purana record delete kiye.
- `writelines(list_of_strings)`: Multiple lines ko ek sath file me write karta hai.

```python
# Append mode ('a') me naya log add karna
f = open("demo_sample.txt", "a")
f.writelines(["Line 3: Training Epoch 1 Complete\n", "Line 4: Model Accuracy: 94.5%\n"])
f.close()

# Updated file check karna
f = open("demo_sample.txt", "r")
print("Updated Content:\n", f.read())
f.close()
```

### Operation 4: Best Practice — Context Manager (`with open(...) as f:`)

**Kyu zaroorat hai:**
Agar hum manual `f.close()` lagana bhool jayein, ya program beech me crash ho jaye, to operating system me file handle open reh jata hai jisse memory leak hoti hai.
- **`with open()`:** Yeh automatic context manager hai jo block khatam hote hi file ko **khud-b-khud safely close** kar deta hai, chahe error bhi kyu na aaye.

```python
# Recommended safe pattern
with open("demo_sample.txt", "r") as f:
    content = f.read()
    print("File read successfully with context manager!")

# Verify karna ki file close hui ya nahi
print("Kya file automatically close ho gayi?", f.closed)  # Returns True
```

### Operation 6: Exception Handling (`try`, `except`)

**Kyu zaroorat hai:**
Exception ek aisa runtime error hota hai jo aane par poora program turant crash ho jata hai.
- **Exception Handling:** Error ko 'catch' karta hai taaki program band na ho aur gracefully aage badhta rahe.
- **Common Errors:** `ZeroDivisionError` (0 se divide), `ValueError` (galat conversion), `FileNotFoundError` (missing file), `TypeError`.

```python
# Specific exception handling demo
try:
    ans = 10 / 0  # Zero division error
except ZeroDivisionError as e:
    print("[Handled Error] Zero se divide nahi kar sakte! Reason:", e)

try:
    val = int("soban_text")  # String to integer error
except ValueError:
    print("[Handled Error] Text string ko integer me convert nahi kiya ja sakta!")
```

### Operation 7: Error Handling me `else` aur `finally` Blocks

**Dono me Farq:**
- `else`: Sirf tab chalta hai jab `try` block me **koi error NA aaya ho**.
- `finally`: **Hamesha chalta hai**, chahe error aaye ya na aaye (resources cleanup aur connections close karne ke liye best hota hai).

```python
# else aur finally ka combined demo
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print(f"Galti: {a} ko 0 se divide nahi kiya ja sakta!")
    else:
        print(f"Success: {a} / {b} = {result}")
    finally:
        print("--- Division routine complete ---")

safe_divide(20, 4)  # Normal run (else chalega)
safe_divide(15, 0)  # Error run (except chalega)
```

### Operation 8: List Comprehensions (Pythonic One-Liners)

**Kyu zaroorat hai:**
Lambe 4-line ke `for` loop aur `.append()` ke badle 1 line me list banane ka elegant aur fast tareeqa:
1. **Basic:** `[expr for item in iterable]`
2. **Filtering (with if):** `[expr for item in iterable if condition]`
3. **Mapping (with if-else):** `[val_true if cond else val_false for item in iterable]`

```python
# 1. Basic: 1 se 5 ke squares
squares = [x ** 2 for x in range(1, 6)]
print("Squares:", squares)

# 2. Filter: Sirf even numbers (1 se 10 me)
evens = [x for x in range(1, 11) if x % 2 == 0]
print("Even numbers:", evens)

# 3. If-else: Even ko 'Even' aur Odd ko 'Odd' label karna
labels = ["Even" if x % 2 == 0 else "Odd" for x in range(1, 6)]
print("Labels for 1..5:", labels)
```

### Operation 9: JSON Module — In-Memory (`dumps` aur `loads`)

**Kyu zaroorat hai:**
JSON (JavaScript Object Notation) internet par APIs ke beech data bhejne ka standard format hai:
- `json.dumps(obj)`: Python dictionary $\rightarrow$ JSON text string me badalta hai (Serialization).
- `json.loads(str)`: JSON text string $\rightarrow$ Python dictionary me badalta hai (Deserialization).

```python
import json

# Python Dictionary
student_profile = {
    "name": "Soban",
    "age": 22,
    "skills": ["Python", "SQL", "Machine Learning"],
    "enrolled": True
}

# 1. dumps(): Dict -> JSON String
json_str = json.dumps(student_profile)
print("JSON String (dumps):", json_str)
print("json_str ka type:", type(json_str))

# 2. loads(): JSON String -> Python Dict
parsed_dict = json.loads(json_str)
print("Python Dict (loads):", parsed_dict["skills"])
```

### Operation 10: JSON File I/O (`dump` aur `load` with `indent=4`)

**Kyu zaroorat hai:**
- `json.dump(data, file, indent=4)`: Seedha file me JSON format me save karta hai (`indent=4` se code sundar aur readable dikhta hai).
- `json.load(file)`: Seedha disk par rakhi JSON file ko Python dictionary me load karta hai.

```python
# JSON file me save karna (dump)
config_data = {
    "model_name": "RandomForestClassifier",
    "n_estimators": 100,
    "max_depth": 10,
    "random_state": 42
}

json_filepath = "model_config.json"
with open(json_filepath, "w") as jf:
    json.dump(config_data, jf, indent=4)
print(f"Hyperparameters '{json_filepath}' me save ho gaye!")

# JSON file se wapas load karna (load)
with open(json_filepath, "r") as jf:
    loaded_config = json.load(jf)
print("File se loaded config:", loaded_config)
print("Loaded Model Name:", loaded_config["model_name"])
```
