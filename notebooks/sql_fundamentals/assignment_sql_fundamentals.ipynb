## ❓ Question 6: Unique Departments Present in Table (2 Solutions)

### 📌 Problem Statement:
Show a list of all unique departments present in the table.

### 💡 Hinglish Explanation & Comparative Solutions:
Aapne apne assignment me iske 2 tarike likhe the, dono valid hain:
1. **Approach 1 (`DISTINCT`):** Yeh standard aur recommended tarika hai duplicates hatane ke liye.
2. **Approach 2 (`GROUP BY`):** Categories ko group karke unique values nikalta hai.


```python
# Solution 1: DISTINCT (Recommended)
print("--- Solution 1: Using SELECT DISTINCT ---")
display(run_query("""
SELECT DISTINCT Department 
FROM Employee;
"""))

# Solution 2: GROUP BY (User's alternative solution)
print("\n--- Solution 2: Using GROUP BY ---")
display(run_query("""
SELECT Department AS All_Unique_Department
FROM Employee
GROUP BY Department;
"""))
```

## ❓ Question 9: Average Salary of All Employees

### 📌 Problem Statement:
Find the average salary of all employees.

### 💡 Hinglish Explanation:
- `AVG(Salary)` sabhi employees ki salary ka mathematical mean nikalta hai.
- `AS All_Emp_Avg_Salary` alias se output column ka naam clean ho jata hai.


```python
# Q9 Query
q9_sql = """
SELECT ROUND(AVG(Salary), 2) AS All_Emp_Avg_Salary
FROM Employee;
"""
display(run_query(q9_sql))
```

## ❓ Question 10: Departments with More Than 3 Employees (`HAVING` Clause)

### 📌 Problem Statement:
List each department along with the number of employees, but only include departments with more than 3 employees.

### 💡 Hinglish Explanation:
- `GROUP BY Department`: Employees ko department-wise group karega.
- `COUNT(*) AS Total_emp`: Har department ke staff count calculate karega.
- `HAVING COUNT(*) > 3`: `WHERE` aggregate functions par nahi lag sakta, isliye grouped results par filtering ke liye `HAVING` clause use hota hai.


```python
# Q10 Query
q10_sql = """
SELECT Department, COUNT(*) AS Total_emp
FROM Employee
GROUP BY Department
HAVING COUNT(*) > 3;
"""
display(run_query(q10_sql))

print("\nObservation: IT aur Finance dono departments me 3-3 employees hain. Agar condition > 3 hai toh koi department nahi aayega (empty set).")
print("Agar condition >= 3 hoti, toh IT aur Finance dono list hote:")
display(run_query("""
SELECT Department, COUNT(*) AS Total_emp
FROM Employee
GROUP BY Department
HAVING COUNT(*) >= 3;
"""))
```
