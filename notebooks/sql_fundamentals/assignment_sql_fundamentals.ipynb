## 🏢 Step 1: Database Setup & Table Creation (`Employee`)
- **Schema:** `EmployeeID`, `FirstName`, `LastName`, `Department`, `Salary`, `HireDate`.
- In-memory SQLite connection ke sath 10 realistic employee records initialize kiye gaye hain.


```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

def run_query(sql):
    return pd.read_sql_query(sql, conn)

cursor.execute("""
CREATE TABLE Employee (
    EmployeeID INT PRIMARY KEY,
    FirstName VARCHAR(50),
    LastName VARCHAR(50),
    Department VARCHAR(50),
    Salary DECIMAL(10,2),
    HireDate DATE
);
""")

employees = [
    (1, 'Aarav', 'Sharma', 'IT', 7500.00, '2021-03-15'),
    (2, 'Isha', 'Verma', 'HR', 4500.00, '2020-07-20'),
    (3, 'Rohan', 'Mehta', 'Finance', 6200.00, '2019-11-01'),
    (4, 'Sneha', 'Patel', 'IT', 8200.00, '2022-01-10'),
    (5, 'Vikram', 'Singh', 'Finance', 5100.00, '2021-08-25'),
    (6, 'Pooja', 'Nair', 'Marketing', 3900.00, '2023-02-14'),
    (7, 'Aditya', 'Rao', 'IT', 6800.00, '2020-04-18'),
    (8, 'Ananya', 'Das', 'Finance', 7100.00, '2018-09-30'),
    (9, 'Amit', 'Kumar', 'IT', 9000.00, '2017-06-12'),
    (10, 'Karan', 'Johar', 'Finance', 4800.00, '2022-10-05')
]

cursor.executemany("""
INSERT INTO Employee (EmployeeID, FirstName, LastName, Department, Salary, HireDate)
VALUES (?, ?, ?, ?, ?, ?);
""", employees)
conn.commit()

print("Database initialized with 10 Employee records.")
```

## ❓ Question 1: Display Every Employee & All Data
- **Task:** Write a query to display every employee and all available details (`SELECT *`).


```python
q1_sql = "SELECT * FROM Employee;"
display(run_query(q1_sql))
```

## ❓ Question 2: List Specific Columns (`FirstName`, `LastName`, `Salary`)
- **Task:** Retrieve only name and salary attributes.


```python
q2_sql = "SELECT FirstName, LastName, Salary FROM Employee;"
display(run_query(q2_sql))
```

## ❓ Question 3: Show Employees in 'IT' Department
- **Task:** Filter employees where `Department = 'IT'`.


```python
q3_sql = """
SELECT FirstName, LastName, Department 
FROM Employee
WHERE Department = 'IT';
"""
display(run_query(q3_sql))
```

## ❓ Question 4: Retrieve Employees with Salary > 6000
- **Task:** Filter employees earning greater than 6000.


```python
q4_sql = """
SELECT FirstName, LastName, Salary 
FROM Employee
WHERE Salary > 6000;
"""
display(run_query(q4_sql))
```

## ❓ Question 5: Order Employees by HireDate (Newest to Oldest)
- **Task:** Sort table chronologically by joining date (`ORDER BY HireDate DESC`).


```python
q5_sql = """
SELECT FirstName, LastName, HireDate 
FROM Employee
ORDER BY HireDate DESC;
"""
display(run_query(q5_sql))
```

## ❓ Question 6: Find Distinct Departments
- **Task:** List unique department names (`SELECT DISTINCT Department`).


```python
q6_sql = """
SELECT DISTINCT Department 
FROM Employee;
"""
display(run_query(q6_sql))
```

## ❓ Question 7: Filter Names Starting with 'A'
- **Task:** Substring wildcard search (`WHERE FirstName LIKE 'A%'`).


```python
q7_sql = """
SELECT FirstName, LastName, Department 
FROM Employee
WHERE FirstName LIKE 'A%';
"""
display(run_query(q7_sql))
```

## ❓ Question 8: Find Employees with Salary Between 4000 and 7000
- **Task:** Inclusive range condition (`WHERE Salary BETWEEN 4000 AND 7000`).


```python
q8_sql = """
SELECT FirstName, LastName, Salary 
FROM Employee
WHERE Salary BETWEEN 4000 AND 7000;
"""
display(run_query(q8_sql))
```

## ❓ Question 9: Calculate Average Salary of All Employees
- **Task:** Overall company-wide average salary calculation (`AVG(Salary)`).


```python
q9_sql = """
SELECT ROUND(AVG(Salary), 2) AS All_Emp_Avg_Salary
FROM Employee;
"""
display(run_query(q9_sql))
```

## ❓ Question 10: Find Departments with More Than 3 Employees
- **Task:** Aggregate filter using `GROUP BY Department HAVING COUNT(*) > 3`.


```python
q10_sql = """
SELECT Department, COUNT(*) AS Total_emp
FROM Employee
GROUP BY Department
HAVING COUNT(*) > 3;
"""
display(run_query(q10_sql))
```
