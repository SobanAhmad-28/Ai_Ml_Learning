### Operation 1: Introduction to Plotting
- **Core Elements:** `plt.figure()`, `plt.plot()`, `plt.title()`, `plt.xlabel()`, `plt.ylabel()`, `plt.show()`.


```python
import matplotlib.pyplot as plt
import numpy as np

# Oscar Winning Movies Revenue Dataset (in $ Millions)
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

# Basic Line Plot
plt.figure(figsize=(8, 4))
plt.plot(years, oscar_revenue)

plt.title("Oscar Winning Movies Revenue (2008-2012)")
plt.xlabel("Release Year")
plt.ylabel("Box Office Revenue ($ Millions)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

### Operation 2: Multi-Series Line Plots
- Ek hi canvas par multiple datasets compare karna using `label` aur `plt.legend()`.


```python
non_oscar_revenue = [378, 2788, 829, 185, 275]

plt.figure(figsize=(9, 4.5))

plt.plot(years, oscar_revenue, label="Oscar Winners", marker="o", linewidth=2)
plt.plot(years, non_oscar_revenue, label="Blockbuster Non-Winners (Avatar/Inception)", marker="s", linestyle="--", linewidth=2)

plt.title("Box Office Comparison: Oscar vs Non-Oscar Blockbusters")
plt.xlabel("Year")
plt.ylabel("Revenue ($ Millions)")
plt.legend(loc="upper right")
plt.grid(True, linestyle=":", alpha=0.7)
plt.show()
```

### Operation 3: Format Strings & Detailed Styling
- Shorthand syntax `'ro--'` (Red, circle marker, dashed line) aur hex colors (`#1f77b4`).


```python
plt.figure(figsize=(9, 4))

plt.plot(years, oscar_revenue, color="#1f77b4", linestyle="-", linewidth=2.5, 
         marker="o", markersize=8, markerfacecolor="red", markeredgecolor="black", label="Oscar Trend")

plt.title("Styled Box Office Trajectory")
plt.xlabel("Year")
plt.ylabel("Revenue ($ Millions)")
plt.legend()
plt.show()
```

### Operation 4: Built-in Themes & Styles (`plt.style`)
- Professional themes apply karna: `'ggplot'`, `'seaborn-v0_8'`, `'bmh'`.


```python
# Reset to default then use ggplot
plt.style.use('ggplot')

plt.figure(figsize=(8, 4))
plt.plot(years, oscar_revenue, 'o-b', label="Revenue Trend")
plt.title("Revenue Trend in ggplot Theme")
plt.xlabel("Year")
plt.ylabel("Revenue ($M)")
plt.legend()
plt.show()

# Reset style back to default
plt.style.use('default')
```

### Operation 5: Vertical Bar Charts & Value Labels
- Discrete categorical variables ke magnitudes compare karna.


```python
movies = ["The Dark Knight", "The Hurt Locker", "The King's Speech", "The Artist", "Argo"]
rev = [1005, 170, 427, 133, 232]

plt.figure(figsize=(9, 4.5))
bars = plt.bar(movies, rev, color="#6f42c1", width=0.55)

# Value Labels directly above bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 15, f"${height}M", ha="center", fontsize=9, fontweight="bold")

plt.title("Oscar Best Picture Box Office Revenues")
plt.ylabel("Revenue ($ Millions)")
plt.ylim(0, 1150)
plt.xticks(rotation=15)
plt.show()
```

### Operation 6: Grouped Side-by-Side Bar Charts
- `x_indices - width/2` aur `x_indices + width/2` se grouped bars align karna.


```python
x_indices = np.arange(len(years))
bar_width = 0.35

plt.figure(figsize=(9, 4.5))

plt.bar(x_indices - bar_width/2, oscar_revenue, width=bar_width, label="Oscar Winners", color="#2ca02c")
plt.bar(x_indices + bar_width/2, non_oscar_revenue, width=bar_width, label="Non-Winners", color="#d62728")

plt.title("Annual Revenue Comparison (Side-by-Side)")
plt.xlabel("Release Year")
plt.ylabel("Revenue ($ Millions)")
plt.xticks(x_indices, years)
plt.legend()
plt.show()
```

### Operation 7: Horizontal Bar Charts (`plt.barh`)
- Lambe labels wale categories ke liye horizontal orientation best rehti hai.


```python
plt.figure(figsize=(8, 4))
plt.barh(movies, rev, color="#17a2b8", height=0.5)

plt.title("Movie Revenues (Horizontal Bar Chart)")
plt.xlabel("Revenue ($ Millions)")
plt.ylabel("Movie Title")
plt.grid(axis="x", linestyle="--", alpha=0.6)
plt.show()
```

### Operation 8: Scatter Plots & Colormaps
- Do numerical variables ke beech correlation visualize karna using `plt.scatter()`.


```python
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.figure(figsize=(8, 4.5))
scatter = plt.scatter(age, bp, c=bp, cmap="OrRd", s=100, edgecolors="black", linewidth=1.2)
plt.colorbar(scatter, label="Blood Pressure (mm Hg)")

plt.title("Age vs Blood Pressure Correlation")
plt.xlabel("Patient Age (Years)")
plt.ylabel("Systolic Blood Pressure")
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

### Operation 9: Pie Charts & Explode Slicing
- Composition aur proportions visualize karna (`autopct='%1.1f%%'`).


```python
expenses = ["Salaries", "Office Rent", "Marketing", "R&D", "Misc"]
amounts = [500, 150, 120, 100, 50]
explode_slice = [0.06, 0, 0, 0.08, 0]
colors = ["#4e79a7", "#f28e2c", "#e15759", "#76b7b2", "#59a14f"]

plt.figure(figsize=(6, 6))
plt.pie(amounts, labels=expenses, autopct="%1.1f%%", startangle=140, 
        explode=explode_slice, colors=colors, shadow=True)

plt.title("Corporate Budget Distribution (FY 2024)")
plt.show()
```

### Operation 10: Subplots Grid Architecture
- Object-oriented API: `fig, (ax1, ax2) = plt.subplots(1, 2)`.


```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))

# Subplot 1
ax1.plot(years, oscar_revenue, 'o-g', linewidth=2)
ax1.set_title("Oscar Winners Trajectory")
ax1.set_xlabel("Year")
ax1.set_ylabel("Revenue ($M)")
ax1.grid(True, linestyle="--")

# Subplot 2
ax2.scatter(age, bp, color="purple", s=80)
ax2.set_title("Clinical Correlation")
ax2.set_xlabel("Age")
ax2.set_ylabel("Blood Pressure")
ax2.grid(True, linestyle=":")

plt.tight_layout()
plt.show()
```
