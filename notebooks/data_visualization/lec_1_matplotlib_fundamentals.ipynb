## 📈 Operation 1: Basic Line Plot & Anatomy of a Figure

### 💡 Kyu Zaroorat Hai?
Time-series data ya do continuous variables ke beech ka trend (jaise saal-dar-saal revenue ya temperature) dekhne ke liye line plot sabse simple aur effective graph hota hai.

### 🔍 Anatomy of a Matplotlib Plot:
- `plt.plot(x, y)`: Line draw karta hai.
- `plt.title("...")`: Plot ka main heading.
- `plt.xlabel("...")` & `plt.ylabel("...")`: X aur Y axes ke labels.
- `plt.show()`: Plot screen par display karta hai.


```python
import matplotlib.pyplot as plt
import numpy as np

# Oscar Winning Movies Revenue Dataset (in $ Millions)
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

# Basic Line Plot
plt.figure(figsize=(8, 4))
plt.plot(years, oscar_revenue)

# Labels and Title
plt.title("Oscar Winning Movies Revenue (2008 - 2012)", fontsize=13, fontweight='bold')
plt.xlabel("Year", fontsize=11)
plt.ylabel("Box Office Revenue ($ Millions)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```

## 🆚 Operation 2: Comparing Multiple Series & Adding Legends

### 💡 Kyu Zaroorat Hai?
Do alag-alag categories ka comparison karne ke liye (jaise Oscar winners vs Non-Oscar blockbusters), hum ek hi chart par multiple lines draw karte hain aur `plt.legend()` se unhe identify karte hain.


```python
# Non-Oscar Movies Data
non_oscar_revenue = [378, 2788, 829, 185, 275]  # Slumdog, Avatar, Inception, Hugo, Lincoln

plt.figure(figsize=(9, 5))

# Multiple lines with labels
plt.plot(years, oscar_revenue, label="Oscar Winners (The Dark Knight, Argo, etc.)", marker='o')
plt.plot(years, non_oscar_revenue, label="Non-Oscar Blockbusters (Avatar, Inception)", marker='s')

plt.title("Box Office Comparison: Oscar Winners vs Blockbusters", fontsize=13, fontweight='bold')
plt.xlabel("Year")
plt.ylabel("Revenue ($ Millions)")
plt.legend(loc="upper right", frameon=True)
plt.grid(True, alpha=0.5)

plt.show()
```

## 🎨 Operation 3: Styling Lines — Colors, Linestyles, Markers & Width

### 💡 Kyu Zaroorat Hai?
Professional reports me black-and-white print ya visual appeal ke liye custom colors, dotted/dashed lines aur prominent markers lagane aane chahiye.

### 🔍 Key Parameters:
- **Format String Syntax:** `"[marker][line][color]"` (jaise `'o-b'` = Circle marker, Solid line, Blue color; `'>--g'` = Triangle marker, Dashed line, Green color).
- `linewidth`: Line ki motai.
- `markersize`: Data points ke points ka size.


```python
plt.figure(figsize=(9, 4.5))

# Detailed styling parameters
plt.plot(years, oscar_revenue, color="#1f77b4", linestyle="-", linewidth=2.5, 
         marker="o", markersize=8, label="Oscar Winners")

plt.plot(years, non_oscar_revenue, color="#d62728", linestyle="--", linewidth=2.5, 
         marker="^", markersize=8, label="Non-Oscar Winners")

plt.title("Custom Styled Box Office Trends", fontsize=13)
plt.xlabel("Release Year")
plt.ylabel("Revenue in Millions ($)")
plt.legend()
plt.tight_layout()

plt.show()
```

## 🎭 Operation 4: Matplotlib Pre-built Themes & XKCD Comic Style

### 💡 Kyu Zaroorat Hai?
Bina manual coloring ke ek line me aesthetic themes apply karne ke liye Matplotlib ke paas pre-built styles hote hain:
- `plt.style.use('ggplot')`: R-language ka popular aesthetic grey grid style.
- `plt.style.use('dark_background')`: Sleek modern dark mode.
- `with plt.xkcd():`: Hand-drawn comic strip sketch style (Fun & Engaging presentations ke liye!).


```python
# Available Styles dekhna
print("Total Available Styles:", len(plt.style.available))
print("Sample styles:", plt.style.available[:6])

# Temporarily ggplot style apply karna
plt.style.use('ggplot')

plt.figure(figsize=(8, 4))
plt.plot(years, oscar_revenue, 'o-b', label="Oscar")
plt.plot(years, non_oscar_revenue, '^--r', label="Non-Oscar")
plt.title("Plotted in 'ggplot' Theme")
plt.legend()
plt.show()

# Reset back to default style
plt.style.use('default')
```

## 📊 Operation 5: Vertical Bar Charts & Bar Label Annotations

### 💡 Kyu Zaroorat Hai?
Categorical data (jaise individual movies ka revenue) compare karne ke liye bar chart best hota hai. Har bar ke upar exact number likhna readability 10x behtar karta hai.


```python
movies = ["The Dark Knight", "The Hurt Locker", "The King's Speech", "The Artist", "Argo"]
rev = [1005, 170, 427, 133, 232]

plt.figure(figsize=(10, 5))
bars = plt.bar(movies, rev, color="#6f42c1", width=0.55)

plt.title("Yearly Revenue of Oscar Movies", fontsize=13, fontweight='bold')
plt.xlabel("Movie Title")
plt.ylabel("Revenue ($ Millions)")
plt.ylim(0, max(rev) + 150)  # Top par text ke liye extra headroom

# Har bar ke upar text label lagana
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 20, f"${height}M", 
             ha="center", fontsize=10, fontweight="bold")

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## 🏛️ Operation 6: Grouped Bar Charts (Side-by-Side Comparison)

### 💡 Kyu Zaroorat Hai?
Do alag-alag metrics ko har saal ke liye side-by-side compare karne ke liye hum X-axis positions ko offset karte hain (`x - width/2` aur `x + width/2`).


```python
x_indices = np.arange(len(years))
bar_width = 0.35

plt.figure(figsize=(9, 5))

# Side-by-Side Bars
plt.bar(x_indices - bar_width/2, oscar_revenue, width=bar_width, label="Oscar Winners", color="#2ca02c")
plt.bar(x_indices + bar_width/2, non_oscar_revenue, width=bar_width, label="Non-Oscar Blockbusters", color="#ff7f0e")

plt.title("Side-by-Side Yearly Revenue Comparison", fontsize=13, fontweight='bold')
plt.xlabel("Year")
plt.ylabel("Revenue ($ Millions)")
plt.xticks(x_indices, years)
plt.legend()
plt.tight_layout()

plt.show()
```

## 📏 Operation 7: Horizontal Bar Charts (`plt.barh`)

### 💡 Kyu Zaroorat Hai?
Jab categories ke naam lambe hote hain (jaise Movie titles ya Job designations), vertical bars overlap ho jati hain. Horizontal bars me text aaram se horizontally read hota hai.


```python
plt.figure(figsize=(9, 4.5))

plt.barh(movies, rev, color="#17a2b8", height=0.5)

plt.title("Movie Revenues (Horizontal Bar Chart)", fontsize=13)
plt.xlabel("Revenue ($ Millions)")
plt.ylabel("Movie Title")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```

## 🔴 Operation 8: Scatter Plots, Point Sizing & Colormaps (`cmap`)

### 💡 Kyu Zaroorat Hai?
Do continuous numerical variables ke beech correlation dekhne ke liye (jaise Age vs Blood Pressure) scatter plot use karte hain. Point ka color aur size 3rd aur 4th dimension ki information dikha sakta hai!


```python
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.figure(figsize=(9, 5))

# Scatter plot: Color mapped according to BP intensity (cmap='OrRd')
scatter = plt.scatter(age, bp, c=bp, cmap="OrRd", s=[x * 1.5 for x in bp], 
                      edgecolor='black', alpha=0.8, marker='o')

plt.title("Patient Age vs Blood Pressure Analysis", fontsize=13, fontweight='bold')
plt.xlabel("Age (Years)")
plt.ylabel("Blood Pressure (mmHg)")
plt.colorbar(scatter, label="Blood Pressure Intensity")
plt.grid(True, linestyle="--", alpha=0.5)

# High BP threshold line
plt.axhline(135, color='red', linestyle=':', label="High BP Threshold (135)")
plt.legend()

plt.show()
```

## 🥧 Operation 9: Pie Charts (Percentages, Slice Explode & Shadows)

### 💡 Kyu Zaroorat Hai?
Overall total me se alag-alag parts ka percentage contribution (jaise Company Expense distribution) show karne ke liye pie chart use karte hain.
- `autopct='%1.1f%%'`: Percentage display karta hai.
- `explode`: Kisi specific important slice ko bahar highlight karna.


```python
expenses = ["Salaries", "Office Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
explode_slice = [0.05, 0, 0, 0.1, 0]  # Salaries aur R&D ko bahar pull karna
colors = ["#4e79a7", "#f28e2b", "#e15759", "#76b7b2", "#59a14f"]

plt.figure(figsize=(7, 7))
plt.pie(amounts, labels=expenses, autopct='%1.1f%%', explode=explode_slice, 
        colors=colors, shadow=True, startangle=140, 
        wedgeprops={"edgecolor": "white", "linewidth": 1.5})

plt.title("Company Operational Budget Distribution", fontsize=13, fontweight='bold')
plt.tight_layout()

plt.show()
```

## 🖼️ Operation 10: Multi-Plot Subplots (`plt.subplots`) & Exporting Images

### 💡 Kyu Zaroorat Hai?
Dashboard style me ek sath 2 ya 4 graphs grid me arrange karne ke liye `plt.subplots(rows, cols)` standard approach hai. `plt.savefig("plot.png")` se plot high-res image me save ho jata hai.


```python
# 1 Row, 2 Columns Subplots Grid
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))

# Subplot 1: Line Chart
ax1.plot(years, oscar_revenue, 'o-g', linewidth=2)
ax1.set_title("Oscar Trend")
ax1.set_xlabel("Year")
ax1.set_ylabel("Revenue ($M)")
ax1.grid(True, alpha=0.5)

# Subplot 2: Bar Chart
ax2.bar(movies, rev, color="#ff9999", width=0.5)
ax2.set_title("Movie Breakdown")
ax2.set_xticklabels(movies, rotation=25, ha="right", fontsize=9)
ax2.grid(axis='y', alpha=0.5)

plt.tight_layout()

# Save Figure to Disk
export_img_path = "oscar_dashboard.png"
plt.savefig(export_img_path, dpi=200)
print(f"✅ Dashboard plot exported to '{export_img_path}' successfully!")

plt.show()
```
