### Q2: 2 Numbers ke Arithmetic Operations

**Sawal:** User se 2 numbers lekar unka Sum, Difference, Product aur Quotient print karo.

**Logic (Hinglish):** Float casting use karenge taaki decimal numbers bhi handle ho sakein.

```python
num1 = 20.0
num2 = 5.0

print(f"Numbers: {num1}, {num2}")
print(f"Sum (Jod):         {num1 + num2}")
print(f"Difference (Ghatav): {num1 - num2}")
print(f"Product (Guna):      {num1 * num2}")
print(f"Quotient (Bhag):     {num1 / num2}")
```

### Q3: 2 Integers aur 1 Float ka Average

**Sawal:** User se 2 integers aur 1 float number lo. Sabhi ko float me badlo aur unka average print karo.

**Logic (Hinglish):** `float()` function se explicit cast karke formula `(f1 + f2 + f3) / 3` chalayenge.

```python
n1 = 10
n2 = 25
n3 = 14.5

f1, f2, f3 = float(n1), float(n2), float(n3)
avg = (f1 + f2 + f3) / 3
print(f"Floats: {f1}, {f2}, {f3}")
print(f"Average: {avg:.4f}")
```

### Q4: String se Integer, Float aur String wapas banana

**Sawal:** User ek number wali string enter karta hai (e.g. `"45"`). Use badlo:
1. Integer me
2. Float me
3. Wapas String me
Aur teeno ke type print karo.

```python
raw = "45"
as_int = int(raw)
as_float = float(raw)
as_str = str(as_int)

print(f"Integer value: {as_int} (Type: {type(as_int).__name__})")
print(f"Float value:   {as_float} (Type: {type(as_float).__name__})")
print(f"String value:  '{as_str}' (Type: {type(as_str).__name__})")
```

### Q5: Expression `10 + 3 * 2 ** 2` ka Order aur Result

### Q6: 2 Variables ki Values Swap (Badalna)

### Q7: Celsius se Fahrenheit Temperature Converter

**Sawal:** Celsius string input ko float me badal kar Fahrenheit calculate karo.

**Formula:** $\text{Fahrenheit} = (\text{Celsius} \times \frac{9}{5}) + 32$

```python
c_str = "37.5"
celsius = float(c_str)
fahrenheit = (celsius * (9 / 5)) + 32
print(f"Celsius:    {celsius:.2f} °C")
print(f"Fahrenheit: {fahrenheit:.2f} °F")
```

### Q8: Circle ka Area (Radius $r$ se)

**Sawal:** Radius $r$ lekar Circle ka area print karo (jahan $\pi = 3.14$).

**Formula:** $\text{Area} = \pi \times r^2$

```python
r = 7.0
pi = 3.14
area = pi * (r ** 2)
print(f"Radius: {r} -> Circle Area: {area:.2f}")
```

### Q9: Simple Interest ($SI$) Calculation

**Sawal:** Principal ($P$), Rate ($R$), aur Time ($T$) lekar Simple Interest calculate karo.

**Formula:** $SI = \frac{P \times R \times T}{100}$

```python
p = 50000.0
r = 7.5
t = 3.0
si = (p * r * t) / 100
print(f"Principal: ₹{p:,.2f} | Rate: {r}% | Time: {t} saal")
print(f"Simple Interest (Byaj): ₹{si:,.2f}")
print(f"Total Wapas Rashi:       ₹{p + si:,.2f}")
```

### Q10: Decimal Number ka Integer aur Fractional Part Alag Karna

**Sawal:** Decimal number (jaise `45.78`) lo aur uska:
- Integer part (`45`)
- Fractional part (`.78`)
alag-alag print karo.

```python
val = 45.78
int_part = int(val)
frac_part = round(val - int_part, 4)
print("Original Decimal:", val)
print("Integer Part:   ", int_part)
print("Fractional Part:", frac_part)
```
