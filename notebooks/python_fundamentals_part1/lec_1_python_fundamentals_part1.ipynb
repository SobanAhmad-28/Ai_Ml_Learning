### Operation 2: Variables, Identifiers aur Naming Rules

**Kyu zaroorat hai:**
Variable ek labeled dabbe (container) jaisa hota hai jo memory me data ya value ko store karta hai taaki hum use baad me use ya change kar sakein.

**Variables ke Naming Rules:**
1. Naam hamesha letter (`a-z`, `A-Z`) ya underscore (`_`) se shuru hona chahiye.
2. Pehle letter ke baad numbers aa sakte hain, lekin naam number se shuru nahi ho sakta (e.g. `2name` galat hai, `name2` sahi hai).
3. Reserved keywords (jaise `if`, `for`, `class`, `def`) ko variable name nahi bana sakte.
4. Python **case-sensitive** hai: yaani `age` aur `Age` do alag-alag variables hain.
5. Python **dynamically typed** hai: yaani variable banate waqt `int` ya `str` alag se likhne ki zaroorat nahi hoti.

```python
# Variables define karna
name = "Shradha"
age = 30
PI = 3.14
word = "artificial intelligence"

print("Name:", name)
print("Age:", age)
print("Value of PI:", PI)
print("Topic:", word)

# Case-sensitivity ka demo
Age = 35
print(f"age hai {age}, jabki Age hai {Age} (dono alag variables hain)")
```

### Operation 3: Built-in Data Types aur `type()` Function

**Kyu zaroorat hai:**
Data type batata hai ki variable kis tarah ki value hold kar raha hai. Kisi bhi variable ka type check karne ke liye hum `type()` function use karte hain.

**Main Data Types:**
1. `int`: Bina decimal wale poore numbers (e.g. `10`, `-5`).
2. `float`: Decimal wale point numbers (e.g. `3.14`, `-0.5`).
3. `str`: Quotes ke andar likha text (e.g. `"shradha"`).
4. `bool`: Boolean truth value yaani sirf `True` ya `False`.
5. `None`: Jab kisi variable me koi value nahi hoti (empty/null state).

```python
x = 10
print("x =", x, "| Type:", type(x))

pi_val = 3.14
print("pi_val =", pi_val, "| Type:", type(pi_val))

student_name = "shradha"
print("student_name =", student_name, "| Type:", type(student_name))

is_adult = True
print("is_adult =", is_adult, "| Type:", type(is_adult))

empty_var = None
print("empty_var =", empty_var, "| Type:", type(empty_var))
```

### Operation 4: Type Conversion (Automatic) vs Type Casting (Manual)

**Kyu zaroorat hai:**
Ek data type ko dusre type me badalna type conversion ya casting kahlata hai. Yeh 2 tareeqe se hota hai:
1. **Implicit Type Conversion (Automatic):** Python khud decimal values ko protect karne ke liye integer ko float me convert kar deta hai (jaise `5 + 3.0 = 8.0`).
2. **Explicit Type Casting (Manual):** Programmer khud function use karke type badalta hai, jaise `int()`, `float()`, `str()`, `bool()`.

```python
# 1. Automatic (Implicit) conversion: int (5) + float (3.0) -> float (8.0)
a = 5
b = 3.0
ans = a + b
print(f"Automatic conversion: {a} + {b} = {ans} (Type: {type(ans).__name__})")

# 2. Manual (Explicit) type casting
num_int = 10
num_float = float(num_int)  # 10 ban gaya 10.0
num_str = str(num_int)      # 10 ban gaya "10"

print(f"Explicit float: {num_float} ({type(num_float).__name__})")
print(f"Explicit string: '{num_str}' ({type(num_str).__name__})")
```

### Operation 5: Arithmetic Operators (Ganit ke Operators)

**Kyu zaroorat hai:**
Numbers ke beech basic math operations karne ke liye use hote hain:

| Operator | Kaam | Meaning | Example (`a=5, b=10`) |
| :--- | :--- | :--- | :--- |
| `+` | Addition | Jodna | `5 + 10 = 15` |
| `-` | Subtraction | Ghatana | `5 - 10 = -5` |
| `*` | Multiplication | Guna karna | `5 * 10 = 50` |
| `/` | Division | Bhag dena (hamesha float answer aayega) | `5 / 10 = 0.5` |
| `%` | Modulo | Bacha hua sheshphal (remainder) | `5 % 10 = 5` |
| `**` | Exponentiation | Power (Ghaat) nikalna ($5^{10}$) | `5 ** 10 = 9765625` |

```python
a = 5
b = 10

print("a + b =", a + b)   # Jod (Addition)
print("a - b =", a - b)   # Ghatana (Subtraction)
print("a * b =", a * b)   # Guna (Multiplication)
print("a / b =", a / b)   # Bhag (Division -> float me)
print("a % b =", a % b)   # Sheshphal (Remainder)
print("a ** b =", a ** b) # Power (5 ki power 10)
```

### Operation 6: Relational / Comparison Operators (Tulna Karne Wale)

**Kyu zaroorat hai:**
Do values ki tulna karne ke liye use hote hain aur result hamesha `True` ya `False` (boolean) aata hai:

| Operator | Meaning | Example (`a=10, b=20`) | Output |
| :--- | :--- | :--- | :--- |
| `==` | Barabar hai? | `a == b` | `False` |
| `!=` | Barabar nahi hai? | `a != b` | `True` |
| `>` | Bada hai? | `a > b` | `False` |
| `<` | Chhota hai? | `a < b` | `True` |
| `>=` | Bada ya barabar hai? | `a >= b` | `False` |
| `<=` | Chhota ya barabar hai? | `a <= b` | `True` |

```python
a = 10
b = 20

print("a == b:", a == b) # Barabar hai?
print("a != b:", a != b) # Barabar nahi hai?
print("a > b:",  a > b)  # Bada hai?
print("a < b:",  a < b)  # Chhota hai?
print("a >= b:", a >= b) # Bada ya barabar?
print("a <= b:", a <= b) # Chhota ya barabar?
```

### Operation 7: Assignment Operators (Value Update Karne Wale)

**Kyu zaroorat hai:**
Variables ki value ko short tareeqe se calculate karke update karne ke liye use hote hain (`x += 5` ka matlab hai `x = x + 5`).

```python
x = 10
print("Starting x =", x)

x += 5   # x me 5 jod diya
print("x += 5 ->", x)

x -= 3   # x me se 3 ghata diya
print("x -= 3 ->", x)

x *= 2   # x ko 2 se multiply kiya
print("x *= 2 ->", x)

x /= 4   # x ko 4 se divide kiya
print("x /= 4 ->", x)

x %= 4   # x ka remainder liya
print("x %= 4 ->", x)

x **= 3  # x ki power 3
print("x **= 3 ->", x)
```

### Operation 8: Logical Operators (`and`, `or`, `not`)

**Kyu zaroorat hai:**
Multiple shartein (conditions) check karne ke liye use hote hain:
- `and`: `True` tabhi aayega jab **dono** shartein sahi hon.
- `or`: `True` tab aayega jab **koi ek bhi** shart sahi ho.
- `not`: Result ko ulta kar deta hai (`True` ko `False`, aur `False` ko `True`).

```python
x = 10
y = 20
z = 5

# Logical AND: dono baatein sahi honi chahiye
print("(x > z) and (y > x):", (x > z) and (y > x))  # (10 > 5) aur (20 > 10) dono True -> True

# Logical OR: koi ek bhi baat sahi ho to chalega
print("(x > y) or (y > z):", (x > y) or (y > z))    # (False) or (True) -> True

# Logical NOT: ulti baat
print("not (x > y):", not (x > y))                  # not(False) -> True
```

### Operation 9: Operator Precedence (Calculation ka Order) aur Unary Operators

**Kyu zaroorat hai:**
Jab ek hi expression me multiple operators hon to pehle kaunsa solve hoga, yeh priority list decide karti hai:

1. `()` - Brackets / Parentheses (Sabse pehle)
2. `**` - Power / Exponent
3. `+x`, `-x` - Unary minus/plus (Single number ka sign)
4. `*`, `/`, `//`, `%` - Guna aur Bhag
5. `+`, `-` - Jod aur Ghatana
6. Comparison operators (`<`, `<=`, `>`, `>=`, `!=`, `==`)
7. `not`
8. `and`
9. `or`
10. Assignment (`=`, `+=`, etc.)

```python
# 1. Unary operator (single value ka sign badalna)
val = 5
neg_val = -val  # Unary minus
print("val =", val, "| -val =", neg_val)

# 2. Operator Precedence order:
# Expression: 10 + 3 * 2 ** 2
# Step 1: Pehle 2 ** 2 hoga = 4
# Step 2: Fir 3 * 4 hoga = 12
# Step 3: Aakhiri me 10 + 12 hoga = 22
res = 10 + 3 * 2 ** 2
print("Result of 10 + 3 * 2 ** 2 =", res)
```
