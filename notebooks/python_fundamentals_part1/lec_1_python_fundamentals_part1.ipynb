### Operation 1: Console Output & The `print()` Function

**Why it is needed:**
The `print()` function is the primary mechanism to display data, diagnostic messages, and computation results to the console.

**Key Rules & Syntax:**
- Text enclosed in single quotes (`''`) or double quotes (`""`) represents a literal string and prints as-is.
- Numerical literals and expressions are evaluated before printing.
- Passing multiple comma-separated arguments causes Python to print each value separated by a space on the same line.

```python
# 1. Printing a string literal
print("hello world")

# 2. Printing numerical values and calculated expressions
print(3.14)
print(2 + 3)

# 3. Printing multiple values on the same line
print("hello world", "from PRIME")
```

### Operation 2: Variables, Identifiers & Naming Rules

**Why it is needed:**
A variable is a labeled container that points to a value stored in memory. In Python, variables are dynamically typed (no type declaration keyword like `int` or `float` is needed).

**Identifier Rules:**
1. Must start with a letter (`a-z`, `A-Z`) or an underscore (`_`).
2. Can contain letters, numbers, or underscores (`age_1`, `_rate`). Cannot start with a digit (e.g., `2name` is invalid).
3. Cannot use reserved keywords (e.g., `if`, `for`, `class`, `while`, `def`, `in`).
4. Variable names are case-sensitive (`age` and `Age` are distinct variables).
5. Standard indentation in Python is 4 spaces.

```python
# Valid variable definitions
name = "Soban"
age = 20
PI = 3.14
word = "artificial intelligence"

print("name:", name)
print("age:", age)
print("value of PI =", PI)
print("topic:", word)

# Demonstrating case-sensitivity
Age = 25
print(f"age is {age}, while Age is {Age} (distinct identifiers)")
print(f"Soban age is {age}, 5 year later soban age will be {Age}")
```

### Operation 3: Built-in Primitive Data Types & `type()` Function

**Why it is needed:**
Data types define the nature of operations that can be performed on a value. The built-in `type()` function inspects the runtime class of any variable.

**Core Types Covered:**
1. `int`: Whole numbers without decimals (e.g., `10`, `-5`).
2. `float`: Real numbers with fractional/decimal points (e.g., `3.14`, `-0.01`).
3. `str`: Textual sequence of characters (e.g., `"soban"`).
4. `bool`: Boolean truth values (`True` or `False`).
5. `None`: Represents the absence of a value or null state (`NoneType`).

```python
x = 10
print("x =", x, "| Type:", type(x))

pi_val = 3.14
print("pi_val =", pi_val, "| Type:", type(pi_val))

student_name = "Soban"
print("student_name =", student_name, "| Type:", type(student_name))

is_teacher = True
print("is_teacher =", is_teacher, "| Type:", type(is_teacher))

empty_var = None
print("empty_var =", empty_var, "| Type:", type(empty_var))
```

### Operation 4: Type Conversion (Implicit) vs Type Casting (Explicit)

**Why it is needed:**
Transforming data between types is critical when parsing numeric input from strings or calculating floating-point metrics.

**Two Conversion Mechanisms:**
1. **Implicit Type Conversion:** Python automatically upcasts types during operations to prevent precision loss (e.g., `int + float -> float`).
2. **Explicit Type Casting:** The developer manually casts using constructor functions such as `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`.

```python
# 1. Implicit conversion: int (5) + float (3.0) -> float (8.0)
a = 5
b = 3.0
result_implicit = a + b
print(f"Implicit conversion: {a} + {b} = {result_implicit} (Type: {type(result_implicit).__name__})")

# 2. Explicit type casting
num_int = 10
num_float = float(num_int)  # converts 10 -> 10.0
num_str = str(num_int)      # converts 10 -> "10"

print(f"Explicit float: {num_float} ({type(num_float).__name__})")
print(f"Explicit str: '{num_str}' ({type(num_str).__name__})")
```

### Operation 5: Arithmetic Operators

**Why it is needed:**
Arithmetic operators execute fundamental mathematical calculations between numeric operands.

| Operator | Operation | Description | Example (`a=5, b=10`) |
| :--- | :--- | :--- | :--- |
| `+` | Addition | Sum of operands | `5 + 10 = 15` |
| `-` | Subtraction | Difference | `5 - 10 = -5` |
| `*` | Multiplication | Product | `5 * 10 = 50` |
| `/` | Division | Quotient (always returns `float`) | `5 / 10 = 0.5` |
| `%` | Modulo | Remainder of division | `5 % 10 = 5` |
| `**` | Exponentiation | Base raised to the power of exponent | `5 ** 10 = 9765625` |

```python
a = 5
b = 10

print("a + b =", a + b)   # Addition
print("a - b =", a - b)   # Subtraction
print("a * b =", a * b)   # Multiplication
print("a / b =", a / b)   # Division
print("a % b =", a % b)   # Modulo (remainder)
print("a ** b =", a ** b) # Power (5^10)
```

### Operation 6: Relational (Comparison) Operators

**Why it is needed:**
Relational operators compare two values and evaluate to a boolean result (`True` or `False`), forming the condition tests for conditional branching and filtering.

| Operator | Meaning | Example (`a=10, b=20`) | Output |
| :--- | :--- | :--- | :--- |
| `==` | Equal to | `a == b` | `False` |
| `!=` | Not equal to | `a != b` | `True` |
| `>` | Greater than | `a > b` | `False` |
| `<` | Less than | `a < b` | `True` |
| `>=` | Greater than or equal to | `a >= b` | `False` |
| `<=` | Less than or equal to | `a <= b` | `True` |

```python
a = 10
b = 20

print("a == b:", a == b) # Equal to
print("a != b:", a != b) # Not equal to
print("a > b:",  a > b)  # Greater than
print("a < b:",  a < b)  # Less than
print("a >= b:", a >= b) # Greater than or equal to
print("a <= b:", a <= b) # Less than or equal to
```

### Operation 7: Assignment Operators (Compound / In-Place)

**Why it is needed:**
Assignment operators update the value stored in a variable by performing an arithmetic operation directly on its existing value (`x += 5` is shorthand for `x = x + 5`).

```python
x = 10
print("Initial x =", x)

x += 5   # x = x + 5
print("x += 5 ->", x)

x -= 3   # x = x - 3
print("x -= 3 ->", x)

x *= 2   # x = x * 2
print("x *= 2 ->", x)

x /= 4   # x = x / 4 (returns float)
print("x /= 4 ->", x)

x %= 4   # x = x % 4
print("x %= 4 ->", x)

x **= 3  # x = x ** 3
print("x **= 3 ->", x)
```

### Operation 8: Logical Operators (`and`, `or`, `not`)

**Why it is needed:**
Logical operators combine or negate boolean conditions:
- `and`: Returns `True` only if **both** operands are `True`.
- `or`: Returns `True` if **at least one** operand is `True`.
- `not`: Reverses the boolean value (`not True -> False`, `not False -> True`).

```python
x = 10
y = 20
z = 5

# Logical AND: both conditions must be True
print("(x > z) and (y > x):", (x > z) and (y > x))  # (10 > 5) and (20 > 10) -> True

# Logical OR: at least one condition must be True
print("(x > y) or (y > z):", (x > y) or (y > z))    # (False) or (True) -> True

# Logical NOT: negates the condition
print("not (x > y):", not (x > y))                  # not(False) -> True
```

### Operation 9: Operator Precedence & Unary Operators

**Why it is needed:**
When multiple operators exist in a single expression, operator precedence determines the exact order of evaluation:

1. `()` - Parentheses (Highest priority)
2. `**` - Exponentiation
3. `+x`, `-x`, `~x` - Unary operators (operate on a single operand, e.g. `-x` for negation)
4. `*`, `/`, `//`, `%` - Multiplication, Division, Floor division, Modulo
5. `+`, `-` - Addition, Subtraction
6. Bitwise shifts (`<<`, `>>`), Bitwise AND (`&`), XOR (`^`), OR (`|`)
7. Comparison operators (`<`, `<=`, `>`, `>=`, `!=`, `==`)
8. `not`
9. `and`
10. `or`
11. Assignment operators (`=`, `+=`, `-=`, etc.)

```python
# 1. Unary operator demonstration
val = 5
neg_val = -val  # Unary minus
print("val =", val, "| -val =", neg_val)

# 2. Operator precedence evaluation
# In 10 + 3 * 2 ** 2:
# Step 1: 2 ** 2 = 4 (Exponent)
# Step 2: 3 * 4 = 12 (Multiplication)
# Step 3: 10 + 12 = 22 (Addition)
res = 10 + 3 * 2 ** 2
print("Result of 10 + 3 * 2 ** 2 =", res)
```
