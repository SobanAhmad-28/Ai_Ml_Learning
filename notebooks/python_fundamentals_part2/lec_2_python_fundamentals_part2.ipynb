### Operation 1: Basic Conditional Branching (`if`, `elif`, `else`)

**Why it is needed:**
Conditional statements evaluate boolean conditions to route code execution:
- `if`: Tests the primary condition; runs the indented block if `True`.
- `elif` ("else if"): Evaluates secondary conditions if previous tests evaluate to `False`.
- `else`: Catch-all fallback block executed when all preceding conditions are `False`.

**Rule:** `if` can exist standalone, but `elif` and `else` mandate a preceding `if` statement.

```python
# Example 1: Age verification
age = 20
if age >= 18:
    print("Eligible: You can vote")
    print("Eligible: You can drive")
else:
    print("Ineligible: You can't vote")
    print("Ineligible: You can't drive")

# Example 2: Traffic Light Router
color = "green"
if color == "red":
    print("Signal: Stop")
elif color == "yellow":
    print("Signal: Look & Prepare")
elif color == "green":
    print("Signal: Go")
else:
    print("Signal: Invalid color")
```

### Operation 2: Compound Logical Conditions & Nested Decision Trees

**Why it is needed:**
Complex decision boundaries require combining conditions with `and`, `or`, and `not`, or placing conditional blocks inside one another (nesting) to resolve hierarchical criteria (e.g. credentials verification).

```python
# Compound conditions with logical operators
person_age = 15
if person_age < 13:
    stage = "child"
elif person_age >= 13 and person_age < 18:
    stage = "teenager"
else:
    stage = "adult"
print(f"Age {person_age} classified as: {stage}")

# Nested conditionals: Authentication guard
user_input = "admin"
pass_input = "wrong_pass"

if user_input == "admin" and pass_input == "pass":
    print("Authentication: Log in successful!")
else:
    if user_input != "admin":
        print("Auth Error: Wrong username, try again.")
    else:
        print("Auth Error: Wrong password, try again.")
```

### Operation 3: Ternary Conditional Expressions & Match-Case Statements

**Why it is needed:**
- **Ternary Operator:** `value_if_true if condition else value_if_false` provides concise, inline variable assignment.
- **Match-Case (Python 3.10+):** Structural pattern matching alternative to deep `if-elif-else` ladders.

```python
# 1. Ternary inline assignment
check_age = 18
status = "Adult" if check_age >= 18 else "Not Adult"
print("Ternary status result:", status)

# 2. Match-Case structural pattern
traffic_light = "yellow"
match traffic_light:
    case "red":
        action = "Stop"
    case "yellow":
        action = "Look"
    case "green":
        action = "Go"
    case _:
        action = "Unknown State"
print(f"Match-case action for '{traffic_light}': {action}")
```

### Operation 4: The `while` Loop & Infinite Loop Safeguards

**Why it is needed:**
`while` loops execute repeatedly as long as a test expression remains `True`. Loop variables must be updated every iteration to prevent infinite runaway loops.

```python
# 1. Incrementing loop: 1 to 5
print("Counting 1 to 5:")
i = 1
while i <= 5:
    print(f"  Step {i}")
    i += 1

# 2. Decrementing loop: 5 to 1
print("Countdown 5 to 1:")
j = 5
while j > 0:
    print(f"  T-minus {j}")
    j -= 1
```

### Operation 5: Loop Controls — `break` and `continue`

**Why it is needed:**
- `break`: Terminates loop execution immediately upon meeting a condition.
- `continue`: Skips the remainder of the active iteration and cycles immediately to the next iteration.

```python
# break: Terminate on multiple of 6
print("Break demonstration (stop when reaching multiple of 6):")
idx = 1
while idx <= 10:
    if idx % 6 == 0:
        print(f"  Break triggered at {idx}!")
        break
    print(f"  Value: {idx}")
    idx += 1

# continue: Skip multiples of 3
print("Continue demonstration (skip multiples of 3):")
k = 0
while k < 10:
    k += 1
    if k % 3 == 0:
        continue
    print(f"  Kept value: {k}")
```

### Operation 6: The `for` Loop, Membership Operator `in` & Sequence Iteration

**Why it is needed:**
`for` loops iterate over sequential collections (strings, lists, tuples, or ranges). The `in` membership operator checks presence and yields consecutive items.

```python
# Iterating over a string
word = "Prime"
print("Characters in 'Prime':", [ch for ch in word])

# Membership check
if 'i' in word:
    print("Letter 'i' exists in word!")

# Character frequency counting
corpus = "artificial intelligence"
target_char = 'i'
char_count = 0
for ch in corpus:
    if ch == target_char:
        char_count += 1
print(f"Character '{target_char}' appears {char_count} times in '{corpus}'")

# Nested loops (Coordinate grid)
print("Nested loop grid:")
for r in range(1, 3):
    for c in range(1, 3):
        print(f"  Coord: ({r}, {c})", end=" ")
    print()
```

### Operation 7: The `range()` Function Anatomy

**Why it is needed:**
`range(start, stop, step)` generates an efficient immutable arithmetic sequence without allocating full array memory up-front:
- `range(stop)`: Starts at `0`, stops at `stop - 1`.
- `range(start, stop)`: Starts at `start`, stops at `stop - 1`.
- `range(start, stop, step)`: Steps by `step` increments.

```python
# 1 argument: stop=5
print("range(5):", list(range(5)))

# 2 arguments: start=1, stop=6
print("range(1, 6):", list(range(1, 6)))

# 3 arguments: start=1, stop=10, step=2
print("range(1, 10, 2):", list(range(1, 10, 2)))
```

### Operation 8: User-Defined Functions (`def`), Parameters vs Arguments & Default Values

**Why it is needed:**
Functions package reusable blocks. Values defined in function signatures are **parameters**; values passed during invocation are **arguments**.

**Rule:** Non-default parameters cannot follow default parameters (`def fnx(a, b=1):` is valid; `def fnx(a=1, b):` is a syntax error).

```python
# Function with return value
def calculate_average(a, b, c):
    """Computes the arithmetic mean of 3 numbers."""
    return (a + b + c) / 3

mean_val = calculate_average(10, 20, 30)
print("Calculated Mean (10, 20, 30):", mean_val)

# Function with default parameter
def increment(num, step=1):
    """Increments a number by step (default 1)."""
    return num + step

print("increment(5):", increment(5))        # uses default step=1
print("increment(5, 10):", increment(5, 10)) # overrides step with 10
```

### Operation 9: Anonymous Lambda Functions & Built-in Utilities

**Why it is needed:**
Lambda functions provide single-expression anonymous callbacks (`lambda args: expression`), frequently passed to `map()`, `filter()`, or sorting keys.

```python
# Lambda function for computing square: x^2
square = lambda x: x * x
print("Square of 5 via lambda:", square(5))

# Lambda for feature thresholding
relu = lambda x: x if x > 0 else 0.0
sample_weights = [-1.5, 0.8, -0.2, 3.4]
transformed = [relu(w) for w in sample_weights]
print(f"Raw Weights: {sample_weights}")
print(f"ReLU Transformed: {transformed}")
```

**Practice Problems Covered in Lecture:**
1. **Set 1:** Odd/Even & Multiple of 5 checker.
2. **Set 2:** Multiplication table of `n`, counting vowels in string, sum of first `n` natural numbers.
3. **Set 3:** Factorial of `n` & Finding largest of 3 numbers.

```python
# 1. Multiple of 5 check
def check_multiple_5(num):
    return "multiple of 5" if num % 5 == 0 else "NOT a multiple of 5"
print("Check 25:", check_multiple_5(25))

# 2. Multiplication table of 4 using while
print("Table of 4 (while loop):")
t_num = 4
t_idx = 1
while t_idx <= 5:  # Showing first 5 for brevity
    print(f"  {t_num} x {t_idx} = {t_num * t_idx}")
    t_idx += 1

# 3. Sum of first n numbers
n_val = 10
total_sum = sum(range(1, n_val + 1))
print(f"Sum of first {n_val} natural numbers:", total_sum)

# 4. Factorial calculation
def factorial(n):
    fact = 1
    for f in range(1, n + 1):
        fact *= f
    return fact
print("Factorial of 5:", factorial(5))

# 5. Largest of 3 numbers
def get_largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= c:
        return b
    else:
        return c
print("Largest of (3, 10, 5):", get_largest(3, 10, 5))
```
