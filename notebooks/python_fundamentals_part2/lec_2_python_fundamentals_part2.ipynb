### Operation 1: Shartein Lagana (`if`, `elif`, `else` Basics)

**Kyu zaroorat hai:**
- `if`: Pehli shart check karta hai; agar sahi (True) hai to indented code chalta hai.
- `elif` ("else if"): Agar pehli shart galat hui to doosri shart check karta hai.
- `else`: Agar upar ki saari shartein galat hui to aakhiri fallback block chalta hai.
- **Rule:** `if` akele bina `else` ke chal sakta hai, lekin `elif` aur `else` bina `if` ke kabhi nahi chal sakte.

```python
# Example 1: Age verification check
age = 20
if age >= 18:
    print("Eligible: Aap vote de sakte hain")
    print("Eligible: Aap drive kar sakte hain")
else:
    print("Not eligible: Aap abhi vote nahi de sakte")

# Example 2: Traffic Light Router
color = "green"
if color == "red":
    print("Signal: Ruko (Stop)")
elif color == "yellow":
    print("Signal: Dekho aur taiyaar raho (Look)")
elif color == "green":
    print("Signal: Jao (Go)")
else:
    print("Signal: Galat color enter kiya")
```

### Operation 2: Compound Conditions aur Nested Conditionals

**Kyu zaroorat hai:**
- **Compound conditions:** Ek se zyada shartein check karna using `and`, `or`, `not`.
- **Nesting:** Ek condition ke andar doosri condition lagana (jaise Login system: pehle username check hua, fir password check hua).

```python
# Compound condition (Logical operators ke sath)
person_age = 15
if person_age < 13:
    print("Category: Child (Bachha)")
elif person_age >= 13 and person_age < 18:
    print("Category: Teenager")
else:
    print("Category: Adult")

# Nested conditionals: Login System
username = "admin"
password = "pass"
if username == "admin" and password == "pass":
    print("Login successful!")
else:
    if username != "admin":
        print("Galti: Galat username")
    else:
        print("Galti: Galat password")
```

### Operation 3: Ternary One-Liner aur Match-Case

**Kyu zaroorat hai:**
- **Ternary Operator:** Ek hi line me `if-else` likhne ka smart tareeqa: `value_if_true if condition else value_if_false`.
- **Match-Case (Python 3.10+):** Lambi `if-elif-else` chain ke badle saaf-suthra pattern matching.

```python
# 1. Ternary statement (One-liner)
check_age = 18
status = "Adult" if check_age >= 18 else "Not Adult"
print("Ternary Result:", status)

# 2. Match-Case pattern matching
light = "yellow"
match light:
    case "red":
        action = "Stop"
    case "yellow":
        action = "Look"
    case "green":
        action = "Go"
    case _:
        action = "Invalid color"
print(f"Light '{light}' par action:", action)
```

### Operation 4: `while` Loop aur Infinite Loop se Bachna

**Kyu zaroorat hai:**
`while` loop tab tak chalta hai jab tak uski shart `True` rehti hai. Hamesha loop ke andar counter badhana (`i += 1`) zaroori hota hai, warna loop kabhi nahi rukega aur **Infinite Loop** ban jayega.

```python
# 1. 1 se 5 tak counting print karna
print("Counting 1 to 5:")
i = 1
while i <= 5:
    print(f"  Number {i}")
    i += 1

# 2. 5 se 1 tak ulti counting (countdown)
print("Countdown 5 to 1:")
j = 5
while j > 0:
    print(f"  Countdown {j}")
    j -= 1
```

### Operation 5: Loop Controls — `break` aur `continue`

**Kyu zaroorat hai:**
- `break`: Loop ko turant wahi par khatam (exit) kar deta hai.
- `continue`: Current round ko wahi chhod kar seedha agle round par jump karta hai (skip karna).

```python
# break: Jaise hi 6 ka multiple mile turant ruk jao
print("Break demo (6 par ruk jao):")
num = 1
while num <= 10:
    if num % 6 == 0:
        print(f"  Break ho gaya {num} par!")
        break
    print(f"  Chala: {num}")
    num += 1

# continue: 3 ke multiples ko skip karo
print("Continue demo (3 ke multiples skip):")
k = 0
while k < 10:
    k += 1
    if k % 3 == 0:
        continue  # skip ho gaya
    print(f"  Bacha number: {k}")
```

### Operation 6: `for` Loop, Membership Operator `in` aur Sequence Traversal

**Kyu zaroorat hai:**
`for` loop kisi bhi sequence (string, list, range) ke har element par ek-ek karke ghoomta hai. `in` keyword se hum check kar sakte hain ki koi character ya item sequence me maujood hai ya nahi.

```python
# 1. String ke har character par loop
word = "Prime"
for ch in word:
    print("Letter:", ch)

# 2. Membership check ('in')
if 'i' in word:
    print("Ha, 'i' word ke andar maujood hai!")

# 3. Character count karna
sentence = "artificial intelligence"
count_i = 0
for c in sentence:
    if c == 'i':
        count_i += 1
print(f"'i' total {count_i} baar aaya hai.")
```

### Operation 7: `range()` Function ka Istemal

**Kyu zaroorat hai:**
`range()` function numbers ki list memory me store kiye bina ek-ek karke numbers generate karta hai, jo memory-efficient hota hai:
- `range(stop)`: 0 se lekar `stop - 1` tak chalta hai.
- `range(start, stop)`: `start` se shuru hokar `stop - 1` par rukta hai.
- `range(start, stop, step)`: `step` ke gap par chalta hai.

```python
print("range(5):", list(range(5)))          # 0, 1, 2, 3, 4
print("range(1, 6):", list(range(1, 6)))     # 1, 2, 3, 4, 5
print("range(1, 10, 2):", list(range(1, 10, 2))) # 1, 3, 5, 7, 9 (2 ka jump)
```

### Operation 8: Functions (`def`), Parameters, Arguments aur Default Values

**Kyu zaroorat hai:**
Code ko reusable banane ke liye hum functions banate hain:
- **Parameters:** Function definition me jo variables likhe hote hain (`def sum(a, b)`).
- **Arguments:** Function call karte waqt jo asli values pass karte hain (`sum(5, 10)`).
- **Default Parameters:** Agar caller argument na de toh default value use ho jati hai (`def sum(a, b=1)`).

```python
# Function with return statement
def calculate_avg(a, b, c):
    return (a + b + c) / 3

print("Average (10, 20, 30):", calculate_avg(10, 20, 30))

# Function with default parameter
def add_numbers(a, b=1):
    return a + b

print("Default b=1 use kiya: ", add_numbers(5))    # 5 + 1 = 6
print("Manual b=10 pass kiya:", add_numbers(5, 10)) # 5 + 10 = 15
```

### Operation 9: Built-in Functions aur Anonymous Lambda Functions

**Kyu zaroorat hai:**
- Built-in functions (`print`, `input`, `type`, `range`, `len`) pehle se bane hote hain.
- **Lambda Function:** Chhota one-liner anonymous function hota hai jo bina naam ke banta hai (`lambda x: x * x`).

```python
# Lambda function: Square nikalna
square = lambda x: x * x
print("5 ka square (lambda se):", square(5))

# Lambda se simple ReLU activation transformation
relu = lambda x: max(0, x)
print("ReLU of -4:", relu(-4))
print("ReLU of 6: ", relu(6))
```

### Operation 10: Lecture ke In-Class Practice Problems ke Solutions

**Solved Practice Sets:**
1. Number 5 ka multiple hai ya nahi?
2. Number Odd hai ya Even?
3. Multiplication table using `while` loop.
4. Factorial of $N$.
5. Teen numbers me se largest nikalna.

```python
# 1. Multiple of 5 check
num = 25
print(f"{num} is:", "Multiple of 5" if num % 5 == 0 else "NOT a multiple of 5")

# 2. Multiplication table of 4 (while loop)
print("Table of 4:")
t = 1
while t <= 5:
    print(f"  4 x {t} = {4 * t}")
    t += 1

# 3. Factorial of N
def factorial(n):
    fact = 1
    for f in range(1, n + 1):
        fact *= f
    return fact
print("5 ka factorial:", factorial(5))

# 4. Largest of 3 numbers
def get_largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= c:
        return b
    else:
        return c
print("Largest of (3, 10, 5):", get_largest(3, 10, 5))
```
