### Operation 1: Conditional Statements in Python (`if`, `elif`, `else` Syntax)

Programming me jab bhi hume kisi shart (condition) ke base par faisla lena ho, to hum conditional statements use karte hain.
- **`if`:** Pehli shart check karta hai. Agar True ho to code block chalega.
- **`elif` (Else If):** Agar pehla `if` False ho, to agla `elif` check hota hai.
- **`else`:** Jab saari shartein False ho jayein, tab aakhri default block chalta hai.

**Dhyan dene wali baat:** Python me curly braces `{}` nahi hote; yaha code block **4 spaces ki indentation** se pehchana jata hai!


````python
# Operation 1: Basic if-elif-else Syntax
marks = 85

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
else:
    grade = "C"

print(f"Student Marks: {marks} | Final Grade: {grade}")
````

### Operation 2: Practice Example — Traffic Light & Age Eligibility System

Lecture transcript me Shradha ma'am ne do real-world examples samjhaye hain:
1. **Traffic Light System:** `"red"` hone par STOP, `"yellow"` hone par LOOK / SLOW, aur `"green"` hone par GO!
2. **Age Eligibility:** Check karna ki user driving license aur voting ke liye eligible hai ya nahi (age >= 18).


````python
# Operation 2: Traffic Light System & Age Check
# Example A: Traffic Light
light = "green"

if light == "red":
    print("Action: STOP (Ruko!)")
elif light == "yellow":
    print("Action: LOOK & SLOW DOWN (Dekho aur dheere chalo)")
elif light == "green":
    print("Action: GO (Ja sakte ho)")
else:
    print("Invalid traffic light color!")

# Example B: Age Eligibility
user_age = 20

if user_age >= 18:
    print(f"Age {user_age}: You are an ADULT. Eligible to vote and drive!")
else:
    print(f"Age {user_age}: You are a MINOR. Not eligible yet.")
````

### Operation 3: Classic Problem — Odd or Even Number Checker

Computer science ka sabse classical problem: Diye gaye number N ko check karna ki wo **Even (Sam)** hai ya **Odd (Visham)**.
- Agar koi number 2 se poora divide ho jaye aur remainder 0 bache (`n % 2 == 0`), to wo **Even** hota hai.
- Agar remainder 1 bache, to wo **Odd** hota hai.


````python
# Operation 3: Odd or Even Checker
def check_odd_even(n):
    if n % 2 == 0:
        return f"{n} is EVEN (2 se poora divide hota hai)"
    else:
        return f"{n} is ODD (2 se divide karne par 1 bachta hai)"

print(check_odd_even(14))
print(check_odd_even(27))
print(check_odd_even(0))
````

### Operation 4: Nesting — Conditions ke andar Conditions

Nesting ka matlab hai ek `if` ya `else` ke andar doosra `if-else` likhna.

**Classroom Example:**
User ki driving eligibility check karni hai:
1. Pehle check karo ki `age >= 18` hai ya nahi.
2. Agar haan, to andar ek aur condition check karo ki kya uske paas `has_license` hai ya nahi!


````python
# Operation 4: Nested Conditionals
age = 22
has_license = True

if age >= 18:
    print("Primary check passed: Age is 18 or above.")
    if has_license:
        print("Final Status: You can legally drive car/bike on road!")
    else:
        print("Final Status: Age valid, but you CANNOT drive without a License. Please apply first.")
else:
    print(f"Primary check failed: Age is {age}. Driving is strictly prohibited for minors.")
````

### Operation 5: Modern Python — `match-case` Statement (Python 3.10+)

Python 3.10 me ek naya powerful feature add kiya gaya: `match-case` (jo doosri languages me `switch-case` jaisa hota hai).
Jab ek hi variable ko 4-5 fixed values se compare karna ho, to baar-baar `elif` likhne ke bajaye `match-case` likhna bohot clean aur readable hota hai.
- `case _`: Default case (agar koi bhi case match na ho).


````python
# Operation 5: Python match-case Syntax
day_number = 3

match day_number:
    case 1:
        day_name = "Monday (Week start)"
    case 2:
        day_name = "Tuesday"
    case 3:
        day_name = "Wednesday (Mid-week)"
    case 4:
        day_name = "Thursday"
    case 5:
        day_name = "Friday (Weekend coming)"
    case 6 | 7:
        day_name = "Weekend (Holiday!)"
    case _:
        day_name = "Invalid day number!"

print(f"Day {day_number} is: {day_name}")
````

### Operation 6: Loops using `while` — Syntax & Avoiding Infinite Loops

Loop ka matlab hota hai kisi code block ko bar-bar repeat karna jab tak ek condition True rahe.

`while` loop ke teen sabse zaroori steps hote hain:
1. **Initialization:** Counter variable banana (e.g. `i = 1`).
2. **Condition:** Check karna ki kab tak chalna hai (e.g. `while i <= 5:`).
3. **Update / Increment:** Counter ko badhana (e.g. `i += 1`).

Warning: Agar aap update step (`i += 1`) bhool gaye, to condition hamesha True rahegi aur loop **Infinite Loop** ban kar program ko crash kar dega!


````python
# Operation 6: while Loop Basics
count = 1

while count <= 5:
    print(f"Iteration {count}: Hello Python!")
    count += 1

print("Loop successfully finished!")
````

### Operation 7: Practice Example — Printing Numbers 1 to 5 & Reverse 5 to 1

Lecture transcript me loop control flow samjhne ke liye forward aur reverse counting solve karwayi gayi hai:
- Forward counting: `count = 1` se start karke `count += 1` karte huye 5 tak jana.
- Reverse counting: `count = 5` se start karke `count -= 1` karte huye 1 tak aana.


````python
# Operation 7: Forward and Reverse Counting
print("--- Forward: 1 to 5 ---")
i = 1
while i <= 5:
    print(i, end=" ")
    i += 1
print()

print("--- Reverse: 5 to 1 ---")
j = 5
while j >= 1:
    print(j, end=" ")
    j -= 1
print()
````

### Operation 8: Classic Problem — Multiplication Table of N

Classroom practice problem: Kisi bhi input number N ka multiplication table (1 se 10 tak) loop ke zariye format karke print karna.


````python
# Operation 8: Multiplication Table of N
n = 7
print(f"=== MULTIPLICATION TABLE OF {n} ===")

i = 1
while i <= 10:
    print(f"{n} x {i:2d} = {n * i}")
    i += 1
````

### Operation 9: Loop Controls — `break` aur `continue` Keywords

Loops ko control karne ke do sabse powerful keywords:
1. **`break`:** Loop ko **turant terminate (khatam)** karke bahar nikal deta hai.
2. **`continue`:** Current iteration ko **skip** karke seedhe agle iteration par chala jata hai.


````python
# Operation 9: break and continue Demonstration
print("--- 'break' Example (Find 3 and Stop) ---")
i = 1
while i <= 5:
    if i == 3:
        print(f"Found target {i}! Breaking out of loop immediately.")
        break
    print(f"Processing number: {i}")
    i += 1

print("--- 'continue' Example (Skip Even Numbers, Print Only Odd) ---")
j = 0
while j < 6:
    j += 1
    if j % 2 == 0:
        continue
    print(f"Odd Number: {j}")
````

### Operation 10: Loops using `for` — Sequence Traversal (Strings & Lists)

`for` loop Python me kisi bhi **sequence (Strings, Lists, Tuples, Ranges)** ke upar ek-ek item ko traverse karne ke liye use hota hai.
- `while` loop me counter initialize aur increment khud karna padta hai.
- `for` loop automatically pehle element se aakhri element tak khud chalta hai!


````python
# Operation 10: for Loop Traversal
word = "Python"
print("Traversing String:")
for char in word:
    print(f"  Character: {char}")

languages = ["Python", "SQL", "Machine Learning"]
print("Traversing List:")
for idx, lang in enumerate(languages, start=1):
    print(f"  {idx}. {lang}")
````

### Operation 11: Classic Problem — Vowel Count in a String

Classroom problem: Diye gaye text/string ke andar kitne **Vowels (`a, e, i, o, u`)** hain, unhe count karna.
- Hum ek `vowel_count = 0` counter banayenge.
- `for` loop se har character par jayenge.
- Agar character vowels set me present hai, to counter badhaenge.


````python
# Operation 11: Vowel Count in String
text = "Artificial Intelligence and Machine Learning"
vowels = "aeiouAEIOU"
count = 0

for ch in text:
    if ch in vowels:
        count += 1

print(f"Input Text: '{text}'")
print(f"Total Vowel Count: {count}")
````

### Operation 12: The `range()` Function — Start, Stop aur Step

`range()` function numbers ki ek sequence generate karta hai:
1. **`range(stop)`:** 0 se shuru hoke $stop - 1$ tak chalta hai.
2. **`range(start, stop)`:** $start$ se shuru hoke $stop - 1$ tak.
3. **`range(start, stop, step)`:** $step$ ke interval se aage badhta hai.


````python
# Operation 12: range() Function 3 Variations
print("1. range(5):", list(range(5)))
print("2. range(2, 7):", list(range(2, 7)))
print("3. range(1, 10, 2) [Odd Numbers]:", list(range(1, 10, 2)))
print("4. range(10, 0, -2) [Reverse Even Numbers]:", list(range(10, 0, -2)))
````

### Operation 13: Classic Problem — Sum of First N Natural Numbers

Classroom problem: Pehle N natural numbers ka sum ($1 + 2 + 3 + \dots + n$) calculate karna:
1. **Loop method:** Loop chala kar ek-ek number ko `total_sum` me add karna.
2. **Formula method:** Mathematical formula n * (n + 1) // 2 se verify karna.


````python
# Operation 13: Sum of First N Numbers
n = 10
loop_sum = 0

for i in range(1, n + 1):
    loop_sum += i

formula_sum = (n * (n + 1)) // 2

print(f"Sum of first {n} numbers using Loop:    {loop_sum}")
print(f"Sum of first {n} numbers using Formula: {formula_sum}")
print("Do both match?:", loop_sum == formula_sum)
````

### Operation 14: Functions in Python — Definition (`def`), Parameters & `return`

Function reusable code ka ek block hota hai jo tabhi execute hota hai jab use call kiya jaye.

**Kyu zaroori hai? (DRY Principle — Don't Repeat Yourself):**
Agar aapko 10 alag-alag jagah par do numbers ko add karke percentage nikalna hai, to baar-baar 10 lines likhne ke bajaye ek function bana do aur use 1 line me call karo!


````python
# Operation 14: Function Definition and Return
def calculate_simple_interest(principal, rate, time):
    si = (principal * rate * time) / 100
    total_amount = principal + si
    return si, total_amount

interest, total = calculate_simple_interest(10000, 7.5, 2)
print("Principal: ₹10,000 | Rate: 7.5% | Time: 2 Years")
print(f"Earned Interest: ₹{interest:.2f}")
print(f"Total Maturity Amount: ₹{total:.2f}")
````

### Operation 15: Practice Example — Average of 3 Numbers (`calc_avg`)

Classroom practice problem: Ek aisa function likhna jo 3 input numbers ($a, b, c$) le aur unka **Average (Ausat)** calculate karke return kare:
Average = (a + b + c) / 3


````python
# Operation 15: Average of 3 Numbers Function
def calc_avg(a, b, c):
    avg = (a + b + c) / 3
    return avg

res1 = calc_avg(10, 20, 30)
res2 = calc_avg(45, 60, 75)

print(f"Average of (10, 20, 30): {res1:.2f}")
print(f"Average of (45, 60, 75): {res2:.2f}")
````

### Operation 16: Types of Functions — Built-in vs User-Defined Functions

Python me functions do categories ke hote hain:
1. **Built-in Functions:** Jo Python ke sath pehle se aate hain (`print()`, `len()`, `type()`, `range()`, `sum()`, `max()`, `min()`).
2. **User-Defined Functions:** Jo developer apni zaroorat ke hisaab se `def` keyword ka use karke banata hai.


````python
# Operation 16: Built-in vs User-Defined Demo
numbers = [12, 45, 78, 23, 56, 89, 34]

print("Built-in Functions:")
print("  len(numbers):", len(numbers))
print("  max(numbers):", max(numbers))
print("  min(numbers):", min(numbers))
print("  sum(numbers):", sum(numbers))

def find_range_span(nums):
    return max(nums) - min(nums)

print("User-Defined Function:")
print("  find_range_span(numbers):", find_range_span(numbers))
````

### Operation 17: Lambda Functions — Single-Line Anonymous Functions

Lambda function ek aisa chhota function hota hai jiska koi naam nahi hota (Anonymous) aur use sirf **ek hi line** me define kiya jata hai.

**Syntax:**
```python
lambda arguments: expression
```


````python
# Operation 17: Lambda Functions
square_lambda = lambda x: x ** 2
print("Square of 6 using lambda:", square_lambda(6))

add_lambda = lambda a, b: a + b
print("Add (15, 25) using lambda:", add_lambda(15, 25))

is_even_lambda = lambda n: n % 2 == 0
print("Is 18 even?:", is_even_lambda(18))
print("Is 21 even?:", is_even_lambda(21))
````

### Operation 18: Classic Problem — Factorial of N (`calc_factorial`)

Day 03 ka final master problem: Kisi bhi number N ka **Factorial (N!)** calculate karne ka function likhna.
N! = N * (N - 1) * ... * 1
- Special Case: $0! = 1$ aur $1! = 1$.
- Agar $N < 0$ ho, to negative numbers ka factorial defined nahi hota.


````python
# Operation 18: Factorial of N Function
def calc_factorial(n):
    if n < 0:
        return "Factorial not defined for negative numbers!"
    elif n == 0 or n == 1:
        return 1
    
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

for test_val in [0, 1, 4, 5, 7]:
    print(f"Factorial of {test_val}! = {calc_factorial(test_val)}")
````
