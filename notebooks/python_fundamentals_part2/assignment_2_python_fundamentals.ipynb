### Q1: Income Tax Rate Calculator

**Sawal:** Salary ke input par tax rate nikalo:
- Salary $< 30,000 \rightarrow 5\%$
- Salary $30,000 - 70,000 \rightarrow 15\%$
- Salary $> 70,000 \rightarrow 25\%$

**Logic (Hinglish):** `if-elif-else` conditions se salary ka bracket check karenge.

```python
def calculate_tax(salary):
    if salary < 30000:
        rate = 5.0
    elif salary <= 70000:
        rate = 15.0
    else:
        rate = 25.0
    tax = salary * (rate / 100.0)
    return rate, tax

for s in [25000, 50000, 95000]:
    r, t = calculate_tax(s)
    print(f"Salary: ₹{s:,} -> Tax Rate: {r}% | Tax Amount: ₹{t:,.2f}")
```

### Q2: `a` aur `b` ke beech ke sabhi Even Numbers Print Karna

**Sawal:** Do integers `a` aur `b` lekar unke beech ke sabhi even numbers (inclusive) print karne ka function likho.

**Logic (Hinglish):** `range(a, b + 1)` par loop chalayenge aur jo number 2 se divide ho jaye (`n % 2 == 0`), use chunenge.

```python
def print_even_numbers(a, b):
    evens = [n for n in range(a, b + 1) if n % 2 == 0]
    print(f"{a} aur {b} ke beech ke even numbers: {evens}")

print_even_numbers(3, 15)
```

### Q3: Number ke Har Digit ko Print Karna

**Sawal:** Kisi number $n$ ke sabhi digits ko print karo (e.g. $n = 312$ me digits `3, 1, 2`).

**Math Hint (Hinglish):**
- Right-most digit nikalne ke liye: `n % 10`
- Right-most digit hatane ke liye: `n // 10`

```python
def print_digits(n):
    digits = []
    temp = abs(n)
    while temp > 0:
        digits.append(temp % 10)
        temp //= 10
    digits.reverse()  # Asli order restore karna
    print(f"{n} ke digits:", digits)

print_digits(312)
```

### Q4: Number ke Digits ka Total Count

**Sawal:** Number $n$ me kitne digits hain unka total count return karne ka function banao.

**Logic (Hinglish):** Number ko 10 se divide karte rahenge jab tak wo 0 na ho jaye aur count badhate rahenge.

```python
def count_digits(n):
    num = abs(n)
    if num == 0:
        return 1
    count = 0
    while num > 0:
        count += 1
        num //= 10
    return count

print("312 me digits:", count_digits(312))
print("100050 me digits:", count_digits(100050))
```

### Q5: Number ke Digits ka Sum

**Sawal:** Number $n$ ke sabhi digits ka jod (sum) return karne ka function banao (e.g. $312 \rightarrow 3+1+2 = 6$).

**Logic (Hinglish):** Har baar `n % 10` se aakhiri digit nikaal kar total me add karenge.

```python
def sum_of_digits(n):
    num = abs(n)
    total = 0
    while num > 0:
        total += num % 10
        num //= 10
    return total

print("312 ke digits ka sum:", sum_of_digits(312))
print("9876 ke digits ka sum:", sum_of_digits(9876))
```

### Q8: Simple Calculator Function

**Sawal:** Ek function `calculator(a, b, operation)` banao jo operation (`'+'`, `'-'`, `'*'`, `'/'`) ke hisaab se answer de.

```python
def calculator(a, b, op):
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        return a / b if b != 0 else "Error: Zero se divide nahi kar sakte"
    return "Unsupported operator"

print("10 + 5 =", calculator(10, 5, '+'))
print("10 * 5 =", calculator(10, 5, '*'))
print("10 / 0 =", calculator(10, 0, '/'))
```

### Q9: Prime Number Checker Function

**Sawal:** Ek function `is_prime(n)` likho jo number prime hai ya nahi bataye.

**Logic (Hinglish):** 2 se chhota koi prime nahi hota. 2 se lekar $\sqrt{n}$ tak check karenge, agar kisi se bhi divide ho gaya to prime nahi hai.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

for test_val in [2, 7, 9, 13, 25, 29]:
    print(f"Number {test_val:2d} prime hai? -> {is_prime(test_val)}")
```

### Q10: Number Guessing Game (Khel)

**Sawal:** Secret number guess karne ka game banao jo bataye:
- "Too high" agar guess bada hai
- "Too low" agar guess chhota hai
- "Correct!" agar guess match ho jaye

```python
def guess_game(secret, guesses):
    print(f"Secret number: {secret}")
    for g in guesses:
        if g > secret:
            print(f"  Guess {g:2d} -> Too high (Bada hai)")
        elif g < secret:
            print(f"  Guess {g:2d} -> Too low (Chhota hai)")
        else:
            print(f"  Guess {g:2d} -> Correct! Jeet gaye!")
            break

guess_game(secret=42, guesses=[20, 55, 38, 42])
```
