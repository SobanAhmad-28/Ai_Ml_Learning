### Q1: Income Tax Rate Calculator

**Problem Statement:**
Write a program that takes `salary` as input. Using conditional statements, calculate the final tax rate based on these rules:
- If salary < 30,000 -> 5%
- If salary is 30,000–70,000 -> 15%
- If salary > 70,000 -> 25%

**Logic:**
Compare salary thresholds using `if-elif-else`.

```python
# Q1 Solution
def calculate_tax_rate(salary: float) -> float:
    if salary < 30000:
        return 5.0
    elif salary <= 70000:
        return 15.0
    else:
        return 25.0

# Testing brackets
test_salaries = [25000, 50000, 95000]
for sal in test_salaries:
    rate = calculate_tax_rate(sal)
    tax_amount = sal * (rate / 100.0)
    print(f"Salary: ₹{sal:,} | Tax Rate: {rate}% | Tax Due: ₹{tax_amount:,.2f}")
```

### Q2: Print All Even Numbers Between `a` and `b` (Inclusive)

**Problem Statement:**
Write a function that takes two integers `a` and `b` and prints all even numbers between them (inclusive).

**Logic:**
Ensure `a <= b` and step through `range(a, b + 1)`, printing numbers where `n % 2 == 0`.

```python
# Q2 Solution
def print_even_numbers(a: int, b: int):
    evens = [n for n in range(a, b + 1) if n % 2 == 0]
    print(f"Even numbers between {a} and {b}: {evens}")

print_even_numbers(3, 15)
print_even_numbers(10, 20)
```

### Q3: Print the Digits of a Number

**Problem Statement:**
Write a function that prints the digits of a number, `n`.
For eg: `n = 312`, there are 3 digits in it: `3`, `1` and `2`, and we need to print them.

**Mathematical Hint:**
- Right-most digit: `N % 10`
- Truncate right-most digit: `N = N // 10`

```python
# Q3 Solution
def print_digits(n: int):
    temp = abs(n)
    digits = []
    if temp == 0:
        digits = [0]
    else:
        while temp > 0:
            digits.append(temp % 10)
            temp //= 10
        digits.reverse()  # Restore original left-to-right order
    print(f"Digits of {n}:", ", ".join(map(str, digits)))

print_digits(312)
print_digits(9045)
```

### Q4: Count the Number of Digits in a Number

**Problem Statement:**
Write a function to return the count of digits in a number `n`.

**Logic:**
Repeatedly divide integer by 10 (`// 10`) until 0 while incrementing a counter.

```python
# Q4 Solution
def count_digits(n: int) -> int:
    num = abs(n)
    if num == 0:
        return 1
    count = 0
    while num > 0:
        count += 1
        num //= 10
    return count

print("Digits in 312:", count_digits(312))
print("Digits in 100050:", count_digits(100050))
print("Digits in 0:", count_digits(0))
```

### Q5: Sum of Digits of a Number

**Problem Statement:**
Write a function to return the sum of digits of a number, `n`.

**Logic:**
Extract each rightmost digit with `n % 10`, accumulate into sum, and reduce `n = n // 10`.

```python
# Q5 Solution
def sum_of_digits(n: int) -> int:
    num = abs(n)
    total = 0
    while num > 0:
        total += num % 10
        num //= 10
    return total

print("Sum of digits of 312 (3+1+2):", sum_of_digits(312))
print("Sum of digits of 9876:", sum_of_digits(9876))
```

### Q7: Continuous Input Loop with Sentinel "Quit"

**Problem Statement:**
Design a program to continuously input a number from user & print if it is positive or negative until the user enters `"Quit"`.

**Logic:**
Use a `while True:` loop, check `if val.lower() == 'quit': break`, and evaluate sign.

```python
# Q7 Solution (Simulated input batch to showcase loop mechanics)
sample_inputs = ["15", "-7", "0", "42", "Quit"]

print("Running continuous input evaluation loop:")
for user_entry in sample_inputs:
    print(f"User entered: '{user_entry}'")
    if user_entry.strip().lower() == "quit":
        print("Exiting program. Goodbye!")
        break
    
    num = float(user_entry)
    if num > 0:
        print("  -> Result: Positive Number")
    elif num < 0:
        print("  -> Result: Negative Number")
    else:
        print("  -> Result: Zero")
```

### Q8: Arithmetic Calculator Function

**Problem Statement:**
Create a function `calculator(a, b, operation)` that performs addition, subtraction, multiplication, or division based on the `operation` parameter (`'+'`, `'-'`, `'*'`, `'/'`).

```python
# Q8 Solution
def calculator(a: float, b: float, operation: str):
    if operation == '+':
        return a + b
    elif operation == '-':
        return a - b
    elif operation == '*':
        return a * b
    elif operation == '/':
        return a / b if b != 0 else "Error: Division by zero"
    else:
        return f"Error: Unsupported operation '{operation}'"

print("10 + 5  =", calculator(10, 5, '+'))
print("10 - 5  =", calculator(10, 5, '-'))
print("10 * 5  =", calculator(10, 5, '*'))
print("10 / 5  =", calculator(10, 5, '/'))
print("10 / 0  =", calculator(10, 0, '/'))
```

### Q9: Prime Number Checker Function

**Problem Statement:**
Write a function `is_prime(n)` that returns `True` if `n` is a prime number and `False` otherwise, using a loop.

**Mathematical Rules:**
1. Numbers $< 2$ are not prime.
2. 2 is the smallest prime number.
3. Check divisors up to $\sqrt{n}$ or in range `[2, n-1]`. If divisible, return `False`.

```python
# Q9 Solution
def is_prime(n: int) -> bool:
    if n < 2:
        return False
    # Check divisors from 2 up to n - 1
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

# Testing prime evaluation
test_nums = [1, 2, 3, 4, 7, 9, 13, 25, 29]
for t in test_nums:
    print(f"Number {t:2d} is prime? {is_prime(t)}")
```

### Q10: Number Guessing Game Logic

**Problem Statement:**
Given a secret number (already decided), write a program that evaluates user guesses and prints:
- `"Too high"` if the guess is above the number
- `"Too low"` if the guess is below
- `"Correct!"` if the guess matches

```python
# Q10 Solution
def number_guessing_game(secret_number: int, guesses: list):
    print(f"Secret number locked. Evaluating guesses: {guesses}")
    for g in guesses:
        if g > secret_number:
            print(f"  Guess {g:2d} -> Too high")
        elif g < secret_number:
            print(f"  Guess {g:2d} -> Too low")
        else:
            print(f"  Guess {g:2d} -> Correct! You won!")
            break

number_guessing_game(secret_number=42, guesses=[15, 60, 35, 48, 42])
```
