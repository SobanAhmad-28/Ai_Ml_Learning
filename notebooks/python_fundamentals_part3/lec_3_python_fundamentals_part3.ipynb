### Operation 1: Strings in Python — Indexing, Negative Indexing aur Immutability

String characters ka ek sequence hota hai jo quotes (`' '`, `" "`, ya `''' '''`) ke andar band hota hai.
- **Positive Indexing:** Left se right ($0$ se shuru hoke $len - 1$ tak).
- **Negative Indexing:** Right se left (Aakhri character hamesha $-1$, usse pehla $-2$).
- **Immutability (Aparivartaniya):** Python me string ko create karne ke baad kisi specific index ke character ko direct change nahi kiya ja sakta (`str[0] = 'X'` error dega!).


````python
# Operation 1: String Indexing and Immutability
word = "ApnaCollege"

print(f"Original Word: {word}")
print(f"Length of Word: {len(word)}")
print(f"First character (Index 0):  {word[0]}")
print(f"Fifth character (Index 4):  {word[4]}")
print(f"Last character  (Index -1): {word[-1]}")
print(f"Second last     (Index -2): {word[-2]}")

# Proving Immutability:
try:
    word[0] = "O"
except TypeError as e:
    print(f"Immutability Proof: TypeError caught -> '{e}'")
````

### Operation 2: Slicing in Strings — `string[start:stop:step]` aur String Reversal

Slicing ka matlab hota hai kisi string ka ek chhota hissa (substring) nikalna.

**Syntax:**
```python
string[start_index : stop_index : step]
```
- `start`: Kaha se shuru karna hai (included).
- `stop`: Kaha tak jana hai (**excluded**, yaani stop se 1 pehle rukega).
- `step`: Kitne steps aage badhna hai (default 1).
- **Trick:** String ko reverse karne ke liye step me `-1` de diya jata hai: `string[::-1]`.


````python
# Operation 2: String Slicing Demonstration
phrase = "MachineLearning"

print("Original Phrase:", phrase)
print("1. phrase[0:7] (0 se 6 tak):    ", phrase[0:7])
print("2. phrase[7:]  (7 se end tak):    ", phrase[7:])
print("3. phrase[:7]  (start se 6 tak):  ", phrase[:7])
print("4. phrase[::2] (Every 2nd char):  ", phrase[::2])
print("5. phrase[::-1] (String Reversal):", phrase[::-1])
print("6. phrase[-8:-1]:                 ", phrase[-8:-1])
````

### Operation 3: Modern String Formatting — `f-strings` vs `.format()` Method

Jab hume text ke beech me variables ya mathematical calculations ko embed karna ho, to hum String Formatting use karte hain.
1. **f-strings (Fastest & Modern Python 3.6+):** String ke aage `f` lagao aur variable ko `{variable_name}` ke andar likho. Number formatting jaise `{val:.2f}` bhi direct ho jati hai.
2. **`.format()` method:** Purana tareeqa jisme `{}` placeholder me values pass hoti hain.


````python
# Operation 3: String Formatting Comparison
student_name = "Soban"
course = "AI & ML Prime"
score = 94.6789

# 1. Modern f-string
msg_fstring = f"Hello {student_name}, welcome to {course}! Your final score is {score:.2f}%."
print("f-string output:")
print(" ", msg_fstring)

# 2. .format() method
msg_format = "Hello {}, welcome to {}! Your final score is {:.2f}%.".format(student_name, course, score)
print(".format() output:")
print(" ", msg_format)
````

### Operation 4: Lists in Python — Creation, Mixed Datatypes aur In-Place Mutability

List Python ka sabse popular data structure hai jo square brackets `[]` ke andar items ko store karta hai.
- **Ordered:** Items ka ek fixed index hota hai.
- **Mutable (Parivartaniya):** Strings ke ulat, List ke andar kisi bhi index ki value ko direct change kiya ja sakta hai!
- **Mixed Types:** Ek hi list ke andar Integer, Float, String, aur Boolean sab ek sath aa sakte hain.


````python
# Operation 4: Lists Creation and Mutability
marks_list = [75, 82, 90, 68, 95]
print("Initial Marks List:", marks_list)

marks_list[1] = 88
print("After updating index 1 to 88:", marks_list)

student_profile = ["Soban", 22, 8.85, True]
print("Mixed List Profile:", student_profile)
print("Profile Types:", [type(item).__name__ for item in student_profile])
````

### Operation 5: Essential List Methods — `append()`, `sort()`, `insert()`, `remove()`, `pop()`

Lists ke upar bar-bar use hone wale standard built-in methods:
- `list.append(val)`: List ke aakhri me naya item jodta hai.
- `list.sort()`: Numbers ya words ko ascending order me sort karta hai.
- `list.sort(reverse=True)`: Descending order me sort karta hai.
- `list.reverse()`: List ko bina sort kiye ulta kar deta hai.
- `list.insert(idx, val)`: Kisi specific index par element daal deta hai.
- `list.remove(val)`: Pehli baar aane wale specific element ko delete karta hai.
- `list.pop(idx)`: Given index wale element ko nikal kar return karta hai.


````python
# Operation 5: List Methods Demonstration
nums = [45, 12, 89, 23, 67, 12]
print("Original Numbers:", nums)

nums.append(99)
print("After append(99):", nums)

nums.sort()
print("After sort():    ", nums)
nums.sort(reverse=True)
print("After sort(desc):", nums)

nums.insert(2, 50)
print("After insert(2, 50):", nums)

nums.remove(12)
print("After remove(12):", nums)

popped_val = nums.pop()
print(f"Popped value: {popped_val} | Remaining list: {nums}")
````

### Operation 6: Using Loops with Lists — Traversal aur Linear Search Algorithm

Classroom problem: Ek list ke andar kisi target element ko dhoondna (**Linear Search**):
- Ek-ek element ko `for` loop se check karo.
- Jaise hi element match ho jaye, uska index print karo aur loop ko `break` kar do.
- Agar poori list khatam ho jaye aur element na mile, to `"Not Found"` report karo.


````python
# Operation 6: Linear Search on List using for loop
def linear_search(arr, target):
    for idx, element in enumerate(arr):
        if element == target:
            return f"Target {target} FOUND at index {idx}!"
    return f"Target {target} NOT FOUND in list."

test_list = [10, 25, 47, 83, 92, 56, 31]
print("Test List:", test_list)
print(linear_search(test_list, 83))
print(linear_search(test_list, 99))
````

### Operation 7: Tuples in Python — Syntax, Single-element Tuple `(1,)` aur Immutability

Tuple parenthesis `()` ke andar likha jane wala ek **Immutable** sequence hota hai.

⚠️ **Interview Trap (Single Element Tuple):**
Agar aap `t = (5)` likhoge, to Python use normal Integer samjhega!
Single element wala Tuple banane ke liye aakhri me comma lagana zaroori hai: `t = (5,)`!


````python
# Operation 7: Tuple Syntax and Immutability
not_a_tuple = (5)
real_tuple = (5,)

print(f"Type of (5):  {type(not_a_tuple)} -> Normal Integer!")
print(f"Type of (5,): {type(real_tuple)}  -> Real Tuple!")

coordinates = (28.6139, 77.2090)
print("Coordinates Tuple:", coordinates)

try:
    coordinates[0] = 30.0
except TypeError as e:
    print(f"Immutability Proof: TypeError caught -> '{e}'")
````

### Operation 8: Tuple Methods — `count()` aur `index()` with Performance Advantage

Kyunki Tuples immutable hote hain, isliye isme sirf 2 methods hote hain:
1. `tup.count(x)`: Element $x$ list me kitni baar aaya hai.
2. `tup.index(x)`: Element $x$ pehli baar kis index par mila.

**Performance Advantage:** Tuples memory me List se kam jagah lete hain aur inki iteration speed List se tez hoti hai.


````python
# Operation 8: Tuple Methods
grades = ("A", "B", "A", "C", "A", "B", "A")

print("Grades Tuple:", grades)
print("Count of 'A':", grades.count("A"))
print("Count of 'B':", grades.count("B"))
print("First index of 'C':", grades.index("C"))
````

### Operation 9: Dictionaries in Python — Key-Value Pairs, Keys Uniqueness aur Lookup

Dictionary curly braces `{}` ke andar `key: value` pairs store karti hai.
- **Fast Lookup:** $O(1)$ time complexity me direct key ke through data nikalta hai.
- **Keys Must Be Unique & Immutable:** Key sirf Strings, Numbers, ya Tuples ho sakti hain.
- **Values Can Be Anything:** Value me string, number, list, ya doosri dictionary bhi ho sakti hai.


````python
# Operation 9: Dictionary Basics
student_info = {
    "name": "Soban Ahmad",
    "course": "AI & ML",
    "semester": 5,
    "skills": ["Python", "SQL", "NumPy", "Pandas"],
    "is_enrolled": True
}

print("Student Info Dictionary:")
for k, v in student_info.items():
    print(f"  {k:12s} : {v}")

print()
print("Accessing 'name':", student_info['name'])
student_info["semester"] = 6
student_info["city"] = "Bhopal"
print("Updated Dictionary Keys:", list(student_info.keys()))
````

### Operation 10: Essential Dictionary Methods — `.keys()`, `.values()`, `.items()`, `.get()`, `.update()`

Dictionary ke essential built-in methods:
- `dict.keys()`: Saari keys ka collection return karta hai.
- `dict.values()`: Saari values return karta hai.
- `dict.items()`: Saare `(key, value)` tuples return karta hai.
- `dict.get("key")`: Safe lookup! Agar key na mile to default value return karta hai.
- `dict.update(new_dict)`: Doosri dictionary ke pairs isme merge kar deta hai.


````python
# Operation 10: Dictionary Methods
marks_dict = {"Maths": 92, "Physics": 85, "Chemistry": 88}

print("Keys:  ", list(marks_dict.keys()))
print("Values:", list(marks_dict.values()))
print("Items: ", list(marks_dict.items()))

print("Safe get('Maths'):  ", marks_dict.get("Maths"))
print("Safe get('English'):", marks_dict.get("English", "Subject Not Registered"))

marks_dict.update({"English": 90, "Computer Science": 98})
print("Updated Marks Dict:", marks_dict)
````

### Operation 11: Sets in Python — Unique Elements, Duplicates Removal aur `set()` Constructor

Set curly braces `{}` ke andar **Unique Elements** ka ek **Unordered** collection hota hai.
- Set me **Duplicate Values allow nahi hoti**!
- **Automatic De-duplication:** Kisi bhi list se duplicates hatane ke liye use Set me convert kar do: `list(set(my_list))`.
- ⚠️ **Empty Set Syntax:** Khali Set banane ke liye `set()` likhna padta hai (`{}` se dictionary banti hai).


````python
# Operation 11: Sets and De-duplication
empty_dict = {}
empty_set = set()
print("Type of {}:    ", type(empty_dict))
print("Type of set(): ", type(empty_set))

raw_logins = ["user1", "user2", "user1", "user3", "user2", "user4", "user1"]
unique_users = set(raw_logins)

print("Raw Logins with duplicates:", raw_logins)
print("Unique Users in Set:        ", unique_users)
print(f"Original count: {len(raw_logins)} -> Unique count: {len(unique_users)}")
````

### Operation 12: Set Operations & Methods — `add()`, `remove()`, `union()`, `intersection()`, `difference()`

Classroom mathematics set operations:
- `set.add(x)`: Naya element jodna.
- `set.remove(x)`: Element hatana.
- **Union (`set1 | set2`):** Dono sets ke saare unique elements combine karna.
- **Intersection (`set1 & set2`):** Sirf common elements.
- **Difference (`set1 - set2`):** Wo elements jo set1 me hain par set2 me nahi hain.


````python
# Operation 12: Set Operations
dev_python = {"Soban", "Rahul", "Aman", "Priya"}
dev_java   = {"Aman", "Priya", "Vikram", "Sneha"}

print("Python Developers:", dev_python)
print("Java Developers:  ", dev_java)

all_devs = dev_python.union(dev_java)
common_devs = dev_python.intersection(dev_java)
only_python = dev_python.difference(dev_java)

print()
print("1. All Unique Developers (Union):", all_devs)
print("2. Multi-Skilled (Intersection): ", common_devs)
print("3. Pure Python Only (Difference):", only_python)
````

### Operation 13: Classroom Practice Problem (Part A & B) — Student Marks via List of Tuples

Day 04 ka classroom capstone problem (Part A & B):
Hume student records diye gaye hain as a list of tuples:
```python
records = [("Alice", 85), ("Bob", 92), ("Charlie", 78), ("David", 95), ("Eva", 88)]
```
Hume program likhna hai jo:
1. Saare students ke naam alag list me extract kare.
2. Sabse highest marks laane wale topper student ka naam aur score print kare!


````python
# Operation 13: Student Records using List of Tuples
records = [("Alice", 85), ("Bob", 92), ("Charlie", 78), ("David", 95), ("Eva", 88)]

student_names = [student[0] for student in records]
print("Student Names List:", student_names)

topper_name = ""
highest_marks = -1

for name, marks in records:
    if marks > highest_marks:
        highest_marks = marks
        topper_name = name

print(f"Class Topper: {topper_name} with {highest_marks} Marks!")
````

### Operation 14: Classroom Practice Problem (Part C) — Multi-Subject Gradebook using Nested Dictionaries

Day 04 ka final practice problem (Part C):
Students ka multi-subject gradebook structure:
```python
gradebook = {
    "Alice": {"Maths": 85, "Science": 90, "English": 78},
    "Bob": {"Maths": 92, "Science": 88, "English": 95},
    "Charlie": {"Maths": 78, "Science": 82, "English": 80}
}
```
Har student ka **Average Score** calculate karke class summary report print karna.


````python
# Operation 14: Multi-Subject Gradebook with Nested Dictionaries
gradebook = {
    "Alice":   {"Maths": 85, "Science": 90, "English": 78},
    "Bob":     {"Maths": 92, "Science": 88, "English": 95},
    "Charlie": {"Maths": 78, "Science": 82, "English": 80}
}

print("=== CLASS GRADEBOOK SUMMARY REPORT ===")
for student, subjects in gradebook.items():
    total = sum(subjects.values())
    count = len(subjects)
    avg_score = total / count
    print(f"Student: {student:10s} | Total: {total:3d}/300 | Average: {avg_score:.2f}%")
````
