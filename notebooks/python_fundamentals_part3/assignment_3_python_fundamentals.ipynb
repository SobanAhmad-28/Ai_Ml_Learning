### Q1: Palindrome String Checker

**Sawal:** User se string lekar check karo ki wo palindrome hai ya nahi (jaise `"madam"`, `"racecar"` aage aur peeche dono se same hoti hain).

**Logic (Hinglish):** String ko loop se ulat kar check karenge ki asli string se match hoti hai ya nahi.

```python
def is_palindrome(text):
    clean = text.strip().lower()
    reversed_str = ""
    for ch in clean:
        reversed_str = ch + reversed_str
    return clean == reversed_str

for word in ["madam", "racecar", "python", "level"]:
    print(f"'{word}' palindrome hai? -> {is_palindrome(word)}")
```

### Q2: List ke Numbers ka Average

**Sawal:** Ek list of integers lekar un sabhi numbers ka average calculate karo.

**Logic (Hinglish):** Sabhi elements ka sum lekar use `len(lst)` se divide karenge.

```python
def list_average(nums):
    return sum(nums) / len(nums) if nums else 0.0

sample = [12, 45, 78, 23, 56, 89]
print("Numbers:", sample)
print(f"Average: {list_average(sample):.2f}")
```

### Q4: Tuple me se Even aur Odd Tuples Alag Karna

**Sawal:** Ek integer tuple lekar do alag tuples banao: (1) Saare even numbers, (2) Saare odd numbers.

```python
orig_tuple = (10, 15, 22, 33, 40, 55, 60, 71)
evens = tuple(x for x in orig_tuple if x % 2 == 0)
odds = tuple(x for x in orig_tuple if x % 2 != 0)
print("Original:", orig_tuple)
print("Even Tuple:", evens)
print("Odd Tuple: ", odds)
```

### Q5: Student Marks Dictionary Menu-Based System

**Sawal:** Dictionary banao jisme `Key = Student Name`, `Value = Marks` hon. Menu banao:
- `A` - Add student
- `B` - Update marks
- `C` - Search student
- `D` - Display all

```python
students = {"Aman": 88, "Pooja": 92, "Karan": 76}

def manage_student(action, name=None, marks=None):
    act = action.upper()
    if act == 'A':
        students[name] = marks
        print(f"Add ho gaya: {name} ({marks} marks)")
    elif act == 'B':
        if name in students:
            students[name] = marks
            print(f"Marks update ho gaye: {name} -> {marks}")
    elif act == 'C':
        print(f"Search result for {name}: {students.get(name, 'Nahi mila')}")
    elif act == 'D':
        print("Sabhi students:", students)

manage_student('D')
manage_student('A', "Soban", 96)
manage_student('B', "Aman", 90)
manage_student('C', "Pooja")
```

### Q6: Words List se Unke Length ki Dictionary Banana

### Q8: Do Lists me Koi Common Element Nahi Hai (Disjoint Check)

**Sawal:** Check karo ki kya do lists ke beech koi bhi common element nahi hai (Sets use karo).

```python
def have_no_common(l1, l2):
    return set(l1).isdisjoint(set(l2))

p1, p2 = [1, 2, 3, 4], [5, 6, 7, 8]
q1, q2 = [1, 2, 3], [3, 4]
print(f"{p1} aur {p2} me koi common nahi hai? -> {have_no_common(p1, p2)}")
print(f"{q1} aur {q2} me koi common nahi hai? -> {have_no_common(q1, q2)}")
```

### Q9: List me se Duplicate (Ek se zyada baar aane wale) Elements

**Sawal:** List me jo elements ek se zyada baar aaye hain sirf unhe print karo (Sets use karo).

```python
def find_duplicates(items):
    seen = set()
    dups = set()
    for x in items:
        if x in seen:
            dups.add(x)
        else:
            seen.add(x)
    return sorted(list(dups))

data = [10, 20, 30, 20, 40, 50, 10, 60, 10]
print("Original list:", data)
print("Duplicate elements:", find_duplicates(data))
```
