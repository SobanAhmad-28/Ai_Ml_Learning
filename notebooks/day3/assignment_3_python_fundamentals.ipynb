### Q1: Palindrome String Checker

**Problem Statement:**
Ask the user for a string and check whether it is a palindrome or not.
A palindrome is a string which is the same when read forward and backward (e.g., `"madam"`, `"racecar"`).

**Logic:**
Reverse the string using a loop (or slicing `s[::-1]`) and check equality with original.

```python
# Q1 Solution
def is_palindrome(text: str) -> bool:
    clean_text = text.strip().lower()
    # Manual loop reversal
    reversed_text = ""
    for ch in clean_text:
        reversed_text = ch + reversed_text
    return clean_text == reversed_text

test_words = ["madam", "racecar", "python", "level"]
for word in test_words:
    print(f"Is '{word}' a palindrome? -> {is_palindrome(word)}")
```

### Q2: List Average Computation

**Problem Statement:**
Given a list of integers, compute the average of all numbers in the list.

**Logic:**
Accumulate sum using a loop (or `sum()`) and divide by `len(lst)`.

```python
# Q2 Solution
def compute_list_average(numbers: list) -> float:
    if not numbers:
        return 0.0
    total = sum(numbers)
    return total / len(numbers)

sample_nums = [12, 45, 78, 23, 56, 89]
print("Sample List:", sample_nums)
print(f"Average: {compute_list_average(sample_nums):.2f}")
```

### Q4: Partition Tuple into Even and Odd Tuples

**Problem Statement:**
Given a tuple of integers, create:
- A tuple of all even numbers
- A tuple of all odd numbers

```python
# Q4 Solution
def partition_tuple(tup: tuple):
    even_tup = tuple(x for x in tup if x % 2 == 0)
    odd_tup = tuple(x for x in tup if x % 2 != 0)
    return even_tup, odd_tup

numbers_tuple = (10, 15, 22, 33, 40, 55, 60, 71)
evens, odds = partition_tuple(numbers_tuple)

print("Original Tuple:", numbers_tuple)
print("Even Tuple:    ", evens)
print("Odd Tuple:     ", odds)
```

### Q5: Menu-Based Student Marks Dictionary System

**Problem Statement:**
Create a dictionary where `Keys = student names` and `Values = marks (integer)`.
Write a menu-based program with options:
- `A` - Add a student
- `B` - Update marks
- `C` - Search for a student
- `D` - Display all students and marks

```python
# Q5 Solution
class StudentRegister:
    def __init__(self):
        self.records = {"Aman": 88, "Pooja": 92, "Karan": 76}
        
    def execute(self, action: str, name: str = None, marks: int = None):
        action = action.upper()
        if action == 'A':  # Add student
            self.records[name] = marks
            return f"Added: {name} with marks {marks}"
        elif action == 'B':  # Update marks
            if name in self.records:
                self.records[name] = marks
                return f"Updated: {name}'s marks to {marks}"
            return f"Error: {name} not found in records"
        elif action == 'C':  # Search student
            if name in self.records:
                return f"Found: {name} -> {self.records[name]} marks"
            return f"Student {name} not found."
        elif action == 'D':  # Display all
            return dict(self.records)
        return "Invalid option."

register = StudentRegister()
print("Initial Records (D):", register.execute('D'))
print(register.execute('A', name="Shradha", marks=95))
print(register.execute('B', name="Aman", marks=91))
print(register.execute('C', name="Pooja"))
print("Final Records (D):  ", register.execute('D'))
```

### Q7: Count Spaces in a String

**Problem Statement:**
Write a program that takes a string from the user and prints the number of spaces in the string.

```python
# Q7 Solution
def count_spaces(text: str) -> int:
    space_count = 0
    for char in text:
        if char == ' ':
            space_count += 1
    return space_count

sample_sentence = "Artificial Intelligence and Machine Learning with Python"
print(f"Sentence: '{sample_sentence}'")
print("Number of spaces:", count_spaces(sample_sentence))
print("Verified via str.count(' '):", sample_sentence.count(' '))
```

### Q8: Check Disjoint Lists (No Common Elements)

**Problem Statement:**
Write a program to check whether two lists share no common elements.  
`# share no common elements: list1 = [1, 2, 3, 4], list2 = [5, 6, 7, 8]`  
`# share common elements: list1 = [1, 2, 3], list2 = [3, 4]`  

**Logic (Hint: Use Sets):**
Check `set(list1).isdisjoint(set(list2))` or whether `len(set(list1) & set(list2)) == 0`.

```python
# Q8 Solution
def check_no_common_elements(l1: list, l2: list) -> bool:
    set1 = set(l1)
    set2 = set(l2)
    # Returns True if intersection is empty
    return set1.isdisjoint(set2)

pair1_a, pair1_b = [1, 2, 3, 4], [5, 6, 7, 8]
pair2_a, pair2_b = [1, 2, 3], [3, 4]

print(f"{pair1_a} and {pair1_b} share NO common elements? -> {check_no_common_elements(pair1_a, pair1_b)}")
print(f"{pair2_a} and {pair2_b} share NO common elements? -> {check_no_common_elements(pair2_a, pair2_b)}")
```

### Q9: Find All Duplicate Elements in a List

**Problem Statement:**
Given a list, print all elements that appear more than once in the list.  

**Logic (Hint: Use Sets):**
Maintain a `seen` set and a `duplicates` set to detect items encountered more than once.

```python
# Q9 Solution
def find_duplicates(items: list) -> list:
    seen = set()
    duplicates = set()
    for item in items:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)
    return sorted(list(duplicates))

sample_data = [10, 20, 30, 20, 40, 50, 10, 60, 10]
print("Sample List:", sample_data)
print("Duplicate Elements (more than once):", find_duplicates(sample_data))
```

### Q10: Unique Characters & Count in a String

**Problem Statement:**
Ask the user for a string and print:
- All unique characters
- The count of unique characters

```python
# Q10 Solution
def extract_unique_chars(text: str):
    # Preserving first-seen order while checking uniqueness
    unique_set = set(text)
    ordered_uniques = [ch for ch in text if ch in unique_set and not (unique_set.remove(ch))]
    return ordered_uniques, len(ordered_uniques)

sample_input = "machine learning"
uniques, count = extract_unique_chars(sample_input)

print(f"Input String: '{sample_input}'")
print("All Unique Characters:", uniques)
print("Count of Unique Characters:", count)
```
