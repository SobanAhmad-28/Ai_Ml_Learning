## 1. Calculus Introduction — AI/ML me iska kya kaam hai?

Machine Learning model ko train karne ka poora process ek line me
summarize kiya ja sakta hai:

$$
\min_{w, b} \text{Loss}(w, b)
$$

Yaani hume aise weights ($w$) aur bias ($b$) dhoondne hain jisse model
ka prediction error (Loss) sabse kam (minimum) ho jaye!

### The Blindfolded Hiker Analogy (Gradient Descent):
Maan lo aapko aankho par patti baandh kar ek pahadi ke sabse nichle
hisse (valley) me utarna hai:
- Aap pair se zameen ka dhalan (slope / gradient) mehsoos karte ho.
- Agar dhalan aage neeche ja rahi hai, to aage kadam badhate ho.
- Agar dhalan peeche neeche ja rahi hai, to peeche kadam lete ho.
- Jab tak chalte ho jab tak zameen bilkul flat (slope = 0) na ho jaye!

Ye jo "dhalan" (slope) nikalne ka kaam hai, wahi **Calculus (Differentiation)**
karta hai!


## 2. Functions & 3. Composite Functions

Function ek machine jaisa hota hai: aap usme ek input $x$ daalte ho,
wo us par rule apply karta hai aur ek unique output $y = f(x)$ deta hai.

- **Domain:** Input values ka set jo valid hain (jaise $\sqrt{x}$ me $x \ge 0$).
- **Range:** Saare possible output values ka set.

### Composite Functions — Deep Learning ka Architecture:
Jab ek function ka output doosre function ka input ban jaye:

$$
(f \circ g)(x) = f(g(x))
$$

Deep Learning me Neural Network composite functions ki ek chain hoti hai:
- Layer 1: $h_1 = \sigma(W_1 x + b_1)$
- Layer 2: $h_2 = \sigma(W_2 h_1 + b_2)$
- Final Output: $\hat{y} = f_2(f_1(x))$

Chalo functions $y = \sqrt{x}$ aur $y = \sin(x)$ ke plots dekhte hain:


```python
# Visualizing Functions: y = sqrt(x) and y = sin(x)
import numpy as np
import matplotlib.pyplot as plt

# Using stacked subplots with compact width (7.5 inches) for GitHub safety
fig, axes = plt.subplots(2, 1, figsize=(7.5, 6.5), dpi=100)

# Plot 1: y = sqrt(x)
x1 = np.linspace(0, 100, 500)
y1 = np.sqrt(x1)
axes[0].plot(x1, y1, color='#e67e22', linewidth=2, label='y = √x')
axes[0].set_title('Graph of y = √x (Domain: x ≥ 0)', fontweight='bold')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].grid(True, linestyle=':', alpha=0.6)
axes[0].legend()

# Plot 2: y = sin(x)
x2 = np.linspace(-2 * np.pi, 2 * np.pi, 200)
y2 = np.sin(x2)
axes[1].plot(x2, y2, color='#2980b9', linewidth=2, label='y = sin(x)')
axes[1].axvline(0, color='black', linewidth=1, linestyle='--')
axes[1].axhline(0, color='black', linewidth=1, linestyle='--')
axes[1].set_title('Graph of y = sin(x) (Periodic Function)', fontweight='bold')
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].grid(True, linestyle=':', alpha=0.6)
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 4. Differentiation — Tangents & Instantaneous Rate of Change

Straight line me dhalan (slope) har jagah ek jaisi rehti hai ($m = \Delta y /
\Delta x$).
Lekin curve me har ek point par dhalan badalti rehti hai!

Kisi curve ke specific point par jo **Tangent Line** banti hai,
uske slope ko hi us point ka **Derivative** kehte hain:

$$
f'(x) = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}
$$

### Geometrical Meaning:
- **$f'(x) > 0$ (Positive slope):** Function us point par badh raha hai
(Increasing).
- **$f'(x) < 0$ (Negative slope):** Function us point par ghat raha hai
(Decreasing).
- **$f'(x) = 0$ (Flat tangent):** Ye ya to Peak (Maximum) hai ya Ghati
(Minimum)!


## 5. Differentiation Rules & The Chain Rule

Standard rules:
- **Power Rule:** $\frac{d}{dx}[x^n] = n x^{n-1}$
- **Constant Rule:** $\frac{d}{dx}[c] = 0$
- **Sum / Diff Rule:** $\frac{d}{dx}[f(x) \pm g(x)] = f'(x) \pm g'(x)$
- **Product Rule:** $\frac{d}{dx}[u \cdot v] = u v' + v u'$

---

### The Chain Rule — Backpropagation Engine:
Jab function ke andar function ho $y = f(g(x))$:

$$
\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}, \quad \text{where } u = g(x)
$$

### Example:
Differentiate $y = (3x^2 + 5)^4$:
1. Andar wale hisse ko $u$ maano: $u = 3x^2 + 5 \implies y = u^4$
2. Bahar wale ka derivative: $\frac{dy}{du} = 4u^3$
3. Andar wale ka derivative: $\frac{du}{dx} = 6x$
4. Multiply:

$$
\frac{dy}{dx} = 4(3x^2 + 5)^3 \cdot (6x) = 24x(3x^2 + 5)^3
$$

Neural networks me error backpropagation ke time yahi Chain Rule repeat hota
hai!


## 6. Finding Minima & Maxima (Gradient Descent Foundation)

Machine Learning me hamara aim hota hai **Loss Function ko MINIMIZE karna**.

### Step-by-Step Algorithm:
1. **First Derivative:** $f'(x)$ calculate karo.
2. **Critical Points:** Tangent ko flat karo: $f'(x) = 0$.
3. **Second Derivative Test ($f''(x)$):**
   - Agar $f''(x) > 0 \implies$ Curve opens upwards (Valley) $\to$ **Local
Minimum**!
   - Agar $f''(x) < 0 \implies$ Curve opens downwards (Peak) $\to$ **Local
Maximum**!

## 7. Solved Practice Problem: Loss Function Minimization

Maan lo ek model ka loss function hai:

$$
J(w) = w^2 - 4w + 6
$$

Hume wo weight $w$ dhoondna hai jaha loss minimum ho!

1. **First Derivative:** $J'(w) = 2w - 4$
2. **Slope ko 0 rakho:** $2w - 4 = 0 \implies w^* = 2$
3. **Second Derivative:** $J''(w) = 2 > 0$ (Confirmed Minimum!)
4. **Minimum Loss Value:** $J(2) = (2)^2 - 4(2) + 6 = 2$

Chalo Python me is curve ko aur optimal point ko plot karke dekhte hain:


```python
# Visualizing Loss Curve & Optimal Minimum Point
import numpy as np
import matplotlib.pyplot as plt

w = np.linspace(-1, 5, 200)
loss = w**2 - 4*w + 6

w_min = 2
loss_min = 2

w_tangent = np.linspace(0.5, 3.5, 50)
loss_tangent = np.ones_like(w_tangent) * loss_min

plt.figure(figsize=(7.5, 4.5), dpi=100)
plt.plot(w, loss, color='#8e44ad', linewidth=2.2, label='J(w) = w² - 4w + 6')
plt.plot(w_tangent, loss_tangent, color='red', linestyle='--', linewidth=1.8, label='Flat Tangent at Minimum')
plt.scatter([w_min], [loss_min], color='red', s=100, zorder=5, label=f'Optimal w* = {w_min}')

plt.title("Optimization: Finding Minimum Loss via Calculus", fontsize=12, fontweight='bold')
plt.xlabel("Weight Parameter (w)", fontsize=10)
plt.ylabel("Loss J(w)", fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
