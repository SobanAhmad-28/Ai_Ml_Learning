## 2. 📏 Straight Lines — The Backbone of Linear Regression

2D plane me kisi bhi seedhi line ki standard equation hoti hai:
$$y = mx + c$$

- **$m$ = Slope (Dhalan):** Line kitni tezi se upar ya neeche ja rahi hai.
  $$m = \tan(\theta) = \frac{y_2 - y_1}{x_2 - x_1} = \frac{\Delta y}{\Delta x}$$
- **$c$ = $y$-Intercept:** Line $y$-axis ko kis point par kaat-ti hai (jab $x = 0$ ho).

### 🤖 ML Connection (Linear Regression!):
Linear regression me hum yahi line dhoondte hain!
$$\hat{y} = w \cdot x + b$$
Yaha weight $w$ slope $m$ hai, aur bias $b$ intercept $c$ hai!

## 3. 📍 Distance between 2 Points — Euclidean & Manhattan Distance

Agar do points $P_1(x_1, y_1)$ aur $P_2(x_2, y_2)$ hain:

### A. Euclidean Distance ($L_2$ Distance):
Seedhi hawayi doori (Shortest straight-line distance, Pythagoras Theorem se):
$$d_{\text{Euclidean}} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$
$n$-dimensional space me:
$$d(\vec{x}, \vec{y}) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}$$

### B. Manhattan Distance ($L_1$ Distance / City Block Distance):
Agar aapko shehar ki grid wali sadko par chalna ho (diagonal shortcut nahi le sakte):
$$d_{\text{Manhattan}} = |x_2 - x_1| + |y_2 - y_1|$$

### 🤖 ML Connection:
- **KNN (K-Nearest Neighbors):** Naye data point ke paas wale $K$ points dhoondne ke liye Euclidean distance calculate hoti hai.
- **K-Means Clustering:** Data points ko cluster centroids ke kareeb laane ke liye distance matrix banti hai!

## 4. 📐 Parallel & Perpendicular Lines

- **Parallel Lines (Samantar Rekhayein):** Do lines jinki dhalan (slope) barabar ho aur wo aapas me kabhi na milein:
  $$m_1 = m_2$$
- **Perpendicular Lines (Lambarup Rekhayein):** Do lines jo $90^\circ$ ke angle par milti hain:
  $$m_1 \cdot m_2 = -1 \implies m_2 = -\frac{1}{m_1}$$

---

## 5. 🛡️ Distance from Point to Line & Parallel Lines (SVM Margin Intuition)

### Point $(x_0, y_0)$ se Line $Ax + By + C = 0$ ki Perpendicular Doori:
$$d = \frac{|Ax_0 + By_0 + C|}{\sqrt{A^2 + B^2}}$$

### Do Parallel Lines $Ax + By + C_1 = 0$ aur $Ax + By + C_2 = 0$ ke beech ki doori:
$$d = \frac{|C_1 - C_2|}{\sqrt{A^2 + B^2}}$$

### 🤖 ML Connection (Support Vector Machines - SVM):
SVM algorithm me hum ek aisi dividing line (hyperplane) dhoondte hain jo do classes (jaise Red dots vs Blue dots) ke beech **maximum margin (distance)** banaye. Ye margin exactly do parallel supporting lines ke beech ki perpendicular distance hoti hai!

Chalo lines aur distance ko visualize karke dekhte hain:


```python
# Visualizing Straight Line, Perpendicular Point & Distance
import numpy as np
import matplotlib.pyplot as plt

# Line: y = 2x + 1  => 2x - y + 1 = 0 (A=2, B=-1, C=1)
x = np.linspace(-2, 4, 100)
y = 2 * x + 1

# Point P(1, 7)
px, py = 1, 7

# Perpendicular distance formula: |2(1) - 1(7) + 1| / sqrt(2^2 + (-1)^2) = |-4| / sqrt(5) = 4 / 2.236 = 1.789
A, B, C = 2, -1, 1
dist = abs(A*px + B*py + C) / np.sqrt(A**2 + B**2)

# Foot of perpendicular point on line:
# Slope of line = 2, so slope of perpendicular = -0.5
# Equation of perp line passing through (1, 7): y - 7 = -0.5(x - 1) => y = -0.5x + 7.5
# Intersection with y = 2x + 1: 2x + 1 = -0.5x + 7.5 => 2.5x = 6.5 => x = 2.6, y = 6.2
foot_x, foot_y = 2.6, 6.2

plt.figure(figsize=(8, 5))
plt.plot(x, y, color='#2980b9', linewidth=2, label='Line: y = 2x + 1')
plt.scatter([px], [py], color='red', s=80, zorder=5, label=f'Point P({px}, {py})')
plt.scatter([foot_x], [foot_y], color='green', s=60, zorder=5, label=f'Foot on Line ({foot_x}, {foot_y})')
plt.plot([px, foot_x], [py, foot_y], color='red', linestyle='--', linewidth=1.5, label=f'Perp Distance = {dist:.2f}')

plt.title("Perpendicular Distance from Point to Line (SVM Margin Concept)", fontsize=12, fontweight='bold')
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()

print(f"Calculated Perpendicular Distance: {dist:.4f}")
```

## 7. 🏹 Vectors — Magnitude & Direction

Vector ek aisi mathematical entity hai jisme **Magnitude (Size/Length)** aur **Direction** dono hote hain.

$$\vec{v} = \begin{bmatrix} v_1 \\ v_2 \end{bmatrix} \quad \text{ya 3D me} \quad \vec{v} = \begin{bmatrix} v_1 \\ v_2 \\ v_3 \end{bmatrix}$$

### Magnitude of Vector (Length $\|\vec{v}\|$):
Euclidean norm ($L_2$ norm):
$$\|\vec{v}\| = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2}$$

### Unit Vector (Direction Vector $\hat{v}$):
Aisa vector jiska magnitude $1$ ho:
$$\hat{v} = \frac{\vec{v}}{\|\vec{v}\|}$$

## 8. ➕ Vector Addition & 9. ✖️ Scalar Multiplication

### A. Vector Addition:
Agar $\vec{a} = [a_1, a_2]$ aur $\vec{b} = [b_1, b_2]$ hain:
$$\vec{a} + \vec{b} = [a_1 + b_1, \ a_2 + b_2]$$
- **Geometric Meaning:** **Parallelogram Law** ya **Head-to-Tail Rule**. Pehle vector ke head par doosre vector ki tail rakhne se jo resultant vector banta hai.

### B. Scalar Multiplication ($c \cdot \vec{v}$):
Vector ke har component ko kisi number $c$ se multiply karna:
$$c \cdot [v_1, v_2] = [c \cdot v_1, \ c \cdot v_2]$$
- Agar $c > 1$: Vector lamba ho jata hai (stretching).
- Agar $0 < c < 1$: Vector chhota ho jata hai (shrinking).
- Agar $c < 0$: Vector ki direction ulati ($180^\circ$ opposite) ho jaati hai!

Chalo visual plot karke vector addition aur scalar multiplication dekhte hain:


```python
# Visualizing Vector Addition (Head-to-Tail) & Scalar Multiplication
import numpy as np
import matplotlib.pyplot as plt

v1 = np.array([3, 2])
v2 = np.array([1, 4])
v_sum = v1 + v2
v_scaled = 2 * v1

fig, axes = plt.subplots(2, 1, figsize=(7.5, 7))

# Plot 1: Vector Addition
ax1 = axes[0]
ax1.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue', label='v1 (3, 2)')
ax1.quiver(v1[0], v1[1], v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='green', label='v2 (1, 4) [Head-to-Tail]')
ax1.quiver(0, 0, v_sum[0], v_sum[1], angles='xy', scale_units='xy', scale=1, color='red', label='Sum v1 + v2 (4, 6)')
ax1.set_xlim(-1, 6)
ax1.set_ylim(-1, 8)
ax1.set_title("Vector Addition (Head-to-Tail)", fontweight='bold')
ax1.grid(True, linestyle=':')
ax1.legend()

# Plot 2: Scalar Multiplication
ax2 = axes[1]
ax2.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue', label='Original v1 (3, 2)')
ax2.quiver(0, 0, v_scaled[0], v_scaled[1], angles='xy', scale_units='xy', scale=1, color='orange', alpha=0.6, label='Scaled 2*v1 (6, 4)')
ax2.quiver(0, 0, -1*v1[0], -1*v1[1], angles='xy', scale_units='xy', scale=1, color='purple', label='Flipped -1*v1 (-3, -2)')
ax2.set_xlim(-4, 7)
ax2.set_ylim(-3, 6)
ax2.set_title("Scalar Multiplication", fontweight='bold')
ax2.grid(True, linestyle=':')
ax2.legend()

plt.tight_layout()
plt.show()
```

## 10. 🎯 Dot Product & Cosine Similarity — Modern AI ka Super-Hero!

Dot Product (Scalar Product) do vectors ko multiply karke ek single number (scalar) generate karta hai.

### Do Tarah Se Likha Ja Sakta Hai:
1. **Algebraic Formula (Component-wise sum):**
   $$\vec{a} \cdot \vec{b} = a_1 b_1 + a_2 b_2 + \dots + a_n b_n = \sum_{i=1}^n a_i b_i$$
2. **Geometric Formula (Projection & Angle):**
   $$\vec{a} \cdot \vec{b} = \|\vec{a}\| \|\vec{b}\| \cos(\theta)$$
   *(Yaha $\theta$ dono vectors ke beech ka angle hai).*

---

### 💡 Cosine Similarity:
Geometric formula ko rearrange karein to **Cosine Similarity** milti hai:
$$\cos(\theta) = \frac{\vec{a} \cdot \vec{b}}{\|\vec{a}\| \|\vec{b}\|}$$

- Agar $\theta = 0^\circ \implies \cos(0^\circ) = 1$ (Vectors bilkul same direction me hain $\to$ **100% Similar**).
- Agar $\theta = 90^\circ \implies \cos(90^\circ) = 0$ (Vectors perpendicular / orthogonal hain $\to$ **Koi relation nahi, 0% Similarity**).
- Agar $\theta = 180^\circ \implies \cos(180^\circ) = -1$ (Vectors ek doosre ke bilkul opposite hain $\to$ **Opposite meaning**).

### 🤖 AI / NLP / LLM Connection:
ChatGPT, Recommendation Systems (YouTube/Netflix), aur Vector Databases (Pinecone/Chroma) me jab do sentences ki meaning compare hoti hai, to unke word embeddings ka **Cosine Similarity** hi calculate kiya jata hai!

---

## 11. ⚔️ Cross Product — $\vec{a} \times \vec{b}$

Dot product se scalar milta hai, jabki **Cross Product se ek naya 3D Vector** milta hai jo dono original vectors ke bilkul **perpendicular ($90^\circ$)** hota hai:

$$\vec{a} \times \vec{b} = \begin{vmatrix} \hat{i} & \hat{j} & \hat{k} \\ a_1 & a_2 & a_3 \\ b_1 & b_2 & b_3 \end{vmatrix}$$
$$\|\vec{a} \times \vec{b}\| = \|\vec{a}\| \|\vec{b}\| \sin(\theta)$$

- Iski direction **Right-Hand Rule** se nikalti hai.
- Iska magnitude dono vectors se banne wale **Parallelogram ke Area** ke barabar hota hai.


```python
# Dot Product & Cosine Similarity in Python
import numpy as np

# Word embeddings example: King, Queen, Apple
vec_king  = np.array([0.9, 0.8, 0.1])
vec_queen = np.array([0.85, 0.82, 0.15])
vec_apple = np.array([0.1, 0.05, 0.95])

def cosine_similarity(a, b):
    dot = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    return dot / (norm_a * norm_b)

sim_king_queen = cosine_similarity(vec_king, vec_queen)
sim_king_apple = cosine_similarity(vec_king, vec_apple)

print(f"Cosine Similarity (King vs Queen): {sim_king_queen:.4f} (Bahut High Similarity!)")
print(f"Cosine Similarity (King vs Apple): {sim_king_apple:.4f} (Low Similarity, Alag Meaning!)")
```

## 12. 🧱 Matrices — Rows, Columns & Special Types

Matrix numbers ka ek 2D rectangular array hota hai jisme $m$ rows aur $n$ columns hote hain ($m \times n$).

### Important Types of Matrices:
1. **Square Matrix:** Jisme Rows aur Columns barabar hon ($m = n$).
2. **Identity Matrix ($I$):** Diagonal par $1$ aur baaki sab $0$. Kisi bhi matrix ko $I$ se multiply karne par koi farak nahi padta ($A \cdot I = A$).
3. **Diagonal Matrix:** Sirf main diagonal par numbers hote hain, baaki sab $0$.
4. **Transpose ($A^T$):** Rows ko Columns bana dena aur Columns ko Rows.
5. **Symmetric Matrix:** Jiska transpose original ke barabar ho: $A^T = A$.

---

## 13. ⚙️ Matrix Operations & Neural Network Connection

### A. Matrix Addition:
Dono matrices ka dimension bilkul same hona chahiye. Har element aamne-saamne add hota hai.

### B. Matrix Multiplication ($A \times B$):
**Golden Rule:** Matrix $A$ ke columns ki sankhya Matrix $B$ ki rows ke barabar honi chahiye:
$$(m \times k) \times (k \times n) \implies (m \times n)$$

Product ka element $(i, j)$ pehle matrix ki $i$-th row aur doosre matrix ke $j$-th column ka **Dot Product** hota hai!

### 🤖 Neural Network Feedforward Formula:
Deep Learning me ek layer ka output calculate karne ka formula hota hai:
$$Z = W \cdot X + b$$
Yaha $W$ Weights ki matrix hai, $X$ Inputs ka vector/matrix hai, aur $b$ Bias hai!


```python
# Matrix Multiplication in NumPy: Neural Network Layer Demo
import numpy as np

# Suppose 3 input features, and 2 neurons in the hidden layer
# Input X (batch of 2 samples, 3 features) -> shape (2, 3)
X = np.array([
    [1.5, 2.0, 0.5],
    [3.0, 1.0, 2.5]
])

# Weights matrix W (3 inputs -> 2 neurons) -> shape (3, 2)
W = np.array([
    [0.2, 0.8],
    [0.5, -0.3],
    [-0.1, 0.4]
])

# Bias for 2 neurons -> shape (2,)
b = np.array([0.1, -0.2])

# Matrix Multiplication: Z = X @ W + b
Z = np.dot(X, W) + b

print("Input X shape:", X.shape)
print("Weight W shape:", W.shape)
print("Output Z (X @ W + b) shape:", Z.shape)
print("\nComputed Layer Output Z:\n", Z)
```

## 14. 🔍 Determinants — Area Scaling Factor & Invertibility

Kisi $2 \times 2$ matrix ka determinant:
$$\det(A) = \begin{vmatrix} a & b \\ c & d \end{vmatrix} = ad - bc$$

### 💡 Geometric Meaning:
Determinant batata hai ki matrix ke transformation lagane par **area kitna guna scale hua**:
- Agar $\det(A) = 2$, to transformation unit square ke area ko double kar deta hai.
- Agar $\det(A) = 0$, to transformation 2D plane ko ek single line ya point me crush kar deta hai!
- **Singular Matrix:** Jiska $\det(A) = 0$ ho. Iska **Inverse $A^{-1}$ exist nahi karta**!

## 15. 🌟 Eigenvectors & Eigenvalues — PCA Dimensionality Reduction Core

Jab hum kisi vector par matrix transformation lagate hain, to zyadatar vectors ki length bhi badalti hai aur direction bhi ghoom jaati hai.

Lekin kuch **Special Vectors** aise hote hain jinki transformation lagane ke baad bhi **direction bilkul nahi badalti**, sirf unki length scale (chhoti ya badi) hoti hai!
Inhi special vectors ko hum **Eigenvectors** kehte hain, aur jitne guna length badli use **Eigenvalue ($\lambda$)** kehte hain!

### 📐 The Master Equation:
$$A \vec{v} = \lambda \vec{v}$$

Isko solve karne ke liye characteristic equation banti hai:
$$\det(A - \lambda I) = 0$$

### 🤖 Why are they critical in AI/ML?
1. **PCA (Principal Component Analysis):** Agar aapke paas 100 features hain aur aapko unhe 2 sabse important features me convert karna hai, to Covariance matrix ke Eigenvectors nikaale jaate hain. Jo Eigenvector sabse badi Eigenvalue wala hota hai, wahi data ki maximum variance (information) capture karta hai!
2. **Google PageRank:** Internet ke saare web pages ka ranking order Transition Probability matrix ka principal Eigenvector nikaal kar hi decide hota hai!

Chalo NumPy se Eigenvalues aur Eigenvectors nikaal kar verify karte hain:


```python
# Computing Eigenvalues and Eigenvectors in Python
import numpy as np

# A symmetric matrix (like a covariance matrix)
A = np.array([
    [4, 2],
    [2, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A:\n", A)
print("\nEigenvalues (lambda):", eigenvalues)
print("\nEigenvectors (columns):\n", eigenvectors)

# Verification of A * v = lambda * v for the first eigenvector
v1 = eigenvectors[:, 0]
lambda1 = eigenvalues[0]

Av1 = np.dot(A, v1)
lambda_v1 = lambda1 * v1

print("\n--- Verification for 1st Eigenvector ---")
print("A @ v1:      ", Av1)
print("lambda1 * v1:", lambda_v1)
print("Matches exactly?:", np.allclose(Av1, lambda_v1))
```
