## 2. 🎲 Understanding Probability — Basics & Terminology

Probability ka matlab simple shabdo me hai: **"Kisi event ke hone ke kitne chances hain?"**

### 🔑 Important Terms:
1. **Random Experiment**: Aisa experiment jiska result pehle se 100% confirm nahi bata sakte, lekin saare possible outcomes pata hote hain. (Jaise coin toss karna, ludo ki dice phekna).
2. **Sample Space ($S$)**: Saare possible outcomes ka set.
   - Coin toss: $S = \{H, T\}$ (Total outcomes = 2)
   - Ludo Dice roll: $S = \{1, 2, 3, 4, 5, 6\}$ (Total outcomes = 6)
   - 2 Coins toss: $S = \{HH, HT, TH, TT\}$ (Total outcomes = $2^2 = 4$)
3. **Event ($A$)**: Sample space ka koi specific subset jisme hume interest ho. (Jaise dice pe even number aana: $A = \{2, 4, 6\}$).

### 📐 Basic Formula:
$$P(A) = \frac{\text{Favorable Outcomes } n(A)}{\text{Total Outcomes } n(S)}$$

- Kisi bhi event ki probability hamesha **$0$ aur $1$ ke beech** hoti hai:
  $$0 \le P(A) \le 1$$
- $P(A) = 0 \implies$ **Impossible Event** (Jaise 6-sided dice pe 7 aana).
- $P(A) = 1 \implies$ **Sure/Certain Event** (Jaise dice pe 7 se chota number aana).

## 3. 🃏 Real-World Examples & 4. ✍️ Solved Practice Problems

### Problem 1: 2 Dice ko ek sath roll kiya gaya. Sum = 7 aane ki probability kya hai?
- **Total Outcomes ($S$):** $6 \times 6 = 36$
- **Favorable Outcomes ($A$):** Sum 7 banne wale pairs:
  $$(1,6), (2,5), (3,4), (4,3), (5,2), (6,1) \implies 6 \text{ pairs}$$
- **Probability:**
  $$P(\text{Sum = 7}) = \frac{6}{36} = \frac{1}{6} \approx 0.1667 \text{ (ya } 16.67\%)$$

### Problem 2: 3 Coins ko ek sath toss kiya gaya. "At least 2 Heads" aane ki probability?
- **Sample Space ($S$):** $\{HHH, HHT, HTH, HTT, THH, THT, TTH, TTT\} \implies 8$ outcomes.
- **Favorable Outcomes ($A$):** Kam se kam 2 Heads (2 ya 3 Heads):
  $$\{HHT, HTH, THH, HHH\} \implies 4 \text{ outcomes}$$
- **Probability:**
  $$P(\text{At least 2 Heads}) = \frac{4}{8} = \frac{1}{2} = 0.5$$

Chalo isko Python me verify karke dekhte hain!


```python
# Python Verification: Dice Roll Simulation for Sum = 7
import numpy as np

# 10,00,000 baar 2 dice roll karke simulate karte hain
num_simulations = 1_000_000
dice1 = np.random.randint(1, 7, num_simulations)
dice2 = np.random.randint(1, 7, num_simulations)
sums = dice1 + dice2

simulated_prob = np.mean(sums == 7)
theoretical_prob = 1 / 6

print(f"Simulated Probability (Sum = 7):   {simulated_prob:.4f}")
print(f"Theoretical Probability (6/36):      {theoretical_prob:.4f}")
```

## 5. 🔀 Types of Events

Probability ke rules samajhne ke liye events ke types samajhna sabse critical hai:

1. **Independent Events**: Jab ek event ke hone se doosre event ke hone par koi asar na pade.
   - *Example:* Coin toss me Head aana aur Ludo me 6 aana. Dono ka aapas me koi lena-dena nahi hai.
2. **Dependent Events**: Jab ek event doosre ke outcome ko effect kare.
   - *Example:* 52 cards ki gaddi se bina replace kiye ek card nikalna aur fir doosra nikalna.
3. **Mutually Exclusive Events (Disjoint Events)**: Jo do events **ek sath kabhi nahi ho sakte**.
   - *Example:* Ek single coin toss me ek sath Head aur Tail dono aana namumkin hai. ($A \cap B = \emptyset$).
4. **Exhaustive Events**: Jab sabhi events milkar poore sample space ko cover kar lein.
   - *Example:* Dice me Even number aana ya Odd number aana. Milkar $\{1,2,3,4,5,6\}$ ban jata hai.

---

## 6. 🔄 Complementary Rule

Kisi event ke **na hone ki probability**:
$$P(A') = 1 - P(A)$$
- *Use case:* Jab kisi question me pucha jaye **"At least one"**, to ulta socho:
  $$P(\text{At least 1}) = 1 - P(\text{None})$$
- *Example:* 3 coins toss kiye. Kam se kam ek Head aane ki probability?
  - $P(\text{No Heads / All Tails}) = P(TTT) = \frac{1}{8}$
  - $P(\text{At least 1 Head}) = 1 - \frac{1}{8} = \frac{7}{8} = 0.875$

---

## 7. ➕ Addition Rule ($OR$ Condition)

Jab hume pucha jaye ki Event $A$ ya Event $B$ me se koi ek ya dono ho jaye ($A \cup B$):

### Case A: General Formula (Non-Mutually Exclusive):
Agar dono me kuch common outcomes ho sakte hain:
$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$
*(Common hisse ko do baar add hone se bachane ke liye ek baar minus karte hain).*

### Case B: Mutually Exclusive Events:
Kyunki dono ek sath nahi ho sakte, $P(A \cap B) = 0$:
$$P(A \cup B) = P(A) + P(B)$$

---

## 8. ✖️ Multiplication Rule ($AND$ Condition)

Jab hume dono events ek ke baad ek sath chahiye ($A \cap B$):

### Case A: Independent Events:
$$P(A \cap B) = P(A) \times P(B)$$
- *Example:* Pehle coin pe Head AND dice pe 6 aane ki probability:
  $$P(H \cap 6) = \frac{1}{2} \times \frac{1}{6} = \frac{1}{12}$$

### Case B: Dependent Events:
$$P(A \cap B) = P(A) \times P(B|A)$$
*(Yaha $P(B|A)$ ka matlab hai: event $A$ hone ke baad $B$ hone ke chances).*


## 10. 🎯 Conditional Probability — $P(A|B)$

Conditional probability ka matlab hai: **"Agar event $B$ pehle hi ho chuka hai, to ab event $A$ ke hone ke kitne chances hain?"**

Iska formula hota hai:
$$P(A|B) = \frac{P(A \cap B)}{P(B)}, \quad \text{where } P(B) > 0$$

### 💡 Intuition (Sample Space Shrink Ho Gaya!):
Normal probability me hamara denominator poora Sample Space $S$ hota hai.
Lekin jab hume pehle hi pata chal gaya ki **$B$ ho chuka hai**, to ab humari duniya sirf $B$ ban gayi! Isliye denominator me $P(B)$ aata hai aur numerator me $B$ ke andar $A$ ka kitna hissa hai ($A \cap B$) wo aata hai.

### Real Life Example:
Maan lo ek classroom me 100 students hain:
- 60 Boys, 40 Girls.
- Total 30 students Python jante hain (20 Boys + 10 Girls).

Agar hum random ek student chunte hain aur hume pata chalta hai ki **wo Boy hai (Event $B$)**, to uske **Python janne (Event $A$)** ki probability kya hogi?
$$P(\text{Python} | \text{Boy}) = \frac{\text{Boys who know Python}}{\text{Total Boys}} = \frac{20}{60} = \frac{1}{3} \approx 33.3\%$$

---

## 11. 🌐 Law of Total Probability

Agar poore sample space ko mutually exclusive events $B_1, B_2, B_3, \dots, B_n$ me baant diya jaye (partition), to kisi bhi common event $A$ ki total probability in sabhi branches ka sum hoti hai:

$$P(A) = P(A|B_1)P(B_1) + P(A|B_2)P(B_2) + \dots + P(A|B_n)P(B_n)$$
$$P(A) = \sum_{i=1}^n P(A|B_i) P(B_i)$$

*Ye law Bayes' Theorem ka denominator banane ke kaam aata hai!*


## 12. 👑 Bayes' Theorem — The Foundation of Probabilistic AI

Bayes' Theorem machine learning ki jaan hai. **Naive Bayes classifier, Spam filters, Medical diagnosis systems, Recommendation engines** sab isi par chalte hain!

### 📐 Bayes' Theorem Formula:
$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$

Agar $B$ ke multiple causes $A_1, A_2, \dots, A_n$ hain:
$$P(A_i|B) = \frac{P(B|A_i) \cdot P(A_i)}{\sum_{j=1}^n P(B|A_j) \cdot P(A_j)}$$

### 🧩 Charo Terms ka Asli Matlab:
| Term | Naam | Matlab (Real Life Meaning) |
| :--- | :--- | :--- |
| **$P(A)$** | **Prior Probability** | Bina koi naya saboot (evidence) dekhe, purani information ke basis par $A$ ke chances. |
| **$P(B\|A)$** | **Likelihood** | Agar sach me $A$ sach hai, to saboot $B$ dikhne ke kitne chances hain? |
| **$P(B)$** | **Evidence / Marginal** | Saboot $B$ aane ki overall probability (har condition me). |
| **$P(A\|B)$** | **Posterior Probability** | Naya saboot $B$ milne ke baad $A$ hone ki updated probability! |

## 13. 🏥 Solved Practice Problem: Medical Test Paradox

Ye AI aur Data Science ka sabse famous interview problem hai:

> **Sawalo Ka Scenario:**
> 1. Kisi bimari ki prevalence population me sirf **$1\%$** hai: $P(D) = 0.01 \implies P(\text{No } D) = 0.99$.
> 2. Ek lab test **$95\%$ accurate** hai:
>    - Agar bimari hai to test Positive aane ke chances: $P(+\mid D) = 0.95$ (True Positive Rate).
>    - Agar bimari nahi hai to bhi test galat Positive (False Positive) bata sakta hai: $P(+\mid \text{No } D) = 0.05$.
> 3. **Question:** Ek person ka test **Positive** aaya. Kya use sach me bimari hone ke chances $95\%$ hain?

Chalo Bayes' Theorem lagate hain:
$$P(D \mid +) = \frac{P(+ \mid D) \cdot P(D)}{P(+)}$$

Pehle total $P(+)$ nikalte hain (Law of Total Probability):
$$P(+) = P(+ \mid D)P(D) + P(+ \mid \text{No } D)P(\text{No } D)$$
$$P(+) = (0.95 \times 0.01) + (0.05 \times 0.99)$$
$$P(+) = 0.0095 + 0.0495 = 0.059$$

Ab Posterior calculate karte hain:
$$P(D \mid +) = \frac{0.0095}{0.059} \approx 0.161 \text{ (yaani sirf } 16.1\%!)$$

**Chauk gaye?** Test 95% accurate hone ke bawajood agar aapka test positive aaya hai, to bhi sach me bimari hone ke chances sirf 16% hain! Kyunki bimari itni rare hai ki false positives ki bheed sach wale positives se zyada ho gayi.

Chalo Python se iska complete simulation karke dekhte hain:


```python
# Bayes' Theorem in Action: Medical Diagnosis Calculation
def bayes_disease(p_disease, sensitivity, false_positive_rate):
    p_no_disease = 1 - p_disease
    
    # Total probability of testing positive P(+)
    p_positive = (sensitivity * p_disease) + (false_positive_rate * p_no_disease)
    
    # Posterior probability P(Disease | +)
    p_disease_given_pos = (sensitivity * p_disease) / p_positive
    return p_disease_given_pos, p_positive

p_d = 0.01       # 1% prevalence
sens = 0.95      # 95% test sensitivity
fpr = 0.05       # 5% false alarm

posterior, evidence = bayes_disease(p_d, sens, fpr)

print(f"Total Probability of Positive Test P(+): {evidence:.4f} ({evidence*100:.2f}%)")
print(f"Actual Probability of having Disease P(D|+): {posterior:.4f} ({posterior*100:.2f}%)")
```

## 14. 🎲 Random Variables — Discrete vs Continuous

Random Variable ek aisa numerical variable hota hai jiska value kisi random experiment ke outcome par depend karta hai. Isko capital letter $X$ se denote karte hain.

1. **Discrete Random Variable**:
   - Jo countable values le sakta hai ($0, 1, 2, 3, \dots$).
   - *Examples:* Coin toss me number of heads, ek ghante me website pe aane wale visitors, kisi family me baccho ki sankhya.
2. **Continuous Random Variable**:
   - Jo kisi continuous range me koi bhi value le sakta hai (decimals bhi).
   - *Examples:* Kisi person ki height (e.g. 172.54 cm), car ki speed, stock price, kisi battery ki lifetime.

## 15. 📊 Measures of Central Tendency — Mean, Median & Mode

Data ke "center" ya "average" ko samajhne ke 3 tareeqe hote hain:

1. **Mean (Average)**:
   $$\bar{x} = \frac{\sum x_i}{N}$$
   - *Dhyan dene wali baat:* Mean **Outliers se bahut sensitive** hota hai! Agar ek aadmi ki salary 1 crore aa gayi to sabka average bigad jayega.
2. **Median (Middle Value)**:
   - Data ko sort karne ke baad jo theek beech me aata hai.
   - *Dhyan dene wali baat:* Median **Outliers se robust** hota hai. Real-world salary ya house price me hamesha median prefer kiya jata hai.
3. **Mode (Most Frequent Value)**:
   - Jo data point sabse zyada baar repeat hota hai. Categorical data me mode kaam aata hai.

## 16. 📏 Measures of Dispersion — Variance & Standard Deviation

Do datasets ka mean same ho sakta hai, lekin unka spread alag ho sakta hai:
- Class A marks: $[50, 50, 50, 50, 50] \implies \text{Mean} = 50$
- Class B marks: $[10, 30, 50, 70, 90] \implies \text{Mean} = 50$

Dono ka average 50 hai, par Class B me kitna zyada variation hai! Isi spread ko naapne ke liye hum Variance aur Standard Deviation nikalte hain.

### 1. Variance ($\sigma^2$):
Data points ka mean se doori ka squared average:
$$\sigma^2 = \frac{\sum (x_i - \mu)^2}{N}$$
*(Square isliye karte hain taaki positive aur negative differences cancel na ho jayein).*

### 2. Standard Deviation ($\sigma$):
Variance ki unit square me hoti hai (jaise $\text{marks}^2$), to original unit me laane ke liye square root lete hain:
$$\sigma = \sqrt{\sigma^2} = \sqrt{\frac{\sum (x_i - \mu)^2}{N}}$$

- **Chhota $\sigma$:** Data points mean ke paas bheed banaye hue hain (consistency high).
- **Bada $\sigma$:** Data points door-door phaile hue hain (high spread).

Chalo code me Mean, Median aur Outlier ka impact dekhte hain:


```python
# Python Demo: Outlier impact on Mean vs Median
import numpy as np

# Normal salaries in LPA
salaries = np.array([4, 5, 5.5, 6, 6.5, 7, 7.5, 8])
print("Original Salaries:", salaries)
print(f"Mean Salary:   {np.mean(salaries):.2f} LPA")
print(f"Median Salary: {np.median(salaries):.2f} LPA")
print(f"Std Dev:       {np.std(salaries):.2f} LPA\n")

# Suppose CEO enters with 100 LPA salary (Extreme Outlier)
salaries_with_ceo = np.append(salaries, 100)
print("With CEO Outlier:", salaries_with_ceo)
print(f"New Mean Salary:   {np.mean(salaries_with_ceo):.2f} LPA (Kharab ho gaya!)")
print(f"New Median Salary: {np.median(salaries_with_ceo):.2f} LPA (Stable raha!)")
print(f"New Std Dev:       {np.std(salaries_with_ceo):.2f} LPA")
```

## 17. 📈 Probability Distributions — PMF vs PDF vs CDF

Probability Distribution hume batata hai ki random variable $X$ ke alag-alag values aane ke kitne chances hain.

| Concept | Kis variable ke liye? | Matlab | Example |
| :--- | :--- | :--- | :--- |
| **PMF** (Probability Mass Function) | **Discrete** | Kisi exact point par probability: $P(X = k)$. | Dice pe exact 4 aane ki probability ($1/6$). |
| **PDF** (Probability Density Function) | **Continuous** | Kisi exact point par probability 0 hoti hai; hum range me area under the curve se probability nikalte hain: $P(a \le X \le b) = \int_a^b f(x)dx$. | Height 170cm se 175cm ke beech aane ke chances. |
| **CDF** (Cumulative Distribution Function) | **Dono** ke liye | Probability ki value $x$ se kam ya barabar ho: $P(X \le x)$. Ye hamesha 0 se start hoke 1 tak badhti hai. | Kisi student ke marks $\le 60$ aane ki probability. |

## 18. 🪙 Binomial Distribution

Binomial Distribution tab use hota hai jab koi experiment **$n$ independent trials** tak repeat kiya jaye aur har trial me sirf do hi possible results ho: **Success ($p$)** ya **Failure ($q = 1 - p$)**.

### 📐 Formula:
$$P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k} = \frac{n!}{k!(n - k)!} p^k (1 - p)^{n - k}$$

- $n$ = Total trials
- $k$ = Kitni successes chahiye ($0 \le k \le n$)
- $p$ = Ek trial me success ki probability

### 🎯 Key Properties:
- **Mean (Expected Value):** $E[X] = n \times p$
- **Variance:** $\text{Var}(X) = n \times p \times (1 - p)$
- **Standard Deviation:** $\sigma = \sqrt{n \cdot p \cdot (1 - p)}$

### Example:
Ek unbiased coin ko 5 baar toss kiya ($n = 5, p = 0.5$). Exactly 3 Heads aane ki probability?
$$P(X = 3) = \binom{5}{3} (0.5)^3 (0.5)^2 = 10 \times 0.125 \times 0.25 = \frac{10}{32} = \frac{5}{16} = 0.3125$$

Chalo `scipy.stats` se isko calculate aur plot karte hain!


```python
# Binomial Distribution with Scipy & Matplotlib
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 10     # 10 trials
p = 0.5    # 50% chance of success (e.g. fair coin)

# Possible number of successes (0 to 10)
k_values = np.arange(0, n + 1)
pmf_values = binom.pmf(k_values, n, p)

plt.figure(figsize=(9, 4.5))
bars = plt.bar(k_values, pmf_values, color='#4A90E2', edgecolor='black', alpha=0.8)
plt.title(f"Binomial Distribution PMF (n={n}, p={p})", fontsize=13, fontweight='bold')
plt.xlabel("Number of Successes (k)", fontsize=11)
plt.ylabel("Probability P(X = k)", fontsize=11)
plt.xticks(k_values)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Value annotations on bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.005, f"{yval:.3f}", ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()

# Verification for n=5, k=3, p=0.5
print("P(X=3 for n=5, p=0.5):", binom.pmf(3, 5, 0.5), "-> 5/16 =", 5/16)
```

## 19. 📏 Uniform Distribution

Uniform Distribution ka matlab hai ki diye gaye interval $[a, b]$ ke andar **har outcome ke aane ke chances bilkul barabar (equal) hote hain**.

### Continuous Uniform Distribution:
Agar $X \sim U(a, b)$:
$$f(x) = \begin{cases} \frac{1}{b - a} & \text{for } a \le x \le b \\ 0 & \text{otherwise} \end{cases}$$

- **Mean:** $E[X] = \frac{a + b}{2}$
- **Variance:** $\text{Var}(X) = \frac{(b - a)^2}{12}$

Iska graph ek bilkul flat rectangle jaisa dikhta hai kyunki density constant hoti hai!


```python
# Continuous Uniform Distribution Simulation
import numpy as np
import matplotlib.pyplot as plt

# 0 se 10 ke beech 500,000 random uniform points generate karte hain
a, b = 0, 10
uniform_data = np.random.uniform(a, b, 500_000)

plt.figure(figsize=(9, 4.5))
count, bins, ignored = plt.hist(uniform_data, bins=50, density=True, color='#2ECC71', edgecolor='black', alpha=0.7)
plt.plot(bins, np.ones_like(bins) * (1 / (b - a)), linewidth=2.5, color='darkred', label='Theoretical PDF: 1/(b-a) = 0.1')

plt.title("Continuous Uniform Distribution between 0 and 10", fontsize=13, fontweight='bold')
plt.xlabel("x", fontsize=11)
plt.ylabel("Probability Density", fontsize=11)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

## 20. 🔔 Normal (Gaussian) Distribution — The King of Distributions

Duniya ki zyadatar natural cheezein Normal Distribution follow karti hain — logo ki height, exam ke scores, blood pressure, measurement errors, stock market returns!

Isko **Bell Curve** bhi kehte hain kyunki ye ghanti ke aakaar ka hota hai.

### 📐 PDF Formula:
$$f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x - \mu}{\sigma}\right)^2}$$

- $\mu$ (Mean): Bell curve ka center (jaha peak hoti hai).
- $\sigma$ (Standard Deviation): Bell curve ki chaudai (spread).

---

### 🛡️ The Empirical Rule (68-95-99.7 Rule):
Ye rule har data scientist ki zubaan par hona chahiye:
1. **$68.27\%$ Data** mean ke $\pm 1\sigma$ ke daayre me rehta hai: $[\mu - \sigma, \mu + \sigma]$.
2. **$95.45\%$ Data** mean ke $\pm 2\sigma$ ke daayre me rehta hai: $[\mu - 2\sigma, \mu + 2\sigma]$.
3. **$99.73\%$ Data** mean ke $\pm 3\sigma$ ke daayre me rehta hai: $[\mu - 3\sigma, \mu + 3\sigma]$.
4. Jo data $\pm 3\sigma$ ke bhi bahar nikal jaye, use hum ML me **Outlier** maante hain!

---

### 📏 Standard Normal Distribution & $Z$-Score:
Kisi bhi normal distribution ko standard banane ke liye jiska Mean $\mu = 0$ aur $\sigma = 1$ ho:
$$Z = \frac{X - \mu}{\sigma}$$

- $Z$-score batata hai ki koi point mean se kitne standard deviation door hai!
- ML me **StandardScaler** isi formula par kaam karta hai!

Chalo ab complete 68-95-99.7 bell curve ka high quality visual plot banate hain:


```python
# Complete Normal Distribution & 68-95-99.7 Empirical Rule Plot
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters (e.g. Exam Scores)
mu = 70        # Mean Score
sigma = 10     # Standard Deviation

# X-axis values
x = np.linspace(30, 110, 1000)
y = norm.pdf(x, mu, sigma)

plt.figure(figsize=(10, 5.5))
plt.plot(x, y, color='#1B1464', linewidth=2.5, label='Normal Distribution Curve')

# 68% region (mu +- 1 sigma) -> [60, 80]
x1 = np.linspace(mu - sigma, mu + sigma, 500)
plt.fill_between(x1, norm.pdf(x1, mu, sigma), color='#2ecc71', alpha=0.45, label='68% Region (μ ± 1σ)')

# 95% region (mu +- 2 sigma) -> [50, 90]
x2_left = np.linspace(mu - 2*sigma, mu - sigma, 250)
x2_right = np.linspace(mu + sigma, mu + 2*sigma, 250)
plt.fill_between(x2_left, norm.pdf(x2_left, mu, sigma), color='#f1c40f', alpha=0.35, label='95% Region (μ ± 2σ)')
plt.fill_between(x2_right, norm.pdf(x2_right, mu, sigma), color='#f1c40f', alpha=0.35)

# 99.7% region (mu +- 3 sigma) -> [40, 100]
x3_left = np.linspace(mu - 3*sigma, mu - 2*sigma, 250)
x3_right = np.linspace(mu + 2*sigma, mu + 3*sigma, 250)
plt.fill_between(x3_left, norm.pdf(x3_left, mu, sigma), color='#e74c3c', alpha=0.25, label='99.7% Region (μ ± 3σ)')
plt.fill_between(x3_right, norm.pdf(x3_right, mu, sigma), color='#e74c3c', alpha=0.25)

# Center Mean Line
plt.axvline(mu, color='black', linestyle='--', linewidth=1.5, label=f'Mean (μ={mu})')

plt.title("Normal Distribution with Empirical Rule (68% - 95% - 99.7%)", fontsize=13, fontweight='bold')
plt.xlabel("Student Marks", fontsize=11)
plt.ylabel("Probability Density", fontsize=11)
plt.legend(loc='upper right', frameon=True)
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```
