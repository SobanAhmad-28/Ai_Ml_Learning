### Operation 1: Class & Object Basics
- **Class:** Object ka blueprint ya template (jaise ghar ka naksha).
- **Object:** Class ka actual runtime instance (jaise bana hua ghar).
- **Memory:** Class definition memory nahi leti; memory object banne par allocate hoti hai.


```python
# Operation 1: Class and Object Creation
class Student:
    name = "Anonymous"
    role = "Learner"

# Object Creation (Instantiation)
s1 = Student()
s1.name = "Soban"

s2 = Student()
s2.name = "Aman"

print(f"s1 Name: {s1.name} | Role: {s1.role}")
print(f"s2 Name: {s2.name} | Role: {s2.role}")
```

### Operation 2: Constructor (`__init__`) & `self`
- **`__init__()`:** Special dunder method jo object create hote hi automatically execute hota hai.
- **`self`:** Current object ka reference hota hai, jisse attributes us object se bind hote hain.


```python
# Operation 2: Constructor and self
class Car:
    def __init__(self, brand, model, price):
        self.brand = brand      # Instance attribute
        self.model = model
        self.price = price

    def display(self):
        return f"{self.brand} {self.model} - ₹{self.price:,}"

c1 = Car("Tata", "Nexon", 1200000)
c2 = Car("Mahindra", "XUV700", 2100000)

print(c1.display())
print(c2.display())
```

### Operation 3: Default vs Parameterized Constructor
- **Default Constructor:** Koi extra argument nahi leta; default values set karta hai.
- **Parameterized Constructor:** Object creation ke time arguments lekar custom attributes initialize karta hai.


```python
# Operation 3: Constructor Types
class UserProfile:
    # Parameterized with default fallback
    def __init__(self, username="Guest", is_admin=False):
        self.username = username
        self.is_admin = is_admin

# Default usage
u1 = UserProfile()
# Parameterized usage
u2 = UserProfile("admin_soban", is_admin=True)

print(f"User 1: {u1.username} | Admin: {u1.is_admin}")
print(f"User 2: {u2.username} | Admin: {u2.is_admin}")
```

### Operation 4: Class Attributes vs Instance Attributes
- **Class Attribute:** Poori class me shared hota hai; ek jagah badalne se sabhi objects par asar hota hai.
- **Instance Attribute:** Har object ka apna personal data hota hai (`self.var`).
- **Priority:** Agar name same ho, to Instance attribute Class attribute ko override karta hai.


```python
# Operation 4: Class vs Instance Attributes
class Employee:
    company_name = "Google"  # Class attribute (Shared)

    def __init__(self, emp_id, name):
        self.emp_id = emp_id  # Instance attribute (Unique)
        self.name = name

e1 = Employee(101, "Soban")
e2 = Employee(102, "Rahul")

print(f"{e1.name} works at {e1.company_name}")
print(f"{e2.name} works at {e2.company_name}")

# Changing class attribute updates for all
Employee.company_name = "DeepMind"
print(f"After Company Rebranding:")
print(f"  {e1.name} -> {e1.company_name}")
print(f"  {e2.name} -> {e2.company_name}")
```

### Operation 5: Methods ke 3 Types
- **1. Instance Method:** Normal method jo `self` lekar object ka data modify karta hai.
- **2. Class Method (`@classmethod`):** `cls` leta hai; class attributes ko modify karta hai.
- **3. Static Method (`@staticmethod`):** Na `self` leta hai na `cls`; independent utility function hota hai.


```python
# Operation 5: Three Types of Methods
class BankAccount:
    interest_rate = 4.0  # Class attribute in %

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    # 1. Instance Method
    def deposit(self, amount):
        self.balance += amount
        return f"Deposited ₹{amount:,}. New Balance: ₹{self.balance:,}"

    # 2. Class Method
    @classmethod
    def set_interest_rate(cls, new_rate):
        cls.interest_rate = new_rate

    # 3. Static Method
    @staticmethod
    def is_valid_pan(pan_number):
        return len(pan_number) == 10 and pan_number.isalnum()

acc = BankAccount("Soban", 50000)
print(acc.deposit(15000))

BankAccount.set_interest_rate(5.5)
print(f"Updated Bank Interest Rate: {BankAccount.interest_rate}%")

print("Is 'ABCDE1234F' valid PAN?:", BankAccount.is_valid_pan("ABCDE1234F"))
print("Is '123' valid PAN?:       ", BankAccount.is_valid_pan("123"))
```

### Operation 6: Classroom Practice Problem — Student Report Card
- Problem: `Student` class banana jo name aur 3 subjects ke marks store kare.
- Methods: Average calculate karna aur Pass/Fail status (`avg >= 40`) return karna.


```python
# Operation 6: Student Report Card System
class StudentRecord:
    def __init__(self, name, marks_list):
        self.name = name
        self.marks_list = marks_list

    def calculate_average(self):
        return sum(self.marks_list) / len(self.marks_list)

    def get_status(self):
        avg = self.calculate_average()
        return "PASS" if avg >= 40 else "FAIL"

s1 = StudentRecord("Soban", [88, 92, 95])
s2 = StudentRecord("Rohan", [35, 42, 38])

print(f"{s1.name:8s} | Average: {s1.calculate_average():.1f}% | Status: {s1.get_status()}")
print(f"{s2.name:8s} | Average: {s2.calculate_average():.1f}% | Status: {s2.get_status()}")
```

### Operation 7: OOP Pillar 1 — Abstraction
- **Definition:** Internal complex implementation ko chupana aur user ko sirf simple interface dikhana.
- **Example:** Car start karte waqt ignition/spark plug ka internal code user ko nahi dikhta, sirf `start()` button milta hai.


```python
# Operation 7: Abstraction Demo
class SmartPhone:
    def __init__(self):
        self._battery_level = 80
        self._is_connected_to_network = True

    # Internal hidden complex process
    def _connect_satellites(self):
        return "Internal GPS & Tower Handshake established."

    # Public simple interface
    def make_call(self, contact_name):
        if self._is_connected_to_network:
            connection = self._connect_satellites()
            return f"Calling {contact_name}... ({connection})"
        return "No network available!"

phone = SmartPhone()
print(phone.make_call("Aman"))
```

### Operation 8: OOP Pillar 2 — Encapsulation & Private Members
- **Definition:** Sensitive data aur methods ko ek unit me band karke unauthorized direct access se bachana.
- **Access Modifiers in Python:**
  - Public: `self.name` (Kahi se bhi access ho sakta hai).
  - Private: `self.__password` (Double underscore — class ke bahar direct access restricted).


```python
# Operation 8: Encapsulation & Private Attributes
class SecureAccount:
    def __init__(self, username, balance, pin):
        self.username = username
        self.balance = balance
        self.__pin = pin  # Private Attribute (Double underscore)

    def withdraw(self, amount, entered_pin):
        if entered_pin == self.__pin:
            if amount <= self.balance:
                self.balance -= amount
                return f"₹{amount:,} withdrawn successfully! Balance: ₹{self.balance:,}"
            return "Insufficient balance!"
        return "Access Denied: Incorrect PIN!"

user_acc = SecureAccount("soban_user", 25000, 1234)

# Correct withdrawal
print(user_acc.withdraw(5000, 1234))
# Wrong pin
print(user_acc.withdraw(5000, 9999))

# Attempting direct access to private attribute
try:
    print(user_acc.__pin)
except AttributeError as e:
    print(f"Encapsulation Protection: AttributeError -> '{e}'")
```

### Operation 9: OOP Pillar 3 — Inheritance
- **Definition:** Parent class ke attributes aur methods ko Child class me re-use karna (`class Child(Parent):`).
- **Types:**
  - Single Inheritance: Parent $	o$ Child
  - Multi-level Inheritance: Grandparent $	o$ Parent $	o$ Child
  - Multiple Inheritance: Parent1 + Parent2 $	o$ Child


```python
# Operation 9: Inheritance Types
# Base Class
class Device:
    def power_on(self):
        return "Device is powered on."

# Single Inheritance
class Laptop(Device):
    def run_code(self):
        return "Executing Python code on Laptop."

# Multi-level Inheritance
class GamingLaptop(Laptop):
    def enable_gpu(self):
        return "Dedicated GPU activated for High FPS gaming."

dev = GamingLaptop()
print(dev.power_on())   # Inherited from Device
print(dev.run_code())   # Inherited from Laptop
print(dev.enable_gpu()) # Defined in GamingLaptop
```

### Operation 10: `super()` Method (Parent Constructor Chaining)
- **`super()`:** Child class ke andar se Parent class ke methods aur `__init__` constructor ko invoke karta hai.
- Isse parent ke initialization code ko duplicate nahi karna padta.


```python
# Operation 10: super() Constructor Chaining
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Developer(Person):
    def __init__(self, name, age, primary_language, experience_years):
        super().__init__(name, age)  # Reusing Person initialization
        self.primary_language = primary_language
        self.experience_years = experience_years

    def profile(self):
        return f"{self.name} ({self.age} yrs) | Language: {self.primary_language} | Exp: {self.experience_years} yrs"

dev = Developer("Soban Ahmad", 22, "Python / AI", 2)
print(dev.profile())
```

### Operation 11: OOP Pillar 4 — Polymorphism & Operator Overloading
- **Polymorphism:** Ek hi method ya operator ka alag-alag objects ke sath alag tarike se behave karna.
- **Operator Overloading:** Built-in operators (`+`, `-`, `*`) ko custom objects ke liye dunder methods (`__add__`, `__sub__`) se custom behavior dena.


```python
# Operation 11: Polymorphism & Operator Overloading
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading '+' operator using __add__ dunder method
    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    # String representation
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector2D(3, 4)
v2 = Vector2D(5, 7)
v3 = v1 + v2  # Calls v1.__add__(v2) automatically!

print("v1:       ", v1)
print("v2:       ", v2)
print("v1 + v2:  ", v3)
```
