### Operation 1: Class aur Object Banana (Class & Object Basics)

**Kyu zaroorat hai:**
- `class`: Ek naya user-defined type define karti hai.
- `object`: Class ka live instance hota hai jo memory me data hold karta hai.
- Dot operator (`.`): Iski madad se hum object ke andar ke attributes aur methods ko access karte hain.

```python
# Class definition (Blueprint)
class Car:
    brand = "Toyota"  # default attribute

# Objects create karna (Instances)
car1 = Car()
car2 = Car()

# Dot operator (.) se access karna
print("Car 1 Brand:", car1.brand)
print("Car 2 Brand:", car2.brand)
print("car1 ka type:", type(car1))
```

### Operation 2: Constructor (`__init__`) aur `self` Parameter

**Kyu zaroorat hai:**
- **Constructor (`__init__`):** Ek special method jo object bante hi automatically call hota hai aur initial values set karta hai.
- **`self`:** Yeh us particular object ko refer karta hai jo method ko call kar raha hai. Isko manually pass karne ki zaroorat nahi hoti, Python khud handle karta hai.
- **Note:** Python me constructor overloading (ek se zyada `__init__`) directly allow nahi hota, jo last likha hota hai wahi execute hota hai.

```python
class Student:
    # Parameterized constructor
    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age
        print(f"[Constructor Called] Student '{self.name}' create ho gaya!")

# Alag-alag values ke sath objects create karna
stu1 = Student("Rahul", 20)
stu2 = Student("Harshita", 22)

print(f"Student 1: {stu1.name}, Age: {stu1.age}")
print(f"Student 2: {stu2.name}, Age: {stu2.age}")
```

### Operation 3: Class Attributes vs Instance Attributes

**Kyu zaroorat hai aur dono me kya farq hai:**
- **Class Attributes:** Class level par define hote hain aur sabhi objects ke beech share hote hain (e.g., college ka naam sabhi students ke liye same hoga).
- **Instance Attributes:** `__init__` ke andar `self` ke sath bante hain, aur har object ki apni personal unique copy hoti hai (e.g., student ka roll number aur GPA).

```python
class StudentRecord:
    college = "ABC Engineering College"  # Class Attribute (Sabka shared)

    def __init__(self, name, gpa):
        self.name = name  # Instance Attribute (Har student ka alag)
        self.gpa = gpa

s1 = StudentRecord("Aman", 8.9)
s2 = StudentRecord("Priya", 9.2)

print(f"{s1.name} | GPA: {s1.gpa} | College: {s1.college}")
print(f"{s2.name} | GPA: {s2.gpa} | College: {s2.college}")
# Class name se direct class attribute access karna
print("Direct via Class:", StudentRecord.college)
```

### Operation 4: Methods ke 3 Types (Instance, Class, Static Methods)

**Dono aur teeno me farq samjhein:**
1. **Instance Method:** Normal method jo pehla argument `self` leta hai aur object ke data ko modify/access karta hai.
2. **Class Method (`@classmethod`):** Pehla argument `cls` leta hai aur pure class level ke variables par kaam karta hai.
3. **Static Method (`@staticmethod`):** Na to `self` leta hai na `cls`. Yeh simple utility function hota hai jise logical grouping ke liye class ke andar rakha jata hai.

```python
class CollegeMember:
    university = "Tech University"  # Class variable

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # 1. Instance Method
    def display_info(self):
        print(f"Instance Method -> Name: {self.name}, Marks: {self.marks}")

    # 2. Class Method
    @classmethod
    def update_university(cls, new_name):
        cls.university = new_name
        print(f"Class Method -> University updated to: {cls.university}")

    # 3. Static Method
    @staticmethod
    def is_passing(marks):
        # Simple calculation, no self or cls needed
        return marks >= 40

member1 = CollegeMember("Soban", 85)
member1.display_info()

CollegeMember.update_university("AI Institute of Technology")
print("Passing Check (Static Method):", CollegeMember.is_passing(85))
```

### Operation 5: OOP Pillar 1 — Encapsulation & Access Modifiers

**Kyu zaroorat hai:**
Encapsulation ka matlab hai data (variables) aur functions ko ek bundle (class) me band karna aur unhe unauthorized changes se protect karna.

**Python me 3 Access Modifiers hote hain:**
1. **Public:** Normal variable (`self.name`) — kahin se bhi access ho sakta hai.
2. **Protected (`_var`):** Single underscore — internal use ya inheritance ke liye hint deta hai.
3. **Private (`__var`):** Double underscore — class ke bahar se direct access nahi hota (Python iski **Name Mangling** karke `_ClassName__var` bana deta hai).
4. **Getters & Setters:** Private variables ko safely read (getter) aur validate karke update (setter) karne ke methods.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner          # Public
        self._account_type = "Savings" # Protected (convention)
        self.__balance = balance    # Private (restricted access)

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter method with validation
    def set_balance(self, new_balance):
        if new_balance >= 0:
            self.__balance = new_balance
        else:
            print("Error: Balance negative nahi ho sakta!")

acc = BankAccount("Soban", 5000)
print("Public Owner:", acc.owner)
print("Protected Account Type:", acc._account_type)
print("Getter se Balance:", acc.get_balance())

# acc.__balance error dega, isliye setter use karenge
acc.set_balance(7500)
print("Updated Balance:", acc.get_balance())
```

### Operation 6: OOP Pillar 2 — Inheritance & `super()` Keyword

**Kyu zaroorat hai:**
Inheritance se ek Child class apne Parent class ke features aur methods ko re-use karti hai bina dobara code likhe.

**Types of Inheritance:**
1. **Single Inheritance:** Ek Parent $\rightarrow$ Ek Child.
2. **Multi-level Inheritance:** Grandparent $\rightarrow$ Parent $\rightarrow$ Child.
3. **Multiple Inheritance:** Do ya zyada Parents $\rightarrow$ Ek Child.
- **`super()`:** Parent class ke methods aur `__init__` constructor ko child class se call karne ke liye use hota hai.

```python
# 1. Base / Parent Class
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def get_details(self):
        return f"ID: {self.emp_id}, Name: {self.name}"

# 2. Derived / Child Class (Single Inheritance)
class DataScientist(Employee):
    def __init__(self, name, emp_id, specialty):
        # super() se parent ka constructor call kiya
        super().__init__(name, emp_id)
        self.specialty = specialty

    def info(self):
        return f"{self.get_details()} | Role: {self.specialty}"

ds = DataScientist("Soban", "AI-101", "Computer Vision & LLMs")
print("Inheritance Demo:", ds.info())
```

### Operation 7: Multiple Inheritance aur Multi-level Inheritance

**Kyu zaroorat hai:**
- **Multi-level:** Hierarchy me chalna (Employee $\rightarrow$ AdminStaff $\rightarrow$ Accountant).
- **Multiple:** Ek se zyada parent classes se qualities inherit karna (e.g. TA jo Teacher aur Student dono hota hai).

```python
# Multiple Inheritance demonstration
class Teacher:
    def __init__(self, salary):
        self.salary = salary

class Student:
    def __init__(self, gpa):
        self.gpa = gpa

# Child class inheriting from BOTH Teacher and Student
class TeachingAssistant(Teacher, Student):
    def __init__(self, name, salary, gpa):
        Teacher.__init__(self, salary)
        Student.__init__(self, gpa)
        self.name = name

ta = TeachingAssistant("Rahul", 45000, 8.8)
print(f"TA Name: {ta.name} | Salary: ₹{ta.salary} | GPA: {ta.gpa}")
```

### Operation 8: OOP Pillar 3 — Abstraction (Abstract Classes & Methods)

**Kyu zaroorat hai:**
Abstraction ka matlab hai zaroori cheezein dikhana aur andar ka complex implementation chhupana.
- **Abstract Class (`ABC`):** Is class ka direct object nahi ban sakta (`Cannot instantiate`). Yeh sirf child classes ke liye compulsory niyam/blueprint set karti hai.
- **`@abstractmethod`:** Jo child class is abstract class ko inherit karegi, use is method ko apna custom code dekar implement karna hi padega.

```python
from abc import ABC, abstractmethod

# Abstract Base Class
class MLModel(ABC):
    @abstractmethod
    def train(self, data):
        pass

    @abstractmethod
    def predict(self, sample):
        pass

# Concrete Subclass
class LinearClassifier(MLModel):
    def train(self, data):
        print(f"[Training] LinearClassifier model trained on {len(data)} samples.")

    def predict(self, sample):
        return f"Predicted class for {sample} -> Positive (1)"

# Instantiating concrete class
model = LinearClassifier()
model.train([1.2, 3.4, 5.6])
print(model.predict(2.5))
```

### Operation 9: OOP Pillar 4 — Polymorphism (Overriding & Duck Typing)

**Kyu zaroorat hai:**
Polymorphism ("Poly" = Many, "Morph" = Forms) ka matlab hai ek hi naam ka function alag-alag situations me alag behave kare:
1. **Operator Overloading:** Jaise `+` operator numbers ko add karta hai (`1 + 2 = 3`) aur strings ko concatenate karta hai (`'1' + '2' = '12'`).
2. **Method Overriding:** Child class apne parent class ke purane method ko naye tareeqe se override karti hai.
3. **Duck Typing:** *"Agar yeh duck jaisa dikhta hai aur duck jaisa quack karta hai, toh yeh duck hi hai."* Yaani Python type nahi dekhta, bas dekhta hai ki object ke paas wo method hai ya nahi.

```python
# 1. Method Overriding
class Animal:
    def sound(self):
        return "Generic animal sound"

class Dog(Animal):
    def sound(self):  # Overriding
        return "Bark! Woof!"

class Cat(Animal):
    def sound(self):  # Overriding
        return "Meow!"

# 2. Duck Typing
def play_sound(entity):
    # entity ka type matter nahi karta, bas uske paas .sound() hona chahiye
    print(f"{entity.__class__.__name__} says: {entity.sound()}")

animals = [Animal(), Dog(), Cat()]
for a in animals:
    play_sound(a)
```

### Operation 10: OOP Architecture Summary & Cheatsheet

| OOP Concept | Meaning in Simple Hinglish | Real-world / AI Example |
| :--- | :--- | :--- |
| **Class** | Blueprint / Naksha | PyTorch ka `nn.Module` |
| **Object** | Memory me bana hua actual instance | `model = ResNet50()` |
| **Constructor (`__init__`)** | Object bante hi initial values set karna | Hyperparameters initialize karna (`learning_rate=0.01`) |
| **Encapsulation** | Private variables (`__var`) + Getters/Setters | API keys aur weights ko hide rakhna |
| **Inheritance** | Parent se qualities lena (`super()`) | Custom dataset class inheriting from `torch.utils.data.Dataset` |
| **Abstraction** | Zaroori methods define karna (`ABC`) | `fit()` aur `predict()` pipeline standard banaye rakhna |
| **Polymorphism** | Ek naam, alag kaam (Method Overriding) | Har layer ka apna custom `forward()` pass |

```python
# Complete Mini Demo summarizing all concepts
class AIProject:
    platform = "Python Prime AI/ML"

    def __init__(self, title, domain):
        self.title = title
        self.domain = domain

    def get_summary(self):
        return f"Project: {self.title} (Domain: {self.domain}, Platform: {self.platform})"

proj = AIProject("Credit Card Fraud Detection", "Machine Learning")
print(proj.get_summary())
```
