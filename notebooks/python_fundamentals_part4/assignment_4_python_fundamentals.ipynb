### Q1: `BankAccount` Class (Deposit, Withdraw, Check Balance)

**Concept:** Classes & Objects  
**Problem Statement:**
Ek `BankAccount` class banayein jisme attributes hon: `account_number`, `owner_name`, aur `balance`.
Aur methods add karein: `deposit(amount)`, `withdraw(amount)`, aur `check_balance()`.

**Logic (Hinglish):**
- `deposit`: Amount add karega agar amount $> 0$ ho.
- `withdraw`: Amount subtract karega agar balance sufficient ho, warna error message dega.
- `check_balance`: Current balance display karega.

```python
class BankAccount:
    def __init__(self, account_number: str, owner_name: str, balance: float = 0.0):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance

    def deposit(self, amount: float):
        if amount > 0:
            self.balance += amount
            print(f"₹{amount:,.2f} deposit ho gaya. Naya balance: ₹{self.balance:,.2f}")
        else:
            print("Deposit amount positive hona chahiye!")

    def withdraw(self, amount: float):
        if amount <= 0:
            print("Withdrawal amount positive hona chahiye!")
        elif amount > self.balance:
            print(f"Insufficient Funds! Aapke account me sirf ₹{self.balance:,.2f} hain.")
        else:
            self.balance -= amount
            print(f"₹{amount:,.2f} nikal liya gaya. Bacha balance: ₹{self.balance:,.2f}")

    def check_balance(self):
        print(f"Account ({self.account_number}) - Owner: {self.owner_name} | Balance: ₹{self.balance:,.2f}")
        return self.balance

# Testing Q1
acc1 = BankAccount("SB-10024", "Soban Ahmad", 10000.0)
acc1.check_balance()
acc1.deposit(5000.0)
acc1.withdraw(3000.0)
acc1.withdraw(20000.0)  # Insufficient funds check
```

### Q2: `Book` Class with Review Management

**Concept:** Classes & Objects  
**Problem Statement:**
Ek `Book` class banayein jisme attributes hon: `title`, `author`, aur `list of reviews`.
Methods add karein:
1. `add_review(review_text)`: Naya review list me jode.
2. `count_reviews()`: Total reviews ki sankhya bataye.
3. `display_reviews()`: Sabhi reviews ko print kare.

**Logic (Hinglish):**
- `self.reviews = []` ek list rakhenge jisme naye strings append honge.

```python
class Book:
    def __init__(self, title: str, author: str):
        self.title = title
        self.author = author
        self.reviews = []  # List of reviews

    def add_review(self, review: str):
        self.reviews.append(review)
        print(f"Review add ho gaya: '{review}'")

    def count_reviews(self) -> int:
        return len(self.reviews)

    def display_reviews(self):
        print(f"\n--- Reviews for '{self.title}' by {self.author} (Total: {self.count_reviews()}) ---")
        if not self.reviews:
            print("Abhi tak koi review nahi hai.")
        for idx, r in enumerate(self.reviews, start=1):
            print(f"  {idx}. {r}")

# Testing Q2
my_book = Book("Hands-On Machine Learning", "Aurélien Géron")
my_book.add_review("Best practical guide for ML algorithms!")
my_book.add_review("Code examples are super clear.")
my_book.display_reviews()
```

### Q3: `Student` Class me Encapsulation & Validation

**Concept:** Encapsulation (Private Attributes + Getters & Setters)  
**Problem Statement:**
Ek `Student` class banayein jisme private attributes hon: `__name`, `__roll_no`, aur `__marks`.
Getter aur Setter methods provide karein strict validations ke sath:
- Marks negative nahi ho sakte ($marks \ge 0$).
- Roll number $1$ aur $100$ ke beech hona chahiye ($1 \le roll \le 100$).
- Name khali (empty string) nahi ho sakta.

```python
class Student:
    def __init__(self, name: str, roll_no: int, marks: float):
        # Direct setter se initialize karenge taaki validation check ho sake
        self.set_name(name)
        self.set_roll_no(roll_no)
        self.set_marks(marks)

    # Name Getter & Setter
    def get_name(self) -> str:
        return self.__name

    def set_name(self, name: str):
        if name and name.strip():
            self.__name = name.strip()
        else:
            raise ValueError("Validation Error: Name khali (empty) nahi ho sakta!")

    # Roll Number Getter & Setter
    def get_roll_no(self) -> int:
        return self.__roll_no

    def set_roll_no(self, roll_no: int):
        if 1 <= roll_no <= 100:
            self.__roll_no = roll_no
        else:
            raise ValueError("Validation Error: Roll number 1 se 100 ke beech hona chahiye!")

    # Marks Getter & Setter
    def get_marks(self) -> float:
        return self.__marks

    def set_marks(self, marks: float):
        if marks >= 0:
            self.__marks = marks
        else:
            raise ValueError("Validation Error: Marks negative nahi ho sakte!")

# Testing Q3
s = Student("Soban", 42, 94.5)
print(f"Valid Student -> Name: {s.get_name()}, Roll: {s.get_roll_no()}, Marks: {s.get_marks()}")

# Validation check test
try:
    s.set_marks(-15)  # Invalid
except ValueError as e:
    print("Caught Expected Exception:", e)
```

### Q4: `Shape` Hierarchy & Function Overriding (`area()`)

**Concept:** Function / Method Overriding (Runtime Polymorphism)  
**Problem Statement:**
Ek base class `Shape` banayein jisme `area()` method ho.  
Teen subclasses banayein: `Circle`, `Rectangle`, aur `Triangle` jo `area()` method ko override karein.

**Formulas:**
- Circle: $\pi \times r^2$ (jahan $\pi = 3.14159$)
- Rectangle: $width \times height$
- Triangle: $0.5 \times base \times height$

```python
class Shape:
    def area(self):
        return 0.0

class Circle(Shape):
    def __init__(self, radius: float):
        self.radius = radius

    def area(self) -> float:
        return 3.14159 * (self.radius ** 2)

class Rectangle(Shape):
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height

class Triangle(Shape):
    def __init__(self, base: float, height: float):
        self.base = base
        self.height = height

    def area(self) -> float:
        return 0.5 * self.base * self.height

# Testing Q4 Polymorphic execution
shapes = [Circle(radius=5), Rectangle(width=4, height=6), Triangle(base=6, height=8)]
for sh in shapes:
    print(f"{sh.__class__.__name__} Area: {sh.area():.2f}")
```

### Q5: `Vehicle` Base Class & Inheritance (`Car`, `Bike`)

**Concept:** Single / Hierarchical Inheritance  
**Problem Statement:**
Base class `Vehicle` banayein jisme `brand` aur `model` attributes hon.  
Do subclasses banayein:
- `Car`: jo extra attribute `seats` add kare.
- `Bike`: jo extra attribute `engine_cc` add kare.

**Logic:** `super().__init__(brand, model)` se parent attributes set honge.

```python
class Vehicle:
    def __init__(self, brand: str, model: str):
        self.brand = brand
        self.model = model

    def get_info(self):
        return f"{self.brand} {self.model}"

class Car(Vehicle):
    def __init__(self, brand: str, model: str, seats: int):
        super().__init__(brand, model)
        self.seats = seats

    def display(self):
        print(f"Car: {self.get_info()} | Seating Capacity: {self.seats} seats")

class Bike(Vehicle):
    def __init__(self, brand: str, model: str, engine_cc: int):
        super().__init__(brand, model)
        self.engine_cc = engine_cc

    def display(self):
        print(f"Bike: {self.get_info()} | Engine: {self.engine_cc}cc")

# Testing Q5
car = Car("Hyundai", "Verna", seats=5)
bike = Bike("Royal Enfield", "Hunter 350", engine_cc=349)
car.display()
bike.display()
```

### Q6: Abstract `Employee` Class & Salary Computation

**Concept:** Abstraction (`ABC`, `@abstractmethod`)  
**Problem Statement:**
Ek abstract class `Employee` banayein jisme abstract method `calculate_salary()` ho.  
Teen subclasses banayein jo is method ko apne mutabiq implement karein:
1. `Intern`: Fixed stipend salary.
2. `FullTimeEmployee`: Base salary + performance bonus.
3. `ContractEmployee`: Hourly rate $\times$ hours worked.

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, name: str):
        self.name = name

    @abstractmethod
    def calculate_salary(self) -> float:
        pass

class Intern(Employee):
    def __init__(self, name: str, stipend: float):
        super().__init__(name)
        self.stipend = stipend

    def calculate_salary(self) -> float:
        return self.stipend

class FullTimeEmployee(Employee):
    def __init__(self, name: str, base_salary: float, bonus: float):
        super().__init__(name)
        self.base_salary = base_salary
        self.bonus = bonus

    def calculate_salary(self) -> float:
        return self.base_salary + self.bonus

class ContractEmployee(Employee):
    def __init__(self, name: str, hourly_rate: float, hours_worked: int):
        super().__init__(name)
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_salary(self) -> float:
        return self.hourly_rate * self.hours_worked

# Testing Q6
staff = [
    Intern("Aman", stipend=25000),
    FullTimeEmployee("Soban", base_salary=80000, bonus=15000),
    ContractEmployee("David", hourly_rate=1200, hours_worked=50)
]
for emp in staff:
    print(f"{emp.__class__.__name__} ({emp.name}) Salary: ₹{emp.calculate_salary():,.2f}")
```

### Q7: Constructor Overloading Simulation (Default Parameters)

**Concept:** Constructor Overloading Simulation  
**Problem Statement:**
Ek `Person` class banayein jiska constructor 3 tareeqon se kaam kare:
1. Sirf `name` ke sath
2. `name + age` ke sath
3. `name + age + address` ke sath

**Logic (Hinglish):**
Python me multiple `__init__` allow nahi hote, isliye hum default parameters (`age=None, address=None`) use karte hain.

```python
class Person:
    def __init__(self, name: str, age: int = None, address: str = None):
        self.name = name
        self.age = age
        self.address = address

    def display(self):
        age_str = f"{self.age} years" if self.age is not None else "Not specified"
        addr_str = self.address if self.address is not None else "Not specified"
        print(f"Person -> Name: {self.name} | Age: {age_str} | Address: {addr_str}")

# Testing Q7 teeno modes me
p1 = Person("Aman")
p2 = Person("Pooja", age=24)
p3 = Person("Soban", age=22, address="Mumbai, India")

p1.display()
p2.display()
p3.display()
```

### Q8: Class vs Instance Attributes (`player_count` Tracker)

**Concept:** Class Attribute vs Instance Attribute  
**Problem Statement:**
Ek `Player` class banayein jisme:
- Class variable: `player_count` (jo count kare kitne total players bane)
- Instance variables: `name` aur `level`

**Logic (Hinglish):**
Har baar jab `__init__` chalega, hum `Player.player_count += 1` karenge.

```python
class Player:
    player_count = 0  # Class variable shared across all instances

    def __init__(self, name: str, level: int = 1):
        self.name = name      # Instance variable
        self.level = level    # Instance variable
        Player.player_count += 1  # Increment global count

    def get_info(self):
        return f"Player '{self.name}' (Level {self.level})"

# Testing Q8
print("Initial Players Count:", Player.player_count)
pl1 = Player("Alpha", level=5)
pl2 = Player("Ghost", level=12)
pl3 = Player("Shadow", level=8)

print(pl1.get_info())
print(pl2.get_info())
print(pl3.get_info())
print("Final Active Players Count:", Player.player_count)
```

### Q9: Multiple Inheritance Showcase (`Herbivore`, `Carnivore`, `Bear`)

**Concept:** Multiple Inheritance  
**Problem Statement:**
Classes banayein: `Herbivore`, `Carnivore`, aur `Omnivore` attributes/methods ke sath.  
Fir ek class `Bear` banayein jo in sabhi classes se inherit kare taaki multiple inheritance ka mechanism showcase ho sake.

```python
class Herbivore:
    def eat_plants(self):
        return "Grass, berries, aur plants khata hai."

class Carnivore:
    def eat_meat(self):
        return "Fish, insect, aur meat khata hai."

class Omnivore(Herbivore, Carnivore):
    def diet_type(self):
        return "Omnivore (Plants aur Meat dono khane wala)"

# Bear inherits from Omnivore (inheriting both plant and meat capabilities)
class Bear(Omnivore):
    def __init__(self, species: str):
        self.species = species

    def describe_diet(self):
        print(f"--- {self.species} Diet Behavior ---")
        print("Diet Category: ", self.diet_type())
        print("Plant intake:  ", self.eat_plants())
        print("Meat intake:   ", self.eat_meat())

# Testing Q9
grizzly = Bear("Grizzly Bear")
grizzly.describe_diet()
```

### Q10: Mini Project — OOP Chat System

**Concept:** Real-world Multi-Class OOP Architecture  
**Problem Statement:**
OOPs concepts use karke ek Chat System banayein jisme teen classes hon:
1. `User`: User ka name, join status.
2. `Message`: Sender name, message text, timestamp.
3. `ChatRoom`: Chat history, users tracking, message broadcast.

**Functions implement karne hain:**
- Sending messages
- Viewing chat history
- User joining and leaving chatroom

```python
from datetime import datetime

class Message:
    def __init__(self, sender_name: str, content: str):
        self.sender_name = sender_name
        self.content = content
        self.timestamp = datetime.now().strftime("%H:%M:%S")

    def format_message(self) -> str:
        return f"[{self.timestamp}] {self.sender_name}: {self.content}"

class User:
    def __init__(self, username: str):
        self.username = username
        self.current_room = None

    def send_message(self, text: str):
        if self.current_room:
            self.current_room.broadcast(self, text)
        else:
            print(f"{self.username} kisi chatroom me connected nahi hai!")

class ChatRoom:
    def __init__(self, room_name: str):
        self.room_name = room_name
        self.members = {}  # username -> User object
        self.history = []  # List of Message objects

    def join(self, user: User):
        self.members[user.username] = user
        user.current_room = self
        system_msg = Message("SYSTEM", f"{user.username} ne chatroom '{self.room_name}' join kiya!")
        self.history.append(system_msg)
        print(system_msg.format_message())

    def leave(self, user: User):
        if user.username in self.members:
            del self.members[user.username]
            user.current_room = None
            system_msg = Message("SYSTEM", f"{user.username} ne chatroom chhod diya.")
            self.history.append(system_msg)
            print(system_msg.format_message())

    def broadcast(self, sender: User, text: str):
        msg = Message(sender.username, text)
        self.history.append(msg)
        print(msg.format_message())

    def view_history(self):
        print(f"\n=== Chat History: #{self.room_name} (Active Members: {len(self.members)}) ===")
        for msg in self.history:
            print(" ", msg.format_message())

# Testing Q10 (Mini Project Simulation)
room = ChatRoom("AI-ML-Developers")
u1 = User("Soban")
u2 = User("Aman")
u3 = User("Priya")

room.join(u1)
room.join(u2)

u1.send_message("Hello Aman! Kya haal hain?")
u2.send_message("All good Soban! Day 4 OOPs assignment solve kar raha hoon.")

room.join(u3)
u3.send_message("Hey everyone! Great chat system.")

room.leave(u2)
room.view_history()
```
