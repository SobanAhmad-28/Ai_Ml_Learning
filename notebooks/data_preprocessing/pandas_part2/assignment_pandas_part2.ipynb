## 🌸 Question 1: IRIS Flower Dataset Exploration & GroupBy Analysis

### 📌 Problem Statement:
Iris Dataset ko load karo aur niche diye gaye 4 tasks perform karo:
- **Part A:** First 10 rows, Shape, Data types (`.info()`), aur Summary statistics (`mean, std, min, max`) display karo.
- **Part B:** Sirf wahi rows select karo jaha:
  - `petal_length > 4.5` AND
  - `species == "Iris-virginica"`
- **Part C:** `species` ke basis par **Group by** karo aur compute karo:
  - Average `sepal_length`
  - Maximum `petal_width`
  - Standard deviation of `sepal_width`
- **Part D:** Ek naya column banao `"petal_ratio" = petal_length / petal_width`, aur har species ke liye average `petal_ratio` find out karo.


```python
import pandas as pd
import numpy as np

# Load Iris Dataset (Seaborn repository se directly or fallback DataFrame)
iris_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
try:
    df_iris = pd.read_csv(iris_url)
    print("Iris dataset loaded successfully from online source!")
except Exception:
    # Offline fallback dataset simulation
    from sklearn.datasets import load_iris
    raw_iris = load_iris()
    df_iris = pd.DataFrame(raw_iris.data, columns=["sepal_length", "sepal_width", "petal_length", "petal_width"])
    species_map = {0: "setosa", 1: "versicolor", 2: "virginica"}
    df_iris["species"] = [species_map[i] for i in raw_iris.target]
    print("Iris dataset loaded from local sklearn fallback!")

# Standardize species column names to match assignment (e.g. Iris-virginica or virginica)
df_iris["species_clean"] = df_iris["species"].apply(lambda s: s if s.startswith("Iris-") else "Iris-" + s)

print("Iris Table Shape:", df_iris.shape)
```

### 🔹 Q1 Part A: First 10 Rows, Shape, Info & Summary Statistics

#### 💡 Hinglish Explanation:
- `.head(10)` se pehli 10 rows dekhenge.
- `.shape` se dimensions (150 rows, 5 columns) verify karenge.
- `.describe()` se numerical columns ka mean, std, min, aur max nikal aayega.


```python
# 1. First 10 rows
print("--- First 10 Rows of Iris Dataset ---\n", df_iris.head(10))

# 2. Shape & Data Types
print(f"\nDataset Shape: {df_iris.shape[0]} Rows x {df_iris.shape[1]} Columns")
print("\n--- Column Data Types (.info()) ---")
df_iris.info()

# 3. Summary Statistics (mean, std, min, max)
stats = df_iris.describe().loc[["mean", "std", "min", "max"]]
print("\n--- Key Summary Statistics (Mean, Std, Min, Max) ---\n", np.round(stats, 2))
```

### 🔹 Q1 Part B: Row Selection (`petal_length > 4.5` & `species == 'Iris-virginica'`)

#### 💡 Hinglish Explanation:
- Do conditions ko compound filter karne ke liye `&` operator use hota hai aur har condition ko parentheses `()` me rakhte hain.


```python
# Filter rows
filtered_iris = df_iris[(df_iris["petal_length"] > 4.5) & (df_iris["species_clean"] == "Iris-virginica")]

print(f"Total matching Virginica flowers: {len(filtered_iris)}")
print("\n--- Filtered Rows Preview (Top 5) ---\n", filtered_iris.head(5))
```

### 🔹 Q1 Part C & D: Species GroupBy Aggregations & `petal_ratio` Feature Engineering

#### 💡 Hinglish Explanation:
- **Part C:** `df.groupby("species")` par `.agg()` use karke custom column metrics calculate karenge.
- **Part D:** Vectorized division `df["petal_length"] / df["petal_width"]` se `petal_ratio` banega, fir har species ka average ratio calculate karenge.


```python
# Part C: Group by species and compute custom metrics
species_summary = df_iris.groupby("species_clean").agg(
    avg_sepal_length=("sepal_length", "mean"),
    max_petal_width=("petal_width", "max"),
    std_sepal_width=("sepal_width", "std")
).round(2)

print("--- Q1 Part C: Species GroupBy Summary Table ---\n", species_summary)

# Part D: Create new column 'petal_ratio' = petal_length / petal_width
df_iris["petal_ratio"] = df_iris["petal_length"] / df_iris["petal_width"]

# Average petal_ratio per species
avg_ratios = df_iris.groupby("species_clean")["petal_ratio"].mean().round(3)
print("\n--- Q1 Part D: Average Petal Ratio per Species ---\n", avg_ratios)
```

## 🚢 Question 2: TITANIC Dataset Analysis & Survival Insights

### 📌 Problem Statement:
Titanic Passengers Dataset load karo aur niche diye gaye analysis questions solve karo:
- **Part A:** Sirf specific columns display karo: `Name, Sex, Age, Fare, Survived`.
- **Part B:** Wahi passengers select karo jo:
  - Female hon (`Sex == 'female'`) AND
  - Jinka ticket Fare $> 30$ ho (`Fare > 30`).
- **Part C:** Passenger Class (`Pclass`) ke basis par **Group by** karo aur compute karo:
  - **Survival rate** (`Survived.mean()`)
  - **Average Fare**
  - **Average Age**


```python
# Load Titanic Dataset
titanic_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
try:
    df_titanic = pd.read_csv(titanic_url)
    print("Titanic dataset loaded successfully from online source!")
except Exception:
    # Offline simulation dataset with exact Titanic columns
    np.random.seed(42)
    df_titanic = pd.DataFrame({
        "PassengerId": range(1, 101),
        "Survived": np.random.choice([0, 1], size=100, p=[0.6, 0.4]),
        "Pclass": np.random.choice([1, 2, 3], size=100, p=[0.25, 0.25, 0.50]),
        "Name": [f"Passenger_{i}" for i in range(1, 101)],
        "Sex": np.random.choice(["female", "male"], size=100, p=[0.35, 0.65]),
        "Age": np.random.randint(1, 75, size=100),
        "Fare": np.random.uniform(7.5, 120.0, size=100)
    })
    print("Titanic dataset loaded from offline simulation fallback!")

print("Titanic Dataset Dimensions:", df_titanic.shape)
```

### 🔹 Q2 Part A & B: Column Selection & High-Fare Female Passenger Filtering

#### 💡 Hinglish Explanation:
- **Part A:** List of column names pass karke specific 5 columns display karenge.
- **Part B:** `(df["Sex"] == "female") & (df["Fare"] > 30)` se privileged female passengers filter honge.


```python
# Part A: Display only specific columns
selected_cols = ["Name", "Sex", "Age", "Fare", "Survived"]
df_subset = df_titanic[selected_cols]
print("--- Q2 Part A: Selected Columns Preview (First 5 Rows) ---\n", df_subset.head(5))

# Part B: Female passengers with Fare > 30
female_high_fare = df_subset[(df_subset["Sex"] == "female") & (df_subset["Fare"] > 30)]
print(f"\n--- Q2 Part B: Female Passengers with Fare > 30 (Total: {len(female_high_fare)}) ---")
print(female_high_fare.head(6))
```

### 🔹 Q2 Part C: Pclass GroupBy Analysis (Survival Rate, Avg Fare, Avg Age)

#### 💡 Hinglish Explanation:
- `Survived` column me `0` (died) aur `1` (survived) hota hai. Iska `mean()` nikalne se direct **Survival Rate percentage** mil jati hai!
- `Pclass 1` (First class - Rich passengers) ka survival rate sabse high expected hota hai.


```python
# Group by Pclass
pclass_analysis = df_titanic.groupby("Pclass").agg(
    survival_rate=("Survived", "mean"),
    average_fare=("Fare", "mean"),
    average_age=("Age", "mean")
).round(2)

# Percentage formatting for survival rate
pclass_analysis["survival_rate_pct"] = (pclass_analysis["survival_rate"] * 100).astype(str) + "%"

print("--- Q2 Part C: Titanic Class-wise Survival & Demographic Analysis ---\n", pclass_analysis)

print("\nKey Insight (Hinglish):")
print("First Class (Pclass 1) ke passengers ka Survival Rate sabse zyada hai aur Average Fare bhi sabse high hai.")
```
