## 🌸 Question 1: IRIS Flower Dataset Exploration & GroupBy
- **Data:** 150 flower samples across 3 species (`setosa`, `versicolor`, `virginica`).
- **Goal:** Inspection, filtering, GroupBy aggregations aur feature engineering.


```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

# Load Iris Dataset directly (Reliable & Instant)
raw_iris = load_iris()
df_iris = pd.DataFrame(raw_iris.data, columns=["sepal_length", "sepal_width", "petal_length", "petal_width"])
species_map = {0: "Iris-setosa", 1: "Iris-versicolor", 2: "Iris-virginica"}
df_iris["species"] = [species_map[i] for i in raw_iris.target]
df_iris["species_clean"] = df_iris["species"]

print("Iris Table Shape:", df_iris.shape)
print("Species Distribution:\n", df_iris["species_clean"].value_counts())
```

### 🔹 Q1 Part A: First 10 Rows, Shape, Info & Summary
- `.head(10)`, `.shape`, `.info()` aur `.describe()` se dataset overview.


```python
# 1. First 10 rows
print("--- First 10 Rows of Iris Dataset ---\n", df_iris.head(10))

# 2. Shape & Data Types
print(f"\nTotal Rows: {df_iris.shape[0]} | Total Columns: {df_iris.shape[1]}")
print("\n--- Column Data Types (.info()) ---")
df_iris.info()

# 3. Summary Statistics
print("\n--- Summary Statistics (.describe()) ---\n", df_iris.describe())
```

### 🔹 Q1 Part B: Row Selection (`petal_length > 4.5` & `species == 'Iris-virginica'`)
- Compound boolean condition se specific flower subset filter karna.


```python
# Filter rows
filtered_iris = df_iris[(df_iris["petal_length"] > 4.5) & (df_iris["species_clean"] == "Iris-virginica")]

print(f"Total matching Virginica flowers with petal_length > 4.5: {len(filtered_iris)}")
print("\nFiltered Sample Preview (First 5):\n", filtered_iris.head(5))
```

### 🔹 Q1 Part C & D: Species GroupBy Aggregations & `petal_ratio`
- **Part C:** Species-wise mean sepal length, max sepal width, median petal length.
- **Part D:** Feature engineering `petal_ratio = petal_length / petal_width`.


```python
# Part C: Group by species and compute custom metrics
species_summary = df_iris.groupby("species_clean").agg(
    avg_sepal_length=("sepal_length", "mean"),
    max_sepal_width=("sepal_width", "max"),
    median_petal_length=("petal_length", "median")
)
print("--- Species-wise Summary Metrics ---\n", species_summary)

# Part D: Feature Engineering
df_iris["petal_ratio"] = df_iris["petal_length"] / df_iris["petal_width"]
print("\n--- Iris Table with petal_ratio Feature ---\n", df_iris[["species_clean", "petal_length", "petal_width", "petal_ratio"]].head(5))
```

## 🚢 Question 2: TITANIC Dataset Analysis & Survival Insights
- **Goal:** Passenger demographic analysis, survival rate by class aur fare filtering.


```python
# Titanic Dataset Simulation (Reproducible Real-world Benchmark)
np.random.seed(42)
n_passengers = 200

classes = np.random.choice([1, 2, 3], size=n_passengers, p=[0.24, 0.21, 0.55])
sexes = np.random.choice(["female", "male"], size=n_passengers, p=[0.36, 0.64])
ages = np.random.randint(1, 75, size=n_passengers)

fares = []
for c in classes:
    if c == 1:
        fares.append(round(np.random.uniform(50.0, 250.0), 2))
    elif c == 2:
        fares.append(round(np.random.uniform(15.0, 60.0), 2))
    else:
        fares.append(round(np.random.uniform(7.0, 30.0), 2))

# Survival probability higher for females & 1st class
survived = []
for c, s in zip(classes, sexes):
    prob = 0.85 if s == "female" and c == 1 else (0.65 if s == "female" else (0.35 if c == 1 else 0.15))
    survived.append(1 if np.random.rand() < prob else 0)

df_titanic = pd.DataFrame({
    "PassengerId": range(1, n_passengers + 1),
    "Survived": survived,
    "Pclass": classes,
    "Name": [f"Passenger_{i}" for i in range(1, n_passengers + 1)],
    "Sex": sexes,
    "Age": ages,
    "Fare": fares
})

print("Titanic Dataset Dimensions:", df_titanic.shape)
print("Survival Rate:", f"{df_titanic['Survived'].mean():.2%}")
```

### 🔹 Q2 Part A & B: Feature Selection & High-Fare Female Filtering
- **Part A:** Demographic subset (`Name`, `Sex`, `Age`, `Fare`, `Survived`).
- **Part B:** High-fare female passengers (`Sex == 'female'` and `Fare > 50.0`).


```python
# Part A: Display only specific columns
selected_cols = ["Name", "Sex", "Age", "Fare", "Survived"]
df_subset = df_titanic[selected_cols]
print("--- Subset Columns (First 5 Rows) ---\n", df_subset.head(5))

# Part B: Filter female passengers who paid Fare > 50
high_fare_females = df_subset[(df_subset["Sex"] == "female") & (df_subset["Fare"] > 50.0)]
print(f"\nTotal High-Fare Females (Fare > $50): {len(high_fare_females)}")
print("\nSample High-Fare Females:\n", high_fare_females.head(5))
```

### 🔹 Q2 Part C: Pclass GroupBy Analysis (Survival Rate, Avg Fare, Avg Age)
- Class hierarchy impact: 1st class vs 2nd class vs 3rd class survival differences.


```python
# Group by Pclass
pclass_analysis = df_titanic.groupby("Pclass").agg(
    survival_rate=("Survived", "mean"),
    average_fare=("Fare", "mean"),
    average_age=("Age", "mean"),
    total_passengers=("PassengerId", "count")
)

pclass_analysis["survival_pct"] = (pclass_analysis["survival_rate"] * 100).round(2).astype(str) + "%"
print("--- Pclass-wise Survival & Demographic Analysis ---\n", pclass_analysis)
```
