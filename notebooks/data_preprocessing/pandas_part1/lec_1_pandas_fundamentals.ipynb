## 📊 Operation 1: Pandas Series (1D Labeled Array)

### 💡 Kyu Zaroorat Hai?
Spreadsheet ka ek single column ek **Series** hota hai. NumPy array se alag, Series me har element ke sath ek readable label (index) judi hoti hai.

### 🔍 Key Points:
1. **Homogeneous Data:** Ek series me same data type hota hai.
2. **Custom Labels:** Hum default `0, 1, 2...` index ki jagah custom names (jaise "Adam", "Eve") assign kar sakte hain.
3. **Mutable Values but Immutable Size:** Elements ki value change kar sakte hain, lekin Series ka size direct modify nahi hota (drop karne par naya series return hota hai).


```python
import pandas as pd
import numpy as np

# 1. Default Index wali Series
ages = pd.Series([23, 25, 28, 32])
print("Default Series:\n", ages)
print("Index labels:", ages.index)

# 2. Custom Labeled Index
salaries = pd.Series([55000, 72000, 68000, 91000], index=["Adam", "Eve", "Charlie", "David"])
print("\nCustom Labeled Series:\n", salaries)

# Label se access karna
print("\nEve's Salary:    ", salaries["Eve"])
print("Charlie's Salary:", salaries["Charlie"])

# 3. Vectorized Operations (Loop ki zaroorat nahi)
bonus = pd.Series([5000, 5000, 5000, 5000], index=["Adam", "Eve", "Charlie", "David"])
total_payout = salaries + bonus
print("\nTotal Payout (Vectorized Sum):\n", total_payout)

# 4. Values update karna vs Drop karna
salaries["Adam"] = 60000  # Value update ho gayi
print("\nUpdated Adam's Salary:", salaries["Adam"])

dropped_series = salaries.drop("David")  # David remove karke nayi series return karega
print("\nOriginal Series still intact:\n", salaries)
print("\nNew Series after drop:\n", dropped_series)
```

## 🗃️ Operation 2: Creating DataFrames (Dictionary, NumPy, & Lists)

### 💡 Kyu Zaroorat Hai?
**DataFrame** 2-dimensional tabular table hota hai jisme rows aur columns hote hain. Real-world me customer records, transactions, aur sensor logs sab DataFrame ke roop me hi handle hote hain.

### 🔍 Kaise Banate Hain:
1. **Dictionary of Lists:** Sabse aasan aur common tarika (Key = Column Name, Value = List of data).
2. **NumPy 2D Array:** Matrix ko DataFrame me wrap karke column names assign karna.
3. **List of Lists:** Nested lists se table banana.


```python
# 1. Dictionary se DataFrame
student_dict = {
    "Name": ["Aarav", "Isha", "Rohan", "Sneha"],
    "Marks": [85, 92, 78, 95],
    "City": ["Delhi", "Mumbai", "Bangalore", "Pune"]
}
df_dict = pd.DataFrame(student_dict)
print("--- DataFrame from Dictionary ---\n", df_dict)
print("\nColumns:", df_dict.columns.tolist())
print("Row Index:", df_dict.index.tolist())

# 2. NumPy 2D Array se DataFrame
matrix = np.array([[10, 20, 30], 
                   [40, 50, 60], 
                   [70, 80, 90]])
df_np = pd.DataFrame(matrix, columns=["Col_A", "Col_B", "Col_C"])
print("\n--- DataFrame from NumPy Matrix ---\n", df_np)

# 3. List of Lists se DataFrame
records = [["Laptop", 65000, 15], 
           ["Smartphone", 25000, 40], 
           ["Headphones", 3500, 120]]
df_list = pd.DataFrame(records, columns=["Product", "Price", "Stock"])
print("\n--- DataFrame from Lists ---\n", df_list)
```

## 🔍 Operation 4: DataFrame Inspection & Summary Methods

### 💡 Kyu Zaroorat Hai?
Bade dataset (10,000+ rows) ko direct print karenge toh screen bhar jayegi. Hum inspection methods se quick overview lete hain:

## 🎯 Operation 5: Selecting Data — Columns, `loc` vs `iloc`, & Scalar Access

### 💡 Kyu Zaroorat Hai? (Most Important Interview Topic!)
Table me se kisi specific row aur column ki value nikalne ke liye Pandas do primary accessors provide karta hai:
1. **`df.loc[...]` (Label-based):** Row label aur Column names se access karta hai.
   - ⚠️ **Dhyan de:** `loc` me slicing karte waqt **ending index INCLUSIVE** hota hai (`0:3` ka matlab rows 0, 1, 2, 3 charon aayengi!).
2. **`df.iloc[...]` (Integer Position-based):** 0-indexed positions se access karta hai.
   - ⚠️ **Dhyan de:** `iloc` me slicing Python ki tarah **exclusive** hoti hai (`0:3` ka matlab sirf 0, 1, 2 aayengi, 3 nahi!).
3. **`df.at[...]` aur `df.iat[...]`:** Agar sirf ek single cell ki value super-fast speed me access karni ho.


```python
# 1. Columns Select Karna
single_col = df_emp["Name"]            # Returns a Series
multi_cols = df_emp[["Name", "Salary"]] # Returns a DataFrame
print("Single column type:", type(single_col))
print("Multi-column type: ", type(multi_cols))

# 2. df.loc (Label-based)
print("\n--- df.loc Selection ---")
print("Row label 0:\n", df_emp.loc[0])
print("\nloc slice 0 to 2 (INCLUSIVE -> 3 rows):\n", df_emp.loc[0:2, ["Name", "Department", "Salary"]])

# 3. df.iloc (Integer Position-based)
print("\n--- df.iloc Selection ---")
print("Row at position 0:\n", df_emp.iloc[0])
print("\niloc slice 0 to 2 (EXCLUSIVE -> 2 rows):\n", df_emp.iloc[0:2, [1, 3, 4]])

# 4. Single Cell Fast Access (at vs iat)
print("\nSpecific cell with .at (Row 2, 'Salary'):", df_emp.at[2, "Salary"])
print("Specific cell with .iat (Row 2, Col 4):   ", df_emp.iat[2, 4])
```

## 🔬 Operation 6: Data Filtering (Boolean Masking & `.query()`)

### 💡 Kyu Zaroorat Hai?
Real-world analysis me hume conditionally data filter karna hota hai (e.g., "IT department ke wo employees jinki salary > 70,000 hai").

### 🔍 Two Approaches:
1. **Direct Boolean Masking (`df[...]`):**
   - Conditions ko parentheses `(...)` me wrap karo.
   - AND ke liye `&` aur OR ke liye `|` use hota hai (Python ka `and`/`or` kaam nahi karta).
2. **`.query("...")` Method:**
   - SQL jaisi readable string condition likhi jati hai.
   - Iske andar English words `and`, `or`, `not` likh sakte hain.
   - External Python variable use karne ke liye `@variable_name` likhte hain.
   - Query hamesha ek **safe copy** return karta hai.


```python
# 1. Boolean Filtering: Salary > 60000
high_salary = df_emp[df_emp["Salary"] > 60000]
print("--- Employees with Salary > 60,000 ---\n", high_salary[["Name", "Department", "Salary"]])

# 2. Compound Condition: IT Department AND Age >= 30
it_seniors = df_emp[(df_emp["Department"] == "IT") & (df_emp["Age"] >= 30)]
print("\n--- IT Department & Age >= 30 ---\n", it_seniors[["Name", "Age", "Salary"]])

# 3. .query() Method (SQL Style)
query_res = df_emp.query("Department == 'Finance' and Salary >= 50000")
print("\n--- Query Method (Finance & Salary >= 50,000) ---\n", query_res[["Name", "Salary"]])

# 4. Query with Python Variable (@var)
target_dept = "Marketing"
min_age = 25
var_query = df_emp.query("Department == @target_dept and Age >= @min_age")
print(f"\n--- Query with Python Variable (@target_dept='{target_dept}') ---\n", var_query[["Name", "Age", "Department"]])
```

## 🧹 Operation 7: Data Cleaning — Handling Missing Values (`NaN`)

### 💡 Kyu Zaroorat Hai?
Real-world raw data me hamesha missing values (empty cells, nulls) hoti hain. Agar inko handle nahi kiya, toh ML models train hone se fail ho jayenge.

### 🔍 Key Techniques on `raw_data.csv`:
- `.isnull()` / `.isna()`: Missing values par `True` return karta hai.
- `.isnull().sum()`: Har column me kitni missing values hain count karta hai.
- `.dropna()`: Jin rows me koi bhi missing value ho, unhe delete karna.
- `.fillna(val)`: Missing values ko kisi specific number ya mean/median se fill karna.
- `.ffill()` (Forward fill) & `.bfill()` (Backward fill): Time-series data me previous/next value copy karna.


```python
# Load the dirty raw dataset
df_raw = pd.read_csv("raw_data.csv")
print("--- Raw Dataset with Missing & Dirty Values ---\n", df_raw)

# 1. Check Missing Values Count
print("\nMissing Values Count per Column:\n", df_raw.isnull().sum())

# 2. Dropping Rows with any missing value
dropped_df = df_raw.dropna()
print(f"\nRows before dropna: {len(df_raw)} | Rows after dropna: {len(dropped_df)}")

# 3. Filling Missing Values with Column Mean / Imputation
df_clean = df_raw.copy()
mean_age = round(df_clean["age"].mean(), 1)
mean_income = round(df_clean["income"].mean(), 1)

df_clean["age"] = df_clean["age"].fillna(mean_age)
df_clean["income"] = df_clean["income"].fillna(mean_income)
df_clean["country"] = df_clean["country"].fillna("Unknown")
df_clean["gender"] = df_clean["gender"].fillna("Not Specified")

print(f"\n--- Imputed Clean Dataset (Age Mean={mean_age}, Income Mean={mean_income}) ---\n", df_clean)
print("\nRemaining missing values:", df_clean.isnull().sum().sum())
```

## 👥 Operation 8: Data Cleaning — Detecting & Dropping Duplicates

### 💡 Kyu Zaroorat Hai?
Data scraping ya form submissions me duplicate rows aa jati hain. Duplicates se model bias aur overcounting hoti hai.

### 🔍 Kaise Handle Karte Hain:
- `.duplicated()`: Duplicate rows par `True` return karta hai.
- `.duplicated(subset=['country'])`: Kisi specific column ke duplicate check karna.
- `.drop_duplicates(inplace=False)`: Duplicate rows ko remove karta hai aur unique dataset return karta hai.


```python
# Check duplicate rows
print("Duplicate Rows Mask:\n", df_raw.duplicated())
print("\nTotal Duplicate Rows Count:", df_raw.duplicated().sum())

# Duplicates display karna
duplicate_entries = df_raw[df_raw.duplicated(keep=False)]
print("\nAll instances of duplicate records:\n", duplicate_entries)

# Duplicates drop karna
deduped_df = df_raw.drop_duplicates()
print(f"\nTotal rows before: {len(df_raw)} | Total rows after drop_duplicates: {len(deduped_df)}")
```

## ⚙️ Operation 9: Data Cleaning — Data Types & Datetime Parsing

### 💡 Kyu Zaroorat Hai?
Aksar numbers string format me aa jate hain ya dates normal text hoti hain. Unhe numerical calculations ya time-series filtering ke liye sahi type me cast karna padta hai.

### 🔍 Methods:
- `.dtypes`: Sabhi columns ke data types show karta hai.
- `.astype(int / float / str)`: Type conversion karta hai.
- `pd.to_datetime(...)`: Text string ko datetime object banata hai jisse saal, mahina, din (`.dt.year`, `.dt.month`) extract kiya ja sake.


```python
# Type conversion on cleaned dataset
df_types = df_clean.copy()
print("Original Data Types:\n", df_types.dtypes)

# Convert income to integer
df_types["income"] = df_types["income"].astype(int)
print("\nIncome column casted to int64:", df_types["income"].dtype)

# Datetime Parsing
date_strings = pd.Series(["2025-01-15", "2025-06-20", "2025-11-04"])
parsed_dates = pd.to_datetime(date_strings)
print("\nParsed Datetime Series:\n", parsed_dates)
print("Extracted Years: ", parsed_dates.dt.year.tolist())
print("Extracted Months:", parsed_dates.dt.month.tolist())
print("Extracted Days:  ", parsed_dates.dt.day.tolist())
```

## 🔤 Operation 10: String Manipulation with `.str` Accessor

### 💡 Kyu Zaroorat Hai?
Textual data me spelling inconsistencies hoti hain (jaise 'Male', 'male', 'MALE ', 'USA', 'usa'). `.str` accessor vector level par strings ko clean karta hai.

### 🔍 Common String Functions:
- `.str.lower()` & `.str.upper()`: Case normalization.
- `.str.strip()`: Unwanted extra leading/trailing whitespace hatana.
- `.str.contains("pattern", case=False)`: Substring match check karna.
- `.str.split(" ")`: Full name ko first name aur last name me split karna.


```python
df_str = df_clean.copy()

# 1. Lowercase aur Uppercase
print("Gender in Lowercase:\n", df_str["gender"].str.lower())

# 2. Text Matching with case=False
usa_residents = df_str[df_str["country"].str.contains("usa", case=False, na=False)]
print("\n--- Filtered USA Residents (Case-Insensitive) ---\n", usa_residents[["id", "name", "country"]])

# 3. Name Splitting
# Fill missing names first
df_str["name"] = df_str["name"].fillna("Unknown User")
name_splits = df_str["name"].str.split(" ", expand=True)
df_str["first_name"] = name_splits[0]
df_str["last_name"] = name_splits[1]

print("\n--- Name Split into First & Last Names ---\n", df_str[["name", "first_name", "last_name"]].head(5))
```
