## ❓ Problem 1: Custom Labeled Series & Scalar Performance
- **Problem:** 5 Indian cities ke AQI data ki Series banao aur highest AQI city nikal kar alert print karo.
- **Concept:** Labeled indexing `series['Delhi']` aur scalar lookup `.max()`.


```python
import pandas as pd
import numpy as np

# 1. Series Creation
aqi_data = pd.Series([180, 140, 95, 165, 110], 
                     index=["Delhi", "Mumbai", "Bangalore", "Kolkata", "Chennai"])
print("AQI Index Series:\n", aqi_data)

# 2. Highest AQI City
worst_city = aqi_data.idxmax()
worst_aqi = aqi_data.max()
print(f"\nCity with Highest AQI: '{worst_city}' ({worst_aqi})")

# 3. Cities with Moderate/Good AQI (< 120)
clean_cities = aqi_data[aqi_data < 120]
print("\nCities with Clean Air (AQI < 120):\n", clean_cities)
```

## ❓ Problem 2: E-Commerce Orders DataFrame & Multi-Criteria Filtering
- **Problem:** Orders dataset par Electronics category aur amount $>10000$ wale orders filter karo.
- **Concept:** Compound boolean masking `(df['Category'] == 'Electronics') & (df['Amount'] > 10000)`.


```python
orders_df = pd.DataFrame({
    "OrderID": [101, 102, 103, 104, 105],
    "Customer": ["Aman", "Pooja", "Rahul", "Simran", "Vikas"],
    "Category": ["Electronics", "Clothing", "Electronics", "Home", "Electronics"],
    "Amount": [25000, 3200, 14500, 6800, 8900],
    "Status": ["Delivered", "Cancelled", "Delivered", "Delivered", "Returned"]
})
print("Orders DataFrame:\n", orders_df)

# Filter: Electronics & Amount > 10000 & Delivered
filtered_orders = orders_df.query("Category == 'Electronics' and Amount > 10000 and Status == 'Delivered'")
print("\nHigh-Value Delivered Electronics Orders:\n", filtered_orders[["OrderID", "Customer", "Amount"]])
```

## ❓ Problem 3: Production Data Cleaning Pipeline (Imputation & Deduplication)
- **Problem:** `raw_data.csv` load karke duplicates drop karo aur numerical columns mean se impute karo.
- **Concept:** `.drop_duplicates()` $\to$ `.isnull().sum()` $\to$ `.fillna(mean)`.


```python
raw = pd.read_csv("raw_data.csv")
print(f"Initial raw rows: {len(raw)}")

# 1. Deduplication
clean_pipe = raw.drop_duplicates().copy()
print(f"Rows after removing duplicates: {len(clean_pipe)}")

# 2. Imputation
clean_pipe["age"] = clean_pipe["age"].fillna(clean_pipe["age"].median())
clean_pipe["income"] = clean_pipe["income"].fillna(clean_pipe["income"].mean())
clean_pipe["country"] = clean_pipe["country"].fillna("Unknown")
clean_pipe["gender"] = clean_pipe["gender"].fillna("Not Specified")

print("\nMissing Values after Cleaning Pipeline:\n", clean_pipe.isnull().sum())
print("\nCleaned Sample Preview:\n", clean_pipe.head(3))
```

## ❓ Problem 4: Text Normalization & Email Domain Extraction
- **Problem:** Messy names standardize karo (strip whitespace, title case) aur email se domain extract karo.
- **Concept:** `.str.strip().str.title()` aur `.str.split('@', expand=True)[1]`.


```python
users = pd.DataFrame({
    "name": ["Amit Sharma ", "  priya verma", "NEHA GUPTA", "karan singh "],
    "email": ["amit.s@gmail.com", "priya_v@yahoo.com", "neha.g@outlook.com", "karan_s@gmail.com"]
})

print("Original Messy Users:\n", users)

# 1. Clean Names
users["clean_name"] = users["name"].str.strip().str.title()

# 2. Extract Email Domains
users["domain"] = users["email"].str.split("@", expand=True)[1]

print("\nNormalized Users DataFrame:\n", users[["clean_name", "email", "domain"]])
```

## ❓ Problem 5: `loc` vs `iloc` Boundary & In-place Mutation
- **Problem:** Custom labeled DataFrame me label slice vs index slice ka difference verify karo.
- **Concept:** `loc['A':'C']` 3 rows return karega (Inclusive); `iloc[0:2]` 2 rows return karega (Exclusive).


```python
test_df = pd.DataFrame({
    "Score": [88, 92, 79, 95, 84],
    "Grade": ["B", "A", "C", "A+", "B+"]
}, index=["A", "B", "C", "D", "E"])

print("Custom Index DataFrame:\n", test_df)

# loc is INCLUSIVE of both ends
loc_slice = test_df.loc["A":"C"]
print("\nloc['A':'C'] (Inclusive -> 3 rows):\n", loc_slice)

# iloc is EXCLUSIVE of end index
iloc_slice = test_df.iloc[0:2]
print("\niloc[0:2] (Exclusive -> 2 rows):\n", iloc_slice)
```
