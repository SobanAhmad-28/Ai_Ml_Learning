## ❓ Problem 1: Custom Labeled Series & Scalar Performance

### 📌 Problem Statement:
5 Indian cities (`Delhi`, `Mumbai`, `Bengaluru`, `Hyderabad`, `Chennai`) ke average pollution AQI levels ka ek Pandas Series create karo.
1. Shuru ke 3 cities ka AQI `loc` aur `iloc` dono se slice karke display karo.
2. AQI value ko 10% badha do using vectorized arithmetic (bina loop ke).
3. Check karo ki kaunse cities me AQI > 150 hai (Hazardous Alert).

### 💡 Hinglish Explanation:
- Custom index string labels hote hain.
- `loc['Delhi':'Bengaluru']` inclusive slicing karta hai, jabki `iloc[0:3]` integer positions use karta hai.
- Series par seedha `* 1.1` karne se vectorized multiplication ho jata hai.


```python
import pandas as pd
import numpy as np

# 1. Series Creation
aqi_data = pd.Series([180, 140, 95, 165, 110], 
                     index=["Delhi", "Mumbai", "Bengaluru", "Hyderabad", "Chennai"])
print("Original AQI Series:\n", aqi_data)

# Slicing with loc (Inclusive) and iloc (Exclusive)
print("\nTop 3 Cities using .loc:\n", aqi_data.loc["Delhi":"Bengaluru"])
print("\nTop 3 Cities using .iloc:\n", aqi_data.iloc[0:3])

# 2. Vectorized 10% Increase
increased_aqi = np.round(aqi_data * 1.1, 1)
print("\nAQI after 10% rise:\n", increased_aqi)

# 3. Hazardous Alert Filter (> 150)
hazardous = increased_aqi[increased_aqi > 150]
print("\n🔥 Hazardous Cities (AQI > 150):\n", hazardous)
```

## ❓ Problem 2: E-Commerce Orders DataFrame & Multi-Criteria Filtering

### 📌 Problem Statement:
Niche diye gaye 5 orders ka DataFrame banao:
- `OrderID`: `[101, 102, 103, 104, 105]`
- `Customer`: `["Aman", "Pooja", "Rahul", "Simran", "Vikas"]`
- `Category`: `["Electronics", "Clothing", "Electronics", "Home", "Clothing"]`
- `Amount`: `[12000, 3500, 45000, 8500, 2200]`
- `Payment`: `["UPI", "Card", "EMI", "UPI", "Cash"]`

1. Boolean masking se wo orders filter karo jinki category `'Electronics'` ho AND amount $> 15000$.
2. `.query()` method se wo orders filter karo jinka payment method `'UPI'` ho YA amount $< 3000$.


```python
orders_df = pd.DataFrame({
    "OrderID": [101, 102, 103, 104, 105],
    "Customer": ["Aman", "Pooja", "Rahul", "Simran", "Vikas"],
    "Category": ["Electronics", "Clothing", "Electronics", "Home", "Clothing"],
    "Amount": [12000, 3500, 45000, 8500, 2200],
    "Payment": ["UPI", "Card", "EMI", "UPI", "Cash"]
})

print("--- E-Commerce Orders Table ---\n", orders_df)

# 1. Boolean Masking: Category == 'Electronics' & Amount > 15000
expensive_elec = orders_df[(orders_df["Category"] == "Electronics") & (orders_df["Amount"] > 15000)]
print("\n--- Expensive Electronics Orders ---\n", expensive_elec)

# 2. Query Method: Payment == 'UPI' or Amount < 3000
upi_or_cheap = orders_df.query("Payment == 'UPI' or Amount < 3000")
print("\n--- Query: UPI or Amount < 3000 ---\n", upi_or_cheap)
```

## ❓ Problem 3: Production Data Cleaning Pipeline (Imputation & Deduplication)

### 📌 Problem Statement:
`raw_data.csv` ko load karo aur ek production data-cleaning pipeline implement karo:
1. Duplicate rows identify karo aur unhe drop karo.
2. Missing `age` ko mean age se fill karo.
3. Missing `income` ko median income se fill karo.
4. Missing `country` ko `'Global'` aur missing `gender` ko `'Unspecified'` mark karo.
5. Final verified cleaned dataframe ko `cleaned_data.csv` me save karo bina row index ke.

### 💡 Hinglish Explanation:
- Income me outliers ho sakte hain, isliye income ko median se fill karna robust practice hoti hai.
- `index=False` lagane se output CSV me unwanted integer index column create nahi hota.


```python
raw = pd.read_csv("raw_data.csv")
print(f"Initial raw rows: {len(raw)}")

# 1. Deduplication
clean_pipe = raw.drop_duplicates().copy()
print(f"Rows after removing duplicates: {len(clean_pipe)}")

# 2. Missing Value Imputation
mean_age = round(clean_pipe["age"].mean(), 1)
median_income = clean_pipe["income"].median()

clean_pipe["age"] = clean_pipe["age"].fillna(mean_age)
clean_pipe["income"] = clean_pipe["income"].fillna(median_income)
clean_pipe["country"] = clean_pipe["country"].fillna("Global")
clean_pipe["gender"] = clean_pipe["gender"].fillna("Unspecified")

print("\nCleaned Dataset Preview:\n", clean_pipe.head(6))
print("\nRemaining Nulls Check:", clean_pipe.isnull().sum().to_dict())

# Save to disk
clean_pipe.to_csv("cleaned_data.csv", index=False)
print("\nSaved to 'cleaned_data.csv' successfully.")
```

## ❓ Problem 4: Text Normalization & Email Domain Extraction

### 📌 Problem Statement:
Niche diye gaye users ke email address dataset se:
- `name`: `["Amit Sharma ", "  priya verma", "NEHA GUPTA", "karan singh "]`
- `email`: `["amit.sharma@gmail.com", "priya.v@yahoo.com", "neha.g@outlook.com", "karan@gmail.com"]`

1. Names ko clean karo: extra whitespace strip karo aur `Title Case` me convert karo.
2. `.str.split()` se har user ka email domain (jaise `gmail.com`, `yahoo.com`) extract karke naya column `domain` banao.


```python
users = pd.DataFrame({
    "name": ["Amit Sharma ", "  priya verma", "NEHA GUPTA", "karan singh "],
    "email": ["amit.sharma@gmail.com", "priya.v@yahoo.com", "neha.g@outlook.com", "karan@gmail.com"]
})

print("Original Dirty Users Table:\n", users)

# 1. Strip and Title Case
users["name"] = users["name"].str.strip().str.title()

# 2. Extract Domain using .str.split('@')
users["domain"] = users["email"].str.split("@").str[1]

print("\nCleaned Users Table with Extracted Domains:\n", users)
```

## ❓ Problem 5: `loc` vs `iloc` Boundary & In-place Mutation Verification

### 📌 Problem Statement:
Ek DataFrame banao jisme index labels `['A', 'B', 'C', 'D', 'E']` hon.
1. `loc['B':'D']` aur `iloc[1:3]` ka slice compare karo aur clear explanation do ki dono ke outputs me kitni rows aati hain aur kyu.
2. Verify karo ki DataFrame slice copy hota hai ya view by modifying a value via `.loc`.


```python
test_df = pd.DataFrame({
    "Score": [88, 92, 79, 95, 84],
    "Grade": ["B", "A", "C", "A+", "B+"]
}, index=["A", "B", "C", "D", "E"])

print("Indexed DataFrame:\n", test_df)

slice_loc = test_df.loc["B":"D"]
slice_iloc = test_df.iloc[1:3]

print(f"\n.loc['B':'D'] count: {len(slice_loc)} rows (INCLUSIVE of label 'D')\n", slice_loc)
print(f"\n.iloc[1:3] count:     {len(slice_iloc)} rows (EXCLUSIVE of stop index 3)\n", slice_iloc)

# In-place cell update test
test_df.loc["A", "Score"] = 100
print("\nUpdated score of student 'A' to 100:\n", test_df)
```
