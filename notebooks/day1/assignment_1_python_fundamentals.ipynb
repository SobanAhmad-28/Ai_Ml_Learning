### Q2: Arithmetic Operations on Two Numbers

**Problem Statement:**
Take two numbers as input from the user and print their sum, difference, product, and quotient.

**Logic:**
- Cast inputs to float to accommodate both integer and decimal inputs.
- Compute `+`, `-`, `*`, and `/`.

```python
# Q2 Solution
num1 = 20.0  # float(input("Enter first number: "))
num2 = 5.0   # float(input("Enter second number: "))

sum_val = num1 + num2
diff_val = num1 - num2
prod_val = num1 * num2
quot_val = num1 / num2 if num2 != 0 else "Undefined (Division by zero)"

print(f"Numbers: {num1}, {num2}")
print(f"Sum:        {sum_val}")
print(f"Difference: {diff_val}")
print(f"Product:    {prod_val}")
print(f"Quotient:   {quot_val}")
```

### Q3: Mixed Type Average Calculation

**Problem Statement:**
Ask the user to enter two integers and one float. Convert them all to floats and print their average.

**Logic:**
- Convert each input explicitly using `float()`.
- Formula: `average = (f1 + f2 + f3) / 3.0`.

```python
# Q3 Solution
int_val1 = 10     # int(input("Enter 1st integer: "))
int_val2 = 25     # int(input("Enter 2nd integer: "))
float_val = 14.5  # float(input("Enter float: "))

# Convert all to floats
f1 = float(int_val1)
f2 = float(int_val2)
f3 = float(float_val)

avg = (f1 + f2 + f3) / 3
print(f"Converted Floats: {f1}, {f2}, {f3}")
print(f"Average: {avg:.4f}")
```

### Q4: Multi-Step Type Conversion

**Problem Statement:**
The user enters a string containing a number (e.g., `"45"`). Convert it to:
- An integer
- A float
- A string again
Print all three values with their types.

**Logic:**
- Use `int(s)`, `float(s)`, and `str()` while inspecting `type().__name__`.

```python
# Q4 Solution
raw_str = "45"  # input("Enter a numeric string: ")

as_int = int(raw_str)
as_float = float(raw_str)
as_str_again = str(as_int)

print(f"Integer:      {as_int} (Type: {type(as_int).__name__})")
print(f"Float:        {as_float} (Type: {type(as_float).__name__})")
print(f"String again: '{as_str_again}' (Type: {type(as_str_again).__name__})")
```

### Q5: Operator Precedence Evaluation

**Problem Statement:**
Evaluate and print the result of the following expression:  
`x = 10 + 3 * 2 ** 2`  
Based on what you learnt in the lecture explain why the output is what it is.

**Precedence Breakdown:**
1. `**` (Exponentiation) has highest precedence: `2 ** 2 = 4`.
2. `*` (Multiplication) is evaluated next: `3 * 4 = 12`.
3. `+` (Addition) has lowest precedence among the three: `10 + 12 = 22`.

```python
# Q5 Solution
x = 10 + 3 * 2 ** 2

print("Result of x = 10 + 3 * 2 ** 2:", x)
print("Explanation:")
print("  Step 1: Exponentiation (2 ** 2) -> 4")
print("  Step 2: Multiplication (3 * 4)  -> 12")
print("  Step 3: Addition (10 + 12)      -> 22")
```

### Q6: Swapping Values of Two Variables

**Problem Statement:**
Write a program to swap values of two numbers entered by the user.

**Logic:**
- Pythonic tuple unpacking: `a, b = b, a` swaps values in memory without requiring a temporary third variable.

```python
# Q6 Solution
a = 15  # int(input("Enter value for a: "))
b = 40  # int(input("Enter value for b: "))

print(f"Before swap: a = {a}, b = {b}")

# Pythonic swap
a, b = b, a

print(f"After swap:  a = {a}, b = {b}")
```

### Q7: Celsius to Fahrenheit Temperature Converter

**Problem Statement:**
Ask the user for a temperature in Celsius (string input). Convert it to `float`, then calculate and print temperature in Fahrenheit.

**Formula:**
$$\text{FahrenheitTemp} = (\text{CelsiusTemp} \times (9 / 5)) + 32$$

```python
# Q7 Solution
celsius_str = "37.5"  # input("Enter temperature in Celsius: ")
celsius = float(celsius_str)

fahrenheit = (celsius * (9 / 5)) + 32

print(f"Celsius Temperature:    {celsius:.2f} °C")
print(f"Fahrenheit Temperature: {fahrenheit:.2f} °F")
```

### Q8: Circle Area Calculation

**Problem Statement:**
Take the radius (`r`) as user input and print the area.

**Formula:**
$$\text{Area} = \pi \times r^2 \quad (\text{value of } \pi = 3.14)$$

```python
# Q8 Solution
r = 7.0  # float(input("Enter radius (r): "))
pi = 3.14

area = pi * (r ** 2)

print(f"Radius (r): {r}")
print(f"Area of Circle: {area:.2f}")
```

### Q9: Simple Interest Calculation

**Problem Statement:**
Ask the user for: Principal ($P$), Rate ($R$), Time ($T$). Convert all to `float` and compute simple interest ($SI$):

**Formula:**
$$SI = \frac{P \times R \times T}{100}$$

```python
# Q9 Solution
p = 50000.0  # float(input("Enter Principal (P): "))
r = 7.5      # float(input("Enter Rate of Interest (R): "))
t = 3.0      # float(input("Enter Time in years (T): "))

si = (p * r * t) / 100
total_amount = p + si

print(f"Principal:       ₹{p:,.2f}")
print(f"Rate:            {r}%")
print(f"Time:            {t} years")
print(f"Simple Interest: ₹{si:,.2f}")
print(f"Total Amount:    ₹{total_amount:,.2f}")
```

### Q10: Extract Integer and Fractional Parts of a Decimal

**Problem Statement:**
Take a decimal number as input (like `45.78`) and output its:
- Integer part (`45`)
- Fractional part (`.78` or `0.78`)

**Logic:**
1. Cast float to `int` to retrieve the integer component: `int_part = int(num)`.
2. Subtract integer component from float (or format remainder): `frac_part = num - int_part`.

```python
# Q10 Solution
raw_decimal = 45.78  # float(input("Enter a decimal number: "))

# Method 1: Mathematical extraction
integer_part = int(raw_decimal)
fractional_part = round(raw_decimal - integer_part, 4)

print(f"Input Decimal:   {raw_decimal}")
print(f"Integer Part:    {integer_part}")
print(f"Fractional Part: {fractional_part} (or .{str(raw_decimal).split('.')[1]})")
```
