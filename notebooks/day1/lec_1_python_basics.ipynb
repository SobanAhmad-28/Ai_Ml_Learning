### Theoretical Overview
In Machine Learning, raw data enters as unstructured text, tabular numbers, or multi-dimensional tensors. Before passing data into NumPy arrays or PyTorch tensors, foundational Python types dictate how memory is allocated, how objects are referenced, and how preprocessing pipelines perform under scale.

Key principles explored in this lecture:
1. **Dynamic Typing & Memory References:** Variables are object pointers, not raw memory slots.
2. **Mutability vs Immutability:** Impact on memory efficiency and defensive copying.
3. **Fast Iteration Constructs:** List comprehensions and hash-table lookups for fast preprocessing.

### Operation 1: Numerical Types & Type Casting in Feature Representation

**Why it is needed:**
In ML pipelines, features such as age, price, or categorical flags arrive with inconsistent types (e.g. integer `25` vs string `'25'`). Strict casting ensures predictable mathematical transformations and prevents silent type errors during matrix multiplications.

**Real-world ML Use Case:**
Normalizing feature scales where integers must be cast to floating-point representations prior to division.

```python
# 1. Feature casting demonstration
raw_age = "24"
raw_income = 65000

# Casting string feature to integer and income to float64 equivalent
age_clean = int(raw_age)
income_norm = float(raw_income) / 100000.0

print(f"Clean Age: {age_clean} (Type: {type(age_clean).__name__})")
print(f"Normalized Income: {income_norm:.4f} (Type: {type(income_norm).__name__})")
```

### Operation 2: Lists vs Tuples (Mutable Sequences vs Immutable Coordinates)

**Why it is needed:**
- **Lists:** Dynamic, mutable arrays. Ideal for collecting sample batches where records are appended incrementally.
- **Tuples:** Fixed-size, immutable sequences with lower memory overhead. Ideal for tensor dimensions (e.g., shape `(32, 224, 224, 3)`) and model hyperparameters that should never change at runtime.

**Real-world ML Use Case:**
Storing fixed spatial coordinates of bounding boxes `(xmin, ymin, xmax, ymax)` as tuples to prevent accidental mutation.

```python
# Dynamic sample collection (List)
feature_batch = [0.12, 0.45, 0.88]
feature_batch.append(0.34)  # Appending dynamic reading

# Fixed image resolution & bounding box (Tuples)
image_shape = (224, 224, 3)
bbox_coords = (12, 45, 180, 210)

print(f"Dynamic Batch: {feature_batch}")
print(f"Fixed Tensor Shape: {image_shape}")
print(f"Target BBox Coordinates: {bbox_coords}")
```

### Operation 3: Dictionaries for Categorical Label Encoding

**Why it is needed:**
Machine Learning algorithms require numerical inputs. Dictionaries provide O(1) constant-time lookup maps to encode textual categories (e.g., `'Cat' -> 0`, `'Dog' -> 1`) and inverse-decode predicted class indices back to human-readable labels.

**Real-world ML Use Case:**
Classification target encoding and vocabulary indexing in Natural Language Processing (NLP).

```python
# Label encoding map for medical diagnosis classes
class_to_idx = {
    "benign": 0,
    "malignant": 1,
    "uncertain": 2
}

# Inverse mapping for prediction decoding
idx_to_class = {idx: label for label, idx in class_to_idx.items()}

# Sample labels from raw dataset
raw_labels = ["benign", "malignant", "benign", "uncertain"]
encoded_targets = [class_to_idx[lbl] for lbl in raw_labels]

print(f"Encoded Targets: {encoded_targets}")
print(f"Decoded Class for index 1: '{idx_to_class[1]}'")
```

### Operation 4: Vector-Ready List Comprehensions for Data Preprocessing

**Why it is needed:**
List comprehensions run in optimized C-level bytecode under CPython, significantly faster than traditional `for` loops with repetitive `.append()` calls. They are the cleanest Pythonic method for applying thresholding, feature scaling, and data filtering.

**Real-world ML Use Case:**
Thresholding noisy sensor signals where values below a threshold are zeroed out (ReLU-like transformation).

```python
# Raw sensor activation readings
raw_signals = [-2.5, 0.4, -0.1, 3.8, 1.2, -4.0, 5.1]

# Applying custom ReLU threshold: max(0, x)
relu_filtered = [round(x, 2) if x > 0 else 0.0 for x in raw_signals]

print(f"Raw Signals: {raw_signals}")
print(f"ReLU Transformed: {relu_filtered}")
```

### Operation 5: Robust Preprocessing Function with Type Hinting

**Why it is needed:**
Modern production ML workflows mandate type hints and explicit error handling to ensure data pipelines do not crash silently mid-training on corrupt samples.

**Real-world ML Use Case:**
A robust Min-Max scaling function that validates non-zero range and returns scaled floats.

```python
from typing import List

def min_max_scaler(values: List[float]) -> List[float]:
    """
    Scales a sequence of numbers into the range [0.0, 1.0].
    Formula: x_scaled = (x - min) / (max - min)
    """
    if not values:
        return []
    
    min_val = min(values)
    max_val = max(values)
    val_range = max_val - min_val
    
    if val_range == 0:
        return [0.0 for _ in values]
        
    return [round((x - min_val) / val_range, 4) for x in values]

# Testing scaler on sample feature distribution
sample_features = [10.0, 20.0, 30.0, 40.0, 50.0]
scaled = min_max_scaler(sample_features)

print(f"Original Distribution: {sample_features}")
print(f"Min-Max Scaled [0, 1]:  {scaled}")
```
