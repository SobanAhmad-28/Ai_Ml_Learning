## 🛠️ Step 0: Setup & Multi-Feature Data

Chalo wahi 5-feature dataset banate hain taaki `Ridge` aur `Lasso` ka behavior direct compare kar sakein.

```python
# 1. Libraries import karo
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso

# 2. 2 useful aur 3 weak/noisy features ka dataset banao
np.random.seed(42)
X = np.random.randn(100, 5)
y = 4.0 * X[:, 0] - 2.5 * X[:, 1] + np.random.normal(0, 0.5, 100)
print("Dataset ready:", X.shape)
```

## 📍 Step 1: Ridge (L2) Cost Function Formula

Ridge Regression me MSE ke saath **weights ka square** ($\theta_j^2$) add hota hai:
$$J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})^2 + \lambda \sum_{j=1}^{n} \theta_j^2$$
- **L2 Regularization:** Agar kisi weight ki value badi ho, toh uska square penalty ko tezi se badha deta hai.

```python
# 1. Ridge L2 cost function helper likho
def ridge_cost(y_true, y_pred, weights, lam):
    m = len(y_true)
    mse_half = (1 / (2 * m)) * np.sum((y_pred - y_true) ** 2)
    l2_penalty = lam * np.sum(weights ** 2)
    return mse_half, l2_penalty, mse_half + l2_penalty

# 2. Squared penalty ka impact dekho
err = np.array([1.0, 2.0, 1.5])
print("Slope=10 L2 cost:", ridge_cost(err, np.zeros(3), np.array([10.0]), lam=0.5))
print("Slope=3  L2 cost:", ridge_cost(err, np.zeros(3), np.array([3.0]), lam=0.5))
```

## 🎯 Step 2: Lasso (L1) vs Ridge (L2) — Exact Difference

| Aspect | Lasso (`L1`) | Ridge (`L2`) |
|---|---|---|
| **Penalty Term** | $\lambda \sum |\theta_j|$ | $\lambda \sum \theta_j^2$ |
| **Unimportant Weights** | **Exact `0`** set kar sakta hai | Chhote karta hai, **exact `0` nahi** |
| **Feature Selection** | Haan (features eliminate karta hai) | Nahi (sab features rakhta hai) |
| **Overfitting Fix** | Haan | Haan |

```python
# 1. OLS, Lasso aur Ridge teeno fit karo
ols = LinearRegression().fit(X, y)
lasso = Lasso(alpha=0.5).fit(X, y)
ridge = Ridge(alpha=50.0).fit(X, y)

# 2. Teeno ke coefficients compare karo
print("OLS   Weights:", np.round(ols.coef_, 3))
print("Lasso Weights:", np.round(lasso.coef_, 3))
print("Ridge Weights:", np.round(ridge.coef_, 3))
```

## 🏆 Step 3: Slope Shrinkage & Generalization Intuition

Lecture me bataya gaya: Ridge thoda sa training error accept kar leta hai taaki slope ($\theta_1$) kam aur line less steep ho sake — isse overall cost $J(\theta)$ minimize hoti hai aur overfitting rukti hai!

```python
# 1. Single feature data par OLS vs Ridge fit karo
X_1d = np.array([[1], [2], [3], [4], [5]])
y_1d = np.array([2, 4, 6, 8, 15])

# 2. Slope comparison dekho
ols_1d = LinearRegression().fit(X_1d, y_1d)
ridge_1d = Ridge(alpha=10.0).fit(X_1d, y_1d)
print(f"OLS Slope  : {ols_1d.coef_[0]:.3f}")
print(f"Ridge Slope: {ridge_1d.coef_[0]:.3f} (Less steep -> generalizes better!)")
```
