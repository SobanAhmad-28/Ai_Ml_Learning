## 🛠️ Step 0: Setup & Temperature vs Rain Example

Lecture me diya gaya binary classification example (`0` = No Rain, `1` = Rain) set karte hain:
- **Spam / Not Spam**, **Loan Approved / Rejected**, ya **Rain / No Rain** sabhi classification problems hain.

```python
# 1. Libraries import karo
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 2. Exact lecture example (Temperature in °C vs Rain 0/1) define karo
temp = np.array([[5], [10], [15], [20], [25], [35]])
rain = np.array([0, 0, 0, 1, 1, 1])
print("Temperatures:", temp.ravel(), "| Rain labels:", rain)
```

## ⚠️ Step 1: Linear Regression Classification Par Kyu Fail Hota Hai?

Agar binary labels (`0` aur `1`) par seedhi `LinearRegression` line chalayein toh **3 badi problems** aati hain:
1. Line ki values true **probabilities** nahi hotin.
2. Output **`< 0` ya `> 1`** ja sakta hai (e.g. `40°C` par `> 1` — nonsensical!).
3. **Outlier sensitivity:** Ek outlier point (`30°C -> 0` ya `60°C -> 1`) poori line shift kar deta hai.

```python
# 1. Classification data par LinearRegression fit karo
lin_clf = LinearRegression().fit(temp, rain)

# 2. Extreme values (0°C aur 45°C) par nonsensical predictions dekho
print("Pred at 0°C  (negative!):", round(lin_clf.predict([[0]])[0], 3))
print("Pred at 45°C (> 1.0!)   :", round(lin_clf.predict([[45]])[0], 3))

# 3. Ek outlier (60°C -> 0) add karke slope shift check karo
temp_out = np.vstack([temp, [[60]]])
rain_out = np.append(rain, [0])
lin_out = LinearRegression().fit(temp_out, rain_out)
print(f"Slope without outlier: {lin_clf.coef_[0]:.4f} | With outlier: {lin_out.coef_[0]:.4f}")
```

## 📍 Step 2: Sigmoid Activation Function (Squashing)

Hume ek aisa math function chahiye jo har real number $(-\infty, +\infty)$ ko `(0, 1)` range me squash kar de:
$$g(z) = \frac{1}{1 + e^{-z}}, \quad e \approx 2.71828$$
- **$z = 0$:** $g(0) = 0.5$
- **$z \ge 0$:** $g(z) \ge 0.5$ (Class `1`)
- **$z < 0$:** $g(z) < 0.5$ (Class `0`)

```python
# 1. Sigmoid activation function likho
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 2. Key boundary values verify karo
for z_val in [-10, -2, 0, 2, 10]:
    print(f"z = {z_val:3d} -> g(z) = {sigmoid(z_val):.6f}")
```

## 🎨 Step 3: S-Shaped Sigmoid Curve vs Straight Line

Dekho Sigmoid curve `0` aur `1` ko approach karta hai (kabhi cross ya exact touch nahi karta) aur `z = 0` par exactly `0.5` se guzarta hai.

```python
# 1. z values par Sigmoid curve plot karo
z_grid = np.linspace(-8, 8, 200)
sig_vals = sigmoid(z_grid)

# 2. Threshold 0.5 aur bounds (0, 1) mark karo
plt.plot(z_grid, sig_vals, color="blue", label="Sigmoid g(z)")
plt.axhline(0.5, color="gray", linestyle="--", label="Threshold = 0.5")
plt.axhline(1.0, color="green", linestyle=":")
plt.axhline(0.0, color="red", linestyle=":")
plt.xlabel("z = theta^T * x")
plt.ylabel("Probability g(z)")
plt.title("Sigmoid Activation Function (S-Curve)")
plt.legend()
plt.show()
```

## 🏆 Step 4: Hypothesis Formula in Vector Form (`theta^T x`)

Logistic Regression me Linear equation ko Sigmoid ke andar pass karte hain:
$$h_\theta(x) = \frac{1}{1 + e^{-\theta^T x}}$$
Jahan $\theta^T x = \theta_0 + \theta_1 x_1 + \theta_2 x_2$ matrix dot product hai.

```python
# 1. Parameter vector theta aur feature vector x (x0=1 bias ke liye) define karo
theta = np.array([-3.5, 0.2])  # theta_0 = -3.5, theta_1 = 0.2
x_22 = np.array([1, 22])       # 22°C temperature

# 2. Matrix dot product z = theta^T * x aur probability nikalo
z = np.dot(theta.T, x_22)
prob = sigmoid(z)
pred_class = 1 if prob >= 0.5 else 0
print(f"At 22°C -> z = {z:.2f}, Probability = {prob:.4f}, Predicted Class = {pred_class}")
```
