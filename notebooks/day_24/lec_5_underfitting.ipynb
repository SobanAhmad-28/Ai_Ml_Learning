## 🛠️ Step 0: Setup & Lecture Quadratic Data

Chalo lecture me diya gaya exact numeric example create karte hain jahan pattern curved ($y \approx x^2$) hai.

```python
# 1. Libraries import karo
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score

# 2. Exact lecture example values define karo
X_train = np.array([[1], [2], [3], [4], [5]])
y_train = np.array([2, 4, 9, 16, 25])

# 3. Test points define karo (x=6, 7 ke liye y=36, 49)
X_test = np.array([[6], [7]])
y_test = np.array([36, 49])
print("Train X:", X_train.ravel(), "| y:", y_train)
```

## 📍 Step 1: Wrong Assumption = High Bias (`LinearRegression`)

Har algorithm kuch assumptions leta hai (jaise `LinearRegression` straight line assume karta hai, Naive Bayes independent features).
- **High Bias:** Jab non-linear data ($y \approx x^2$) par straight line lagayi jaye, toh galat assumption ke kaaran dono Train aur Test error bade aate hain.

```python
# 1. Simple straight-line LinearRegression model fit karo
lin_model = LinearRegression()
lin_model.fit(X_train, y_train)

# 2. Train aur Test predictions compute karo
y_tr_pred = lin_model.predict(X_train)
y_te_pred = lin_model.predict(X_test)

# 3. Train aur Test MSE + R2 check karo
print(f"Linear (Underfit) Train MSE: {mean_squared_error(y_train, y_tr_pred):.2f} | R2: {r2_score(y_train, y_tr_pred):.3f}")
print(f"Linear (Underfit) Test MSE : {mean_squared_error(y_test, y_te_pred):.2f} | R2: {r2_score(y_test, y_te_pred):.3f}")
```

## 🎨 Step 2: Underfit Line vs True Quadratic Curve

Dekho graph par straight line asli points (`[2, 4, 9, 16, 25]`) ko kaise miss karti hai jabki degree-2 curve perfect fit deta hai.

```python
# 1. Better quadratic model (degree 2) fit karo
poly = PolynomialFeatures(degree=2)
quad_model = LinearRegression()
quad_model.fit(poly.fit_transform(X_train), y_train)

# 2. Straight line vs quadratic curve plot karo
X_grid = np.linspace(1, 7, 100).reshape(-1, 1)
plt.scatter(X_train, y_train, color="black", label="Train Points")
plt.plot(X_grid, lin_model.predict(X_grid), color="orange", label="Underfit Line (Deg 1)")
plt.plot(X_grid, quad_model.predict(poly.transform(X_grid)), color="green", label="Good Fit (Deg 2)")
plt.legend()
plt.title("Underfitting: Straight Line on Curved Data")
plt.show()
```

## 🎯 Step 3: Low Variance — Crucial Interview Point

Lecture me bataya gaya: Kai online articles confuse hokar kehte hain ki underfitting me "High Variance" hota hai kyunki test accuracy kam hoti hai — **ye galat hai!**
- **Sahi Definition:** Variance = sensitivity to small changes in data.
- **Underfit Truth:** Model itna rigid aur simple hai ki naye data points add karne par bhi line lagbhag same rehti hai -> **Low Variance**.

```python
# 1. Data me kuch extra noisy points add karo
X_mod = np.vstack([X_train, [[1.5], [2.5], [3.5]]])
y_mod = np.array([2, 4, 9, 16, 25, 3, 7, 13])

# 2. Underfit linear model ko retrain karo
lin_mod = LinearRegression()
lin_mod.fit(X_mod, y_mod)

# 3. Check karo ki naye data par line kitni kam shift hui (Low Variance)
shift = np.mean(np.abs(lin_model.predict(X_grid) - lin_mod.predict(X_grid)))
print(f"Slope Before: {lin_model.coef_[0]:.2f} | After: {lin_mod.coef_[0]:.2f}")
print(f"Average Line Shift (Low Variance): {shift:.3f}")
```

## 🏆 Step 4: Underfit vs Optimal vs Overfit Comparison

| Metric / Property | Underfitting | Optimal Model (Goal) | Overfitting |
|---|---|---|---|
| **Train Accuracy** | Low (~70%) | High | Very High (~95%+) |
| **Test Accuracy** | Low (~65%) | High | Low (~65%–75%) |
| **Bias** | **High Bias** | **Low Bias** | Low Bias |
| **Variance** | **Low Variance** | **Low Variance** | High Variance |

```python
# 1. Quadratic (Good Fit) model ka Test MSE verify karo
y_te_quad = quad_model.predict(poly.transform(X_test))

# 2. Underfit vs Optimal Test R2 compare karo
print(f"Underfit Test R2 : {r2_score(y_test, lin_model.predict(X_test)):.3f}")
print(f"Good Fit Test R2 : {r2_score(y_test, y_te_quad):.3f}")
```
