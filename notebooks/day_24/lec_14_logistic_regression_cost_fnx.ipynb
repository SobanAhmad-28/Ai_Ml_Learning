## 🛠️ Step 0: Setup & Sigmoid Helper

Chalo `numpy`, `matplotlib` aur `sklearn.metrics.log_loss` import karke `sigmoid()` helper define karte hain.

```python
# 1. Libraries import karo
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import log_loss

# 2. Sigmoid helper function likho
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
print("Setup complete!")
```

## ⚠️ Step 1: Problem — Sigmoid + MSE = Non-Convex Curve

Linear Regression me MSE ek smooth bowl-shaped **convex curve** tha (ek hi global minimum).
- **Problem:** Agar MSE me $\hat{y} = \frac{1}{1 + e^{-\theta^T x}}$ daalein, toh curve wavy (**non-convex**) ho jata hai jisme kai **local minima** bante hain — Gradient Descent local minimum me phans sakta hai!

```python
# 1. Sigmoid MSE vs w plot karke non-convex behavior dekho
w_vals = np.linspace(-10, 10, 200)
x_pt, y_pt = np.array([1.0, -2.0]), np.array([1, 0])
mse_surface = [0.5 * np.mean((sigmoid(w * x_pt) - y_pt) ** 2) for w in w_vals]

# 2. Non-convex curve aur flat regions plot karo
plt.plot(w_vals, mse_surface, color="purple")
plt.title("MSE with Sigmoid: Non-Convex / Flat Regions")
plt.xlabel("Weight w")
plt.ylabel("MSE Cost")
plt.show()
```

## 📍 Step 2: Log Loss (Binary Cross Entropy) Piecewise Logic

Isliye hum logarithmic loss use karte hain:
- **Agar $y = 1$:** $\text{Cost} = -\log(\hat{y})$ (`pred=1` par `0`, `pred -> 0` par $\infty$)
- **Agar $y = 0$:** $\text{Cost} = -\log(1 - \hat{y})$ (`pred=0` par `0`, `pred -> 1` par $\infty$)
- **Why?** Chhoti galti par kam penalty, par badi galti (jaise actual `0` par `0.9` predict karna) par **bohot badi penalty**!

```python
# 1. Predicted probabilities [0.05..0.95] par y=1 aur y=0 ki penalty dekho
for p in [0.05, 0.2, 0.5, 0.8, 0.95]:
    loss_y1 = -np.log(p)
    loss_y0 = -np.log(1 - p)
    print(f"y_hat={p:.2f} -> Loss if y=1: {loss_y1:.3f} | Loss if y=0: {loss_y0:.3f}")
```

## 🎨 Step 3: Logarithmic Penalty Curves Plot Karo

Dekho dono curves kaise ek taraf `0` par aate hain aur galat end par infinity ki taraf shoot karte hain.

```python
# 1. Probability grid (0.01 se 0.99) banao
p_grid = np.linspace(0.01, 0.99, 100)

# 2. Dono piecewise log-loss curves ek graph me plot karo
plt.plot(p_grid, -np.log(p_grid), label="y = 1: -log(y_hat)", color="green")
plt.plot(p_grid, -np.log(1 - p_grid), label="y = 0: -log(1 - y_hat)", color="red")
plt.xlabel("Predicted Probability (y_hat)")
plt.ylabel("Log Loss Penalty")
plt.title("Binary Cross Entropy Curves")
plt.legend()
plt.show()
```

## 🏆 Step 4: Combined Cost Function $J(\theta)$ & Gradient Descent

Dono cases ko ek single convex equation me likh sakte hain:
$$J(\theta) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(\hat{y}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{y}^{(i)}) \right]$$
Aur phir wahi Gradient Descent update chalta hai: $\theta_j := \theta_j - \alpha \frac{\partial}{\partial \theta_j} J(\theta)$.

```python
# 1. Combined Binary Cross Entropy (Log Loss) function likho
def binary_cross_entropy(y_true, y_hat):
    return -np.mean(y_true * np.log(y_hat) + (1 - y_true) * np.log(1 - y_hat))

# 2. Manual calculation ko sklearn ke log_loss se verify karo
y_true = np.array([1, 0, 1, 0])
y_hat = np.array([0.9, 0.1, 0.8, 0.2])
print("Manual Log Loss :", round(binary_cross_entropy(y_true, y_hat), 5))
print("Sklearn log_loss:", round(log_loss(y_true, y_hat), 5))
```
